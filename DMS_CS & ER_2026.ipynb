# 1. **SymPy**

**Purpose:**

```python
from sympy import symbols, And, Or, Not, Implies, simplify

# Propositions
p, q = symbols('p q')

# Logical Expression
expr = And(p, Or(Not(p), q))

# Simplify the expression
simplified_expr = simplify(expr)
print(f"Simplified Expression: {expr}")
print(f"Simplified Expression: {simplified_expr}")
```

```python
# Truth Table
from sympy.logic.boolalg import truth_table

tt = truth_table(expr, [p, q])
for row in tt:
    #print(row)
    display(row)
```

# **Constructing a truth table of different compound statements in propositional logic.**

# **Step 1: Import Required Libraries**

* *pandas:* This library is used to handle and manipulate tabular data, and here it's used to create and display the truth table in a DataFrame format.
* *IPython.display*: This module provides a way to render LaTeX in Jupyter environments. Math is used for rendering mathematical formulas.



```python
import pandas as pd
from IPython.display import display, Math
```

# **Step 2: Define the generate_truth_table Function**

**Step-by-Step Explanation of the Code in function**

---


**Step 2.1: Define the function**

* generate_truth_table(*expressions): This function takes multiple compound logical expressions as input. The *expressions parameter allows you to pass any number of arguments, where each argument is a tuple of the form (name, function).

**Step 2.2: Extract variables and define the number of rows**
* variables = list(expressions[0][1].__code__.co_varnames): Extracts the variable names used in the first logical function (using co_varnames from the function's code).
* n = len(variables): Calculates the number of variables involved in the logical expressions.
* rows = 2 ** n: Calculates the number of rows in the truth table, which is
2^n, where, 'n' is the number of variables.

**Step 2.3: Generate truth values for each variable**

The data dictionary is populated with truth values for each variable. The dictionary comprehension loops through the variables and generates all possible combinations of truth values (True or False) for them.

---


**Step 2.5: Return the truth table**

After calculating all logical expressions, the data dictionary is converted into a Pandas DataFrame and returned.


```python
def generate_truth_table(*expressions):
    # Get variable names from the first function
    variables = list(expressions[0][1].__code__.co_varnames)
    n = len(variables)
    rows = 2 ** n  # Number of rows in the truth table

    # Generate truth values for each variable
    data = {var: [(i // (2 ** (n - j - 1))) % 2 == 1 for i in range(rows)] for j, var in enumerate(variables)}

    # Compute the values of each expression
    for name, func in expressions:
        data[name] = [func(**{var: row[var] for var in variables}) for row in pd.DataFrame(data).to_dict('records')]

    # Return the truth table as a Pandas DataFrame
    return pd.DataFrame(data)
```


# **Step 3: Define Compound Statements**


```python
truth_table = generate_truth_table(
    ("A ∨ B", lambda A, B: A or B),             # Logical OR
    ("A ∧ B", lambda A, B: A and B),            # Logical AND
    ("¬A", lambda A, B: not A),                 # Logical NOT
    ("A → B", lambda A, B: not A or B),         # Implication
    ("A ↔ B", lambda A, B: A == B),             # Biconditional
)
```

# ***Example 2***

Using the generate_truth_table() function shown above, construct the truth table for a login system's access rule A ∧ (B ∨ C), where A = "username is valid", B = "password is correct", and C = "valid OTP provided", and identify all input combinations under which access is granted.

```python
import pandas as pd
from IPython.display import display, Math

def generate_truth_table(*expressions):
    variables = list(expressions[0][1].__code__.co_varnames)  # Extract variable names
    n = len(variables)  # Number of variables
    rows = 2 ** n  # Number of rows (2^n)

    # Generate all possible combinations of truth values for variables
    data = {var: [(i // (2 ** (n - j - 1))) % 2 == 1 for i in range(rows)] for j, var in enumerate(variables)}

    # Apply each logical expression function
    for name, func in expressions:
        data[name] = [func(**{var: row[var] for var in variables}) for row in pd.DataFrame(data).to_dict('records')]

    return pd.DataFrame(data)

# Define compound logical expressions
truth_table = generate_truth_table(
    ("A ∧ (B ∨ C)", lambda A, B, C: A and (B or C)),  # Logical AND and OR
    ("(A → B) ∧ (B → C)", lambda A, B, C: (not A or B) and (not B or C)),  # Logical AND of implications
    ("¬A ∨ ¬B", lambda A, B, C: not A or not B)  # Logical OR of negations
)

# Display the truth table
display(truth_table)
```

# **Discussion**
The function generalises to any number of variables by reading __code__.co_varnames from the first lambda, so a new expression needs no code changes — only a new (name, function) pair. This scales as O(2ⁿ) rows, which is exactly why automating truth tables matters once n ≥ 3: it replaces error-prone manual enumeration.
Separating "logic" (the lambda) from "truth-table machinery" (generate_truth_table) lets one function handle OR, AND, NOT, implication, biconditional, and arbitrary nested combinations unchanged — the same exhaustive-evaluation principle used by SAT solvers and hardware verification tools to prove or disprove properties like "this gate is always True" or "this access rule never over-grants."

# Objective
To verify if a given compound propositional logic statement is a tautology (true for all cases) or a contradiction (false for all cases) using Python.
# **Theory**
* **Tautology:** A propositional statement is called a tautology if it is true for all possible truth value assignments to its variables.
* **Contradiction:**  A propositional statement is a contradiction if it is false for all possible truth value assignments to its variables.
# **Logical Operators:**
* ¬ (NOT): Negates a proposition.
* ∧ (AND): True if both propositions are true.
* ∨ (OR): True if at least one proposition is true.
* → (Implies): True except when the first proposition is true, and the second is false.
* ↔ (Equivalent): True if both propositions are either true or false.

## **Example 1**

Verify whether the following compound proposition:

 **((P∨Q)∧(¬Q∨R)) →((P∧R)∨(¬Q))**
is a Tautology, Contradiction, or Neither.


---


**Objective**

To verify the logical proposition by constructing a truth table and analyzing its validity under all possible truth value assignments for the propositional variables P, Q, and R.


---
**Theory**

* Logical Operations:

  * OR (∨): True if at least one operand is true.
  * AND (∧): True if both operands are true.
  * NOT (¬): Inverts the truth value of the operand.
  * Implication (→): Defined as ¬A∨B.
* Tautology: A proposition that is always true for all possible truth value assignments.

* Contradiction: A proposition that is always false for all possible truth value assignments.

* Satisfiable: A proposition that is true for at least one truth value assignment.

**Procedure**

* Define the Logical Proposition:

  * The proposition is ((P∨Q)∧(¬Q∨R))→((P∧R)∨(¬Q)).
  * Decompose it into the following parts:
    * LHS:(P∨Q)∧(¬Q∨R).
    * RHS:(P∧R)∨(¬Q).
    * The implication: LHS→RHS, equivalent to ¬LHS∨RHS.
* List Variables:

  * Identify the propositional variables:c

* Create Truth Table:

    * List all possible combinations of truth values for P, Q, R
    * Evaluate each subexpression:
        * P∨Q
        * ¬Q
        * ¬Q∨R
        * LHS:(P∨Q)∧(¬Q∨R)
        * RHS:(P∧R)∨(¬Q)
        * Full Proposition: ((P∨Q)∧(¬Q∨R))→((P∧R)∨(¬Q)).
* Evaluate the Proposition:

    * Analyze the final column in the truth table:
    * If all values are True, the proposition is a Tautology.
    * If all values are False, it is a Contradiction.
    * Otherwise, it is Satisfiable.
* Write Python Code:

    * Use Python to automate the truth table creation and analysis.

# **Observations**

* **Truth Table:** The code generates a table showing all truth value combinations and evaluations for the subexpressions and the full proposition.
* **Final Analysis:**
    * If the final column has all True values, the proposition is a Tautology.
    * If the final column has mixed values, the proposition is Satisfiable.


###**Result**
 For the given proposition:

#  ((P∨Q)∧(¬Q∨R))→((P∧R)∨(¬Q))

the result of the analysis indicates whether it is a Tautology, Contradiction, or Satisfiable.

```python
import pandas as pd
from IPython.display import display

def generate_truth_table(*expressions):
    # Extract variable names from the first expression
    variables = list(expressions[0][1].__code__.co_varnames)
    n = len(variables)  # Number of variables
    rows = 2 ** n  # Number of rows (2^n)

    # Generate all possible combinations of truth values
    data = {var: [(i // (2 ** (n - j - 1))) % 2 == 1 for i in range(rows)] for j, var in enumerate(variables)}

    # Apply logical expressions
    for name, func in expressions:
        data[name] = [func(**{var: row[var] for var in variables}) for row in pd.DataFrame(data).to_dict('records')]

    return pd.DataFrame(data)

# Define the logical proposition
truth_table = generate_truth_table(
    ("(P ∨ Q) ∧ (¬Q ∨ R)", lambda P, Q, R: (P or Q) and (not Q or R)),  # Left-hand side of the implication
    ("(P ∧ R) ∨ (¬Q)", lambda P, Q, R: (P and R) or (not Q)),           # Right-hand side of the implication
    ("((P ∨ Q) ∧ (¬Q ∨ R)) → ((P ∧ R) ∨ (¬Q))", lambda P, Q, R: not ((P or Q) and (not Q or R)) or ((P and R) or (not Q)))  # Full proposition
)

# Display the truth table
display(truth_table)

# Check if the proposition is a tautology
proposition_column = truth_table["((P ∨ Q) ∧ (¬Q ∨ R)) → ((P ∧ R) ∨ (¬Q))"]
if all(proposition_column):
    print("The proposition is a Tautology.")
elif not any(proposition_column):
    print("The proposition is a Contradiction.")
else:
    print("The proposition is Satisfiable.")
```

# **Hasse Diagram of Posets**
A Hasse diagram is a type of diagram used to represent partially ordered sets (Posets). It displays elements as vertices and order relations as edges. The diagram follows specific conventions:

* An edge between two vertices indicates that one element is directly less than another.
* The diagram is drawn such that the elements are arranged in levels, with lower elements at the bottom.

In this example, we will draw a Hasse diagram of the following partially ordered sets (Posets):

* **Poset 1: Divisibility of Numbers**

  * Set S={1,2,3,4,6,12}, with the divisibility relation.
Poset 2: Subsets of a Set

  * Set S={a,b,c}, with the subset relation.

# **Procedure:**
* Install necessary libraries.
* Define the Poset using set elements and their relations.
* Use networkx and matplotlib to draw the Hasse diagram.
* Display the diagram.

# **Example 1: Divisibility of Numbers Poset**
Consider the set S={1,2,3,4,6,12} with the divisibility relation. This means that element a is less than element b if a divides b.

```python
## Divisibility of Numbers Poset
import matplotlib.pyplot as plt
import networkx as nx

# Create the Poset for Divisibility
G = nx.DiGraph()

# Add edges based on divisibility relation
G.add_edges_from([(1, 2), (1, 3), (1, 4), (1, 6), (1, 12),
                  (2, 4), (2, 6), (2, 12), (3, 6), (3, 12),
                  (4, 12), (6, 12)])

# Draw the Hasse diagram
pos = nx.spring_layout(G)  # Layout for node positioning
plt.figure(figsize=(8, 6))
nx.draw(G, pos, with_labels=True, node_size=3000, node_color="lightblue", font_size=12, font_weight="bold", arrows=True)
#plt.title("Hasse Diagram of Divisibility Poset", fontsize=14)
plt.show()
```

# **Example 2: Subsets of a Set Poset**
Consider the set S={a,b,c}, with the subset relation. This means element A is less than element B if A is a subset of B.

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.DiGraph()

nodes = [
    (),
    ('a',), ('b',), ('c',),
    ('a','b'), ('a','c'), ('b','c'),
    ('a','b','c')
]
G.add_nodes_from(nodes)

edges = [
    ((), ('a',)),
    ((), ('b',)),
    ((), ('c',)),
    (('a',), ('a','b')),
    (('a',), ('a','c')),
    (('b',), ('a','b')),
    (('b',), ('b','c')),
    (('c',), ('a','c')),
    (('c',), ('b','c')),
    (('a','b'), ('a','b','c')),
    (('a','c'), ('a','b','c')),
    (('b','c'), ('a','b','c'))
]
G.add_edges_from(edges)

pos = nx.spring_layout(G, seed=1)
labels = {n: str(set(n)) if n else '∅' for n in G}

plt.figure(figsize=(8,6))
nx.draw(G, pos, labels=labels, with_labels=True,
        node_size=2500, node_color='lightgreen', font_size=10, arrows=True)
plt.title("Hasse Diagram of Subsets of {a,b,c}")
plt.show()
```

# **Explanation of Code:**
* **Create a Directed Graph:**

  * A Directed Graph (nx.DiGraph()) is used to represent the partially ordered set (Poset) because it allows for directed edges between elements.

* **Add Edges:**

  * For each pair of elements in the set, an edge is added based on the given relation (divisibility or subset). For example, (1, 2) means 1 divides 2 or
1⊆2.

* **Layout:**

    * font_size=12 and font_weight="bold": Set the font properties.
    * arrows=True: Display arrows to show the direction of edges.

* **Display the Graph:**

  * plt.show() displays the Hasse diagram.

# **Example simple tree**

**Explanation of the Code:**

* Tree Structure (Parent-Child Relations):

    * The tree structure is defined by a list of parent-child pairs: [(1, 2), (1, 3), (2, 4), (2, 5), (3, 6), (3, 7)].

* Manual Node Positioning:

    * We manually set the positions of the nodes in a dictionary (pos) to ensure the tree structure is maintained:
      * Node 1 is at the top of the tree.
      * Nodes 2 and 3 are directly beneath node 1, and so on.
      * Nodes 4, 5, 6, and 7 are placed as leaves of the tree.

* Drawing the Tree:

    * The tree is drawn using networkx's draw function, with arrows indicating the direction of the edges (from parent to child).

* Visualization:

    * The layout positions the nodes in a tree-like manner, which reflects the exact hierarchical structure.

* **Expected Output:**

  * The resulting plot will display the exact tree structure where:
    * Node 1 is at the top (root).
    * Nodes 2 and 3 are directly below node 1.
    * Nodes 4, 5, 6, and 7 are the leaf nodes placed at the lowest level.

This approach guarantees that the tree is visualized accurately, with a clear hierarchical structure.








```python
import networkx as nx
import matplotlib.pyplot as plt

# Function to draw the exact tree structure (Hasse diagram)
def draw_exact_tree(tree_structure):
    G = nx.DiGraph()  # Directed graph for Poset (Hasse diagram)

    # Add edges based on parent-child relations
    G.add_edges_from(tree_structure)

    # Manually set node positions to reflect the tree structure
    pos = {
        1: (0, 3),
        2: (-1, 2),
        3: (1, 2),
        4: (-1.5, 1),
        5: (-0.5, 1),
        6: (0.5, 1),
        7: (1.5, 1)
    }

    plt.figure(figsize=(8, 6))

    # Draw the tree graph
    nx.draw(G, pos, with_labels=True, node_size=3000, node_color='lightgreen', font_size=12, font_weight="bold", arrows=True)

    # Display the title
    plt.title("Exact Tree Structure as Hasse Diagram", fontsize=16)
    plt.show()

# Simple tree structure: (parent, child) pairs
tree_structure = [(1, 2), (1, 3), (2, 4), (2, 5), (3, 6), (3, 7)]

# Draw the exact tree structure
draw_exact_tree(tree_structure)
```

# **Matrix Representation of Graphs**
# Objective
To understand and implement the matrix representation of graphs (Adjacency Matrix and Incidence Matrix) using Python and analyze the relationships between vertices and edges.

# **Theory**
A graph is a collection of vertices (nodes) and edges (connections between nodes).

* # **Adjacency Matrix:**

  * A square matrix where:
    * Rows and columns represent vertices.
    * The element a_{ij} is 1 if there is an edge between vertex i and j, otherwise 0.
  * For weighted graphs, a_{ij}  represents the weight of the edge.

* # Incidence Matrix:

  * A rectangular matrix where:
    * Rows represent vertices.
    * Columns represent edges.
    * The element b_{ij} is:
      * 1 if vertex i is connected to edge j in an undirected graph.
      * 1 for incoming and -1 for outgoing vertices in directed graphs.

# **Procedure**
 * Define a graph using vertices and edges.
 * Represent the graph using an Adjacency Matrix.
 * Represent the graph using an Incidence Matrix.
 * Write a Python program to generate and display both matrices.

```python
# Matrix Representation of a Graph

import numpy as np

# Define a graph with 4 vertices and edges
# Graph:
# 0 -- 1
# |    |
# 2 -- 3
# Edges: (0-1), (0-2), (1-3), (2-3)

# Adjacency Matrix Representation
adj_matrix = np.array([
    [0, 1, 1, 0],  # Vertex 0 connections
    [1, 0, 0, 1],  # Vertex 1 connections
    [1, 0, 0, 1],  # Vertex 2 connections
    [0, 1, 1, 0]   # Vertex 3 connections
])

print("Adjacency Matrix:")
print(adj_matrix)

# Incidence Matrix Representation
# Edges: (0-1), (0-2), (1-3), (2-3)
inc_matrix = np.array([
    [1, 1, 0, 0],  # Vertex 0
    [1, 0, 1, 0],  # Vertex 1
    [0, 1, 0, 1],  # Vertex 2
    [0, 0, 1, 1]   # Vertex 3
])

print("\nIncidence Matrix:")
print(inc_matrix)
```
