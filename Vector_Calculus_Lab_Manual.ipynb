# **1.5 Loops**

A **loop** repeats a block of code. A `for` loop runs over a sequence (the number of repetitions is known); a `while` loop repeats as long as a condition is true. `break` leaves a loop early and `continue` skips to the next repetition.

As a mathematical example, a **Riemann sum** approximates $\displaystyle\int_0^1x^2\,dx=\frac13$ by adding the areas of $n$ thin rectangles — the loop makes the idea of integration concrete.

````python
print("For loop: 1 to 5")
for i in range(1, 6):
    print(i, end=" ")

print("\n\nWhile loop: powers of 2 below 100")
p = 1
while p < 100:
    print(p, end=" ")
    p *= 2

print("\n\nbreak / continue: odd numbers up to 9, stop after 9")
for n in range(1, 20):
    if n % 2 == 0:
        continue
    print(n, end=" ")
    if n == 9:
        break

print("\n\nNested loops: multiplication table")
for i in range(1, 4):
    print([i * j for j in range(1, 6)])

print("\nRiemann sum of x^2 on [0, 1]  (exact value 1/3 = 0.333333):")
for n in (10, 100, 1000, 10000):
    h = 1 / n
    total = 0.0
    for k in range(n):
        total += ((k + 0.5) * h) ** 2 * h        # midpoint rule
    print(f"  n = {n:6d}   sum = {total:.8f}   error = {abs(total - 1 / 3):.2e}")
````

# **1.6 Conditional statements**

`if`, `elif` and `else` choose which block runs, depending on conditions combined with `and`, `or`, `not`. Example: the nature of the roots of $ax^2+bx+c=0$ depends on the sign of the discriminant $D=b^2-4ac$.

````python
x = 10
if x > 5:
    print("x is greater than 5")

marks = 72
if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
else:
    grade = "C"
print("Marks", marks, "-> grade", grade)

def nature_of_roots(a, b, c):
    D = b * b - 4 * a * c
    if D > 0:
        return f"D = {D}: two distinct real roots"
    elif D == 0:
        return f"D = {D}: two equal real roots"
    else:
        return f"D = {D}: complex conjugate roots"

for coeffs in ((1, -5, 6), (1, -2, 1), (1, 2, 5)):
    print(f"{coeffs[0]}x² + ({coeffs[1]})x + {coeffs[2]} = 0  ->", nature_of_roots(*coeffs))
````

# **1.7 Functions**

A **function** is a reusable block of code defined with `def`. It can take parameters (with default values) and send back a result with `return`. Small one-line functions can be written with `lambda`. Below, functions for the *dot product*, *cross product* and *length* of vectors in $\mathbb R^3$ are written — they are used throughout this course.

````python
def greet():
    print("Hello, welcome to Python programming!")

def power(base, exponent=2):                  # default value of the parameter
    return base ** exponent

def dot(u, v):
    return sum(ui * vi for ui, vi in zip(u, v))

def cross(u, v):
    return [u[1] * v[2] - u[2] * v[1],
            u[2] * v[0] - u[0] * v[2],
            u[0] * v[1] - u[1] * v[0]]

def norm(u):
    return dot(u, u) ** 0.5

greet()
print("power(5) =", power(5), "   power(2, 10) =", power(2, 10))

u, v = [1, 0, 2], [0, 3, -1]
w = cross(u, v)
print("\nu =", u, " v =", v)
print("u . v =", dot(u, v), "   |u| =", round(norm(u), 4))
print("u x v =", w, "   (u x v) . u =", dot(w, u), " and (u x v) . v =", dot(w, v), " -> perpendicular to both")

f = lambda x: x ** 2 + 1                      # lambda function
print("\nlambda f(x) = x² + 1:  f(3) =", f(3))
````

# **1.10 Basic exception handling**

An **exception** is a run-time error (division by zero, wrong type, missing file …). The `try` block contains the risky code; `except` handles the error; `finally` always runs. Specific exceptions allow different handling for different errors. The input is again simulated with fixed values so that the cell runs without typing.

````python
def safe_inverse(text):
    try:
        number = float(text)
        return 1 / number
    except ValueError:
        return "Error: not a number"
    except ZeroDivisionError:
        return "Error: division by zero"
    finally:
        pass                      # code here always runs (e.g. closing a file)

for entry in ("4", "0", "abc", "-0.5"):
    print(f"1/{entry!s:4} ->", safe_inverse(entry))

# handling a missing file
try:
    open("does_not_exist.txt")
except FileNotFoundError as err:
    print("\nFileNotFoundError handled:", err)

# raising an exception on purpose
def sqrt_checked(x):
    if x < 0:
        raise ValueError("square root of a negative number")
    return math.sqrt(x)

try:
    print(sqrt_checked(9)); print(sqrt_checked(-4))
except ValueError as err:
    print("ValueError handled:", err)
````

# **Part 2 – SymPy**

# **2.1 Importing SymPy, symbols and expressions**

SymPy (imported as `sp`) performs **exact, symbolic** mathematics. Variables are declared as *symbols*; expressions can then be simplified, factorised, expanded, differentiated, integrated …

````python
import sympy as sp

x, y = sp.symbols('x y')
expr = x ** 2 + 3 * x + 2
print("Expression      :", expr)
print("Simplified      :", sp.simplify((x ** 2 + 2 * x + 1) / (x + 1)))
print("Factored        :", sp.factor(expr))
print("Expanded        :", sp.expand((x + 2) * (x + 1)))
print("Substitute x=2  :", expr.subs(x, 2))
print("Trigonometric   :", sp.simplify(sp.sin(x) ** 2 + sp.cos(x) ** 2), "   ", sp.simplify(sp.sin(2 * x) / sp.sin(x)))
````

# **2.2 Solving equations**

`sp.Eq(lhs, rhs)` forms an equation and `sp.solve` solves it — linear, quadratic, systems and trigonometric equations.

````python
print("2x + 3 = 7              ->", sp.solve(sp.Eq(2 * x + 3, 7), x))
print("x² - 5x + 6 = 0         ->", sp.solve(sp.Eq(x ** 2 - 5 * x + 6, 0), x))
print("x² + 1 = 0 (complex)    ->", sp.solve(sp.Eq(x ** 2 + 1, 0), x))
print("system x + y = 5, x - y = 1 ->", sp.solve([sp.Eq(x + y, 5), sp.Eq(x - y, 1)], [x, y]))
print("sin(x) = 1/2 on [0, 2π] ->", sp.solve(sp.Eq(sp.sin(x), sp.Rational(1, 2)), x))
````

# **2.3 Calculus with SymPy – derivatives, integrals, limits and series**

`diff` gives ordinary, higher-order and *partial* derivatives; `integrate` gives indefinite, definite and multiple integrals; `limit` evaluates limits; `series` gives Taylor/Maclaurin expansions.

````python
f = x ** 3 + 2 * x ** 2 + x + 1
print("f(x)   =", f)
print("f'(x)  =", sp.diff(f, x), "     f''(x) =", sp.diff(f, x, 2))

g = x ** 2 * y + y ** 3
print("\ng(x,y) =", g, "   ∂g/∂x =", sp.diff(g, x), "   ∂g/∂y =", sp.diff(g, y), "   ∂²g/∂x∂y =", sp.diff(g, x, y))

print("\nIndefinite ∫ x² dx          =", sp.integrate(x ** 2, x))
print("Definite   ∫₀¹ x² dx        =", sp.integrate(x ** 2, (x, 0, 1)))
print("Definite   ∫₀^π sin x dx    =", sp.integrate(sp.sin(x), (x, 0, sp.pi)))
print("Double     ∫₀¹∫₀² xy dy dx  =", sp.integrate(x * y, (y, 0, 2), (x, 0, 1)))

print("\nlim (x²-4)/(x-2), x→2       =", sp.limit((x ** 2 - 4) / (x - 2), x, 2))
print("lim (2x²+3)/(x²-1), x→∞     =", sp.limit((2 * x ** 2 + 3) / (x ** 2 - 1), x, sp.oo))
print("lim sin(x)/x, x→0           =", sp.limit(sp.sin(x) / x, x, 0))

print("\nMaclaurin series of e^x     :", sp.series(sp.exp(x), x, 0, 6))
t = sp.symbols('t')
print("Taylor series of ln x at x=1, with t = x-1:", sp.series(sp.ln(1 + t), t, 0, 5))
````

# **2.4 From SymPy to NumPy – `lambdify`, and the `sympy.vector` module**

`sp.lambdify` converts a symbolic expression into a fast NumPy function — this is the bridge used in every following experiment. The `sympy.vector` module defines a 3-D coordinate system `N` and provides `gradient`, `divergence` and `curl`.

````python
from sympy.vector import CoordSys3D, gradient, divergence, curl

expr = sp.exp(-(x ** 2 + y ** 2) / 2)
f_np = sp.lambdify((x, y), expr, 'numpy')
print("lambdify: f(0, 0) =", f_np(0, 0), "   f on an array:", f_np(np.array([0, 1, 2]), np.array([0, 1, 2])).round(4))

N = CoordSys3D('N')
phi = N.x ** 2 * N.y + N.z
F = N.x * N.y * N.i + N.y * N.z * N.j + N.z * N.x * N.k
print("\nφ = x²y + z      ->  grad φ =", gradient(phi))
print("F = (xy, yz, zx) ->  div F  =", divergence(F), "   curl F =", curl(F))
````

# **3.2 Plots used for vector calculus – contour, quiver (vector field) and 3-D surface**

* `contour` / `contourf` draw the **level curves** of a scalar function of two variables;
* `quiver` draws a **vector field** as arrows $(U,V)$ placed at the points $(X,Y)$;

# **Evaluation and Visualization of the Gradient of a Scalar Field**

# **Aim:**

To compute the gradient of a given scalar field and to visualise the corresponding gradient vector field using Python.

# **Objectives:**

* To understand the concept of the gradient of a scalar field.
* To compute the gradient using SymPy.
* To evaluate the gradient numerically using NumPy.
* To visualise the gradient vector field using Matplotlib.

# **Software required:**

Python 3, SymPy, NumPy, Matplotlib

# **Theory:**

The **gradient** of a scalar field $f(x,y,z)$ is the vector field

$$\nabla f=\left(\frac{\partial f}{\partial x},\,\frac{\partial f}{\partial y},\,\frac{\partial f}{\partial z}\right).$$

**Properties**

* $\nabla f$ points in the direction of the **maximum rate of increase** of $f$, and its magnitude $|\nabla f|$ is that maximum rate of change.
* The **directional derivative** in the direction of a unit vector $\mathbf u$ is $D_{\mathbf u}f=\nabla f\cdot\mathbf u$; it is largest ($=|\nabla f|$) when $\mathbf u$ is along $\nabla f$.
* $\nabla f$ is **perpendicular to the level curves** (surfaces) $f=\text{const}$.
* $\nabla f=\mathbf 0$ at critical points (maxima, minima, saddle points).
* If a field is the gradient of a scalar, $\mathbf F=\nabla\phi$, it is called **conservative**; in physics the electric field is $\mathbf E=-\nabla V$, the negative gradient of the potential.

**Case study.** For $f(x,y,z)=x^2+y^2+z^2$ the gradient is $\nabla f=(2x,2y,2z)$; for visualisation the field is plotted in the plane $z=0$.

# **Algorithm:**

1. Import SymPy, NumPy and Matplotlib.
2. Define the symbolic variables $x,y,z$ and the scalar field $f$.
3. Compute the gradient with `sp.diff` for each variable.
4. Create a grid of points with `np.linspace` and `np.meshgrid`.
5. Convert the symbolic components into numerical functions with `sp.lambdify`.
6. Evaluate the gradient on the grid.
7. Plot the vector field with `plt.quiver` (and the level curves with `plt.contour`).

# **3.1 Symbolic gradient of the case-study field**

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# STEP 1: symbolic variables
x, y, z = sp.symbols('x y z')

# STEP 2: scalar field  f(x, y, z) = x² + y² + z²
scalar_field = x ** 2 + y ** 2 + z ** 2

# STEP 3: gradient = (∂f/∂x, ∂f/∂y, ∂f/∂z)
def gradient(f, variables=(x, y, z)):
    return sp.Matrix([sp.diff(f, v) for v in variables])

grad_f = gradient(scalar_field)
print("Scalar field   f   =", scalar_field)
print("Gradient       ∇f  =", list(grad_f))

# value at a point
point = {x: 2, y: 3, z: 0}
grad_at_point = grad_f.subs(point)
print("\nGradient at (2, 3, 0):", list(grad_at_point), "   magnitude |∇f| =", sp.sqrt(grad_at_point.dot(grad_at_point)), "=", float(sp.sqrt(grad_at_point.dot(grad_at_point))))
````

# **3.2 Visualising the gradient field**

The gradient is evaluated on a $20\times20$ grid in the plane $z=0$ and drawn as arrows; the **level curves** $x^2+y^2=c$ (circles) are drawn behind. The arrows are radial (pointing away from the origin, where $f$ increases) and perpendicular to the circles; their length grows with the distance from the origin because $|\nabla f|=2r$.

````python
# STEP 4: grid of points in the plane z = 0
x_vals = np.linspace(-10, 10, 20)
y_vals = np.linspace(-10, 10, 20)
X, Y = np.meshgrid(x_vals, y_vals)

# STEP 5: convert the symbolic components to numerical functions
gx = sp.lambdify((x, y, z), grad_f[0], 'numpy')
gy = sp.lambdify((x, y, z), grad_f[1], 'numpy')
gz = sp.lambdify((x, y, z), grad_f[2], 'numpy')

# STEP 6: evaluate on the grid (z = 0).  Multiplying by np.ones_like(X) makes a constant component (e.g. 0) an array of the grid's shape
U = gx(X, Y, 0) * np.ones_like(X)
V = gy(X, Y, 0) * np.ones_like(X)
magnitude = np.hypot(U, V)

# STEP 7: plot
fig, ax = plt.subplots(1, 2, figsize=(15, 6.5))

xf, yf = np.meshgrid(np.linspace(-10, 10, 200), np.linspace(-10, 10, 200))
cs = ax[0].contour(xf, yf, xf ** 2 + yf ** 2, levels=[10, 25, 50, 100, 150], colors='gray', linestyles='--')
ax[0].clabel(cs, fmt='f = %g', fontsize=8)
q = ax[0].quiver(X, Y, U, V, magnitude, cmap='viridis')
plt.colorbar(q, ax=ax[0], label='|∇f|')
ax[0].set_title('Gradient field of f = x² + y² + z²  in the plane z = 0'); ax[0].set_xlabel('x'); ax[0].set_ylabel('y'); ax[0].set_aspect('equal')

ax[1].quiver(X, Y, U / np.where(magnitude == 0, 1, magnitude), V / np.where(magnitude == 0, 1, magnitude), magnitude, cmap='viridis', pivot='mid')
ax[1].set_title('Same field, arrows normalised (direction only)'); ax[1].set_xlabel('x'); ax[1].set_ylabel('y'); ax[1].set_aspect('equal')
for a in ax: a.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

# **3.3 Checking the properties of the gradient**

At the point $P=(2,3,0)$:

* the directional derivative in the direction $\mathbf u=(1,1,0)/\sqrt2$ is $\nabla f\cdot\mathbf u=(4+6)/\sqrt2\approx7.07$, which is less than the maximum rate $|\nabla f|=\sqrt{52}\approx7.21$;
* $\nabla f$ is perpendicular to the level curve through $P$ (a circle of radius $\sqrt{13}$, whose tangent at $P$ is $(-3,2,0)$);
* a numerical finite-difference derivative gives the same directional derivative.

````python
f_num = sp.lambdify((x, y, z), scalar_field, 'numpy')
P = np.array([2.0, 3.0, 0.0])
g = np.array([float(c) for c in grad_f.subs(point)])

u = np.array([1.0, 1.0, 0.0]) / np.sqrt(2)                      # unit direction
tangent = np.array([-3.0, 2.0, 0.0])                            # tangent to the circle x² + y² = 13 at P

h = 1e-6
finite_difference = (f_num(*(P + h * u)) - f_num(*(P - h * u))) / (2 * h)

print(f"∇f at P                      = {g}")
print(f"|∇f| (maximum rate of change) = {np.linalg.norm(g):.4f}")
print(f"D_u f = ∇f · u               = {g @ u:.4f}   (finite difference: {finite_difference:.4f})")
print(f"D_u f in the direction of ∇f  = {g @ (g / np.linalg.norm(g)):.4f}  = |∇f|  (the maximum)")
print(f"∇f · tangent                 = {g @ tangent:.4f}   -> the gradient is perpendicular to the level curve")

# 'steepest ascent': a few small steps along the gradient increase f fastest
p = P.copy()
print("\nSteepest-ascent steps from P  (f increases fastest along ∇f):")
for step in range(4):
    print(f"  step {step}: point = {p.round(3)},  f = {f_num(*p):.3f}")
    p = p + 0.05 * np.array([float(c) for c in grad_f.subs({x: p[0], y: p[1], z: p[2]})])
````

# **3.4 The gradient is the normal to a level surface**

In 3-D the level set $f=1$ is the unit sphere and $\nabla f=(2x,2y,2z)$ is the outward **normal** vector at each of its points. This is how normals of surfaces are found in the surface-integral experiments that follow.

````python
phi_, theta_ = np.meshgrid(np.linspace(0, np.pi, 18), np.linspace(0, 2 * np.pi, 24))
Xs = np.sin(phi_) * np.cos(theta_); Ys = np.sin(phi_) * np.sin(theta_); Zs = np.cos(phi_)

fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(projection='3d')
ax.plot_surface(Xs, Ys, Zs, alpha=0.25, color='lightblue')
sel = (slice(None, None, 3), slice(None, None, 3))
ax.quiver(Xs[sel], Ys[sel], Zs[sel], 2 * Xs[sel], 2 * Ys[sel], 2 * Zs[sel], length=0.25, normalize=True, color='red')
ax.set_title('Level surface f = 1 (unit sphere) and the gradient ∇f = normal vectors')
ax.set_box_aspect((1, 1, 1))
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')
plt.show()
````

# **Application Problem 1 – Temperature distribution in a metal plate**

````python
x, y = sp.symbols('x y')
T = 100 * sp.exp(-(x ** 2 + y ** 2) / 2)
grad_T = sp.Matrix([sp.diff(T, x), sp.diff(T, y)])

print("Temperature        T(x, y) =", T)
print("Temperature gradient ∇T    =", list(grad_T), "   i.e.  ∇T = -T·(x, y)")
print("\nT(0, 0) =", T.subs({x: 0, y: 0}), "°C        ∇T(0, 0) =", list(grad_T.subs({x: 0, y: 0})))
print("T(1, 1) =", round(float(T.subs({x: 1, y: 1})), 3), "°C   ∇T(1, 1) =", [round(float(c), 3) for c in grad_T.subs({x: 1, y: 1})], " °C/m")

# where is |∇T| largest?   |∇T| = 100 r e^(-r²/2)  ->  maximum at r = 1
r = sp.symbols('r', positive=True)
mag = sp.simplify(sp.sqrt(sum(c ** 2 for c in grad_T)).subs({x: r, y: 0}))
r_max = sp.solve(sp.diff(mag, r), r)
print("\n|∇T| as a function of the distance r from the centre:", mag)
print("Largest rate of change at r =", r_max[0], "m, where |∇T| =", sp.N(mag.subs(r, r_max[0]), 5), "°C/m")
````

````python
X, Y = np.meshgrid(np.linspace(-3, 3, 200), np.linspace(-3, 3, 200))
T_f = sp.lambdify((x, y), T, 'numpy')
gTx = sp.lambdify((x, y), grad_T[0], 'numpy'); gTy = sp.lambdify((x, y), grad_T[1], 'numpy')
Xq, Yq = np.meshgrid(np.linspace(-3, 3, 21), np.linspace(-3, 3, 21))
Uq, Vq = gTx(Xq, Yq), gTy(Xq, Yq)

fig, ax = plt.subplots(1, 3, figsize=(20, 6))
c = ax[0].contourf(X, Y, T_f(X, Y), levels=20, cmap='hot'); plt.colorbar(c, ax=ax[0], label='T (°C)')
ax[0].set_title('Temperature distribution T(x, y)');

ax[1].contour(X, Y, T_f(X, Y), levels=10, colors='gray', linewidths=0.8)
q = ax[1].quiver(Xq, Yq, Uq, Vq, np.hypot(Uq, Vq), cmap='coolwarm')
plt.colorbar(q, ax=ax[1], label='|∇T| (°C/m)')
ax[1].set_title('Temperature gradient ∇T (points towards the hot centre)')

rr = np.linspace(0, 4, 300)
ax[2].plot(rr, 100 * rr * np.exp(-rr ** 2 / 2), 'r', label='|∇T| = 100 r e^(−r²/2)')
ax[2].plot(rr, 100 * np.exp(-rr ** 2 / 2), 'b--', label='T(r)')
ax[2].axvline(1, color='gray', ls=':'); ax[2].set_xlabel('distance r from the centre (m)'); ax[2].legend(); ax[2].set_title('Temperature and its steepest slope'); ax[2].grid(True)
for a in ax[:2]: a.set_xlabel('x (m)'); a.set_ylabel('y (m)'); a.set_aspect('equal')
plt.tight_layout()
plt.show()
````

# **Evaluating the Divergence and Curl of a Vector Field**

# **Aim:**

To compute the divergence and curl of a given vector field and to visualise both the original vector field and its curl.

# **Software required:**

Python 3, SymPy (including `sympy.vector`), NumPy, Matplotlib (with `mpl_toolkits.mplot3d`)

# **Theory:**

For a vector field $\mathbf F=P\,\mathbf i+Q\,\mathbf j+R\,\mathbf k$:

$$\nabla\!\cdot\!\mathbf F=\frac{\partial P}{\partial x}+\frac{\partial Q}{\partial y}+\frac{\partial R}{\partial z}\quad(\text{a scalar field}),$$

$$\nabla\times\mathbf F=\begin{vmatrix}\mathbf i&\mathbf j&\mathbf k\\ \partial_x&\partial_y&\partial_z\\ P&Q&R\end{vmatrix}=\left(\frac{\partial R}{\partial y}-\frac{\partial Q}{\partial z},\ \frac{\partial P}{\partial z}-\frac{\partial R}{\partial x},\ \frac{\partial Q}{\partial x}-\frac{\partial P}{\partial y}\right)\quad(\text{a vector field}).$$

**Physical meaning**

* **Divergence** measures the net rate at which the field "flows out of" a point (a *source* if positive, a *sink* if negative). If $\nabla\cdot\mathbf F=0$ the field is **solenoidal** (incompressible; e.g. $\nabla\cdot\mathbf B=0$ for the magnetic field).
* **Curl** measures the local rotation (swirl) of the field; for a fluid the angular velocity of a small paddle wheel is $\boldsymbol\omega=\tfrac12\nabla\times\mathbf v$. If $\nabla\times\mathbf F=\mathbf 0$ the field is **irrotational**, and on a simply connected region it is then **conservative**, $\mathbf F=\nabla\phi$.

**Identities:** $\nabla\cdot(\nabla\times\mathbf F)=0$ and $\nabla\times(\nabla\phi)=\mathbf 0$; $\nabla\cdot\nabla f=\nabla^2f$ is the Laplacian.

**Case.** Find the divergence and curl of $\mathbf F=xy\,\mathbf i+yz\,\mathbf j+zx\,\mathbf k$ and plot both the vector field and its curl.

# **Algorithm:**

1. Import the required libraries (SymPy, NumPy, Matplotlib and `Axes3D`).
2. Define the symbols (or the coordinate system) and the vector field.
3. Compute the divergence of the vector field.
4. Compute the curl of the vector field.
5. Define a grid of points for evaluation.
6. Evaluate the field and the curl at every grid point (after `lambdify`).
7. Plot the original vector field.
8. Plot the curl of the vector field.

# **4.1 Divergence and curl with SymPy**

Two independent ways: (i) our own functions `div` and `curl` written from the formulae with `sp.diff`, (ii) the ready-made `sympy.vector` module with the coordinate system `N = CoordSys3D('N')`. The two must agree.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sympy.vector import CoordSys3D, divergence, curl as sv_curl

x, y, z = sp.symbols('x y z')

def div(F, v=(x, y, z)):
    return sum(sp.diff(F[i], v[i]) for i in range(3))

def curl(F, v=(x, y, z)):
    return sp.Matrix([sp.diff(F[2], v[1]) - sp.diff(F[1], v[2]),
                      sp.diff(F[0], v[2]) - sp.diff(F[2], v[0]),
                      sp.diff(F[1], v[0]) - sp.diff(F[0], v[1])])

# ---- Case: F = (xy, yz, zx)
F = sp.Matrix([x * y, y * z, z * x])
print("Vector field F     =", list(F))
print("Divergence ∇·F     =", div(F))
print("Curl ∇×F           =", list(curl(F)))
print("At (1, 1, 1):  F =", list(F.subs({x: 1, y: 1, z: 1})), "  div F =", div(F).subs({x: 1, y: 1, z: 1}), "  curl F =", list(curl(F).subs({x: 1, y: 1, z: 1})))

# ---- the same with the sympy.vector module
N = CoordSys3D('N')
F_v = N.x * N.y * N.i + N.y * N.z * N.j + N.z * N.x * N.k
print("\nsympy.vector:   div F =", divergence(F_v), "      curl F =", sv_curl(F_v))
````

# **4.2 Visualising the field and its curl**

The field $\mathbf F$ (blue) and its curl $\nabla\times\mathbf F=(-y,-z,-x)$ (red) are evaluated on a $5\times5\times5$ grid and drawn with 3-D `quiver` plots.

````python
g = np.linspace(-2, 2, 5)
X, Y, Z = np.meshgrid(g, g, g)

Fx, Fy, Fz = [sp.lambdify((x, y, z), c, 'numpy') for c in F]
Cx, Cy, Cz = [sp.lambdify((x, y, z), c, 'numpy') for c in curl(F)]

ones = np.ones_like(X)
fig = plt.figure(figsize=(16, 7))
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax1.quiver(X, Y, Z, Fx(X, Y, Z) * ones, Fy(X, Y, Z) * ones, Fz(X, Y, Z) * ones, length=0.35, normalize=True, color='tab:blue')
ax1.set_title(r'Vector field  $\mathbf{F} = (xy,\ yz,\ zx)$')

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.quiver(X, Y, Z, Cx(X, Y, Z) * ones, Cy(X, Y, Z) * ones, Cz(X, Y, Z) * ones, length=0.35, normalize=True, color='tab:red')
ax2.set_title(r'Curl  $\nabla\times\mathbf{F} = (-y,\ -z,\ -x)$')
for a in (ax1, ax2): a.set_xlabel('x'); a.set_ylabel('y'); a.set_zlabel('z')
plt.tight_layout()
plt.show()
````

# **4.3 Checking the identities**

````python
def grad(f, v=(x, y, z)):
    return sp.Matrix([sp.diff(f, c) for c in v])

print("div(curl F) =", sp.simplify(div(curl(F))), "   for F = (xy, yz, zx)")
G = sp.Matrix([sp.sin(x) * y, x * z ** 2, sp.exp(y) * z])
print("div(curl G) =", sp.simplify(div(curl(G))), "   for G = (y sin x, x z², z e^y)")

phi = x ** 2 * y * z + sp.sin(x * y)
print("\ncurl(grad φ) =", list(sp.simplify(curl(grad(phi)))), "   for φ = x²yz + sin(xy)")
f = x ** 2 + y ** 2 + z ** 2
print("Laplacian ∇²f = div(grad f) =", div(grad(f)), "   for f = x² + y² + z²")
````

**Problem statement.** The velocity field of a fluid in a cylindrical pipe is $\mathbf V(x,y,z)=(-y,\,x,\,0)$.

1. Determine the divergence $\nabla\cdot\mathbf V$, which measures the net rate of fluid flow out of a point.
2. Determine the curl $\nabla\times\mathbf V$, which measures the rotation (swirl) of the fluid.
3. Visualise the velocity field and its curl.

**Theory.** If $\nabla\cdot\mathbf V=0$ the fluid is incompressible at that point; the magnitude of $\nabla\times\mathbf V$ shows how strongly the fluid rotates, and the angular velocity of a small paddle wheel is half of it.

````python
V = sp.Matrix([-y, x, 0])
div_V, curl_V = div(V), curl(V)

print("Velocity field V       =", list(V))
print("Divergence ∇·V         =", div_V, "   -> incompressible: no sources or sinks")
print("Curl ∇×V               =", list(curl_V), "   -> constant swirl about the z-axis")
print("Angular velocity ω = ½ curl V =", [c / 2 for c in curl_V], "  -> the fluid rotates like a rigid body with 1 rad/s")

# consistency check: circulation around a circle of radius r equals (curl)·(area)
t, r = sp.symbols('t r', positive=True)
xt, yt = r * sp.cos(t), r * sp.sin(t)
circulation = sp.integrate(sp.simplify((V[0] * sp.diff(xt, t) + V[1] * sp.diff(yt, t)).subs({x: xt, y: yt})), (t, 0, 2 * sp.pi))
print(f"\nCirculation ∮ V·dr around a circle of radius r = {circulation}   and   (curl V)_z · area = 2·πr² = {2 * sp.pi * r ** 2}")
````

# **Application Problem 2 – Magnetic field around a current-carrying wire**

**Problem statement.** A long straight wire along the $z$-axis carries a steady current $I$. The magnetic field around it is

$$\mathbf B(x,y,z)=\left(-\frac{\mu_0Iy}{2\pi(x^2+y^2)},\ \frac{\mu_0Ix}{2\pi(x^2+y^2)},\ 0\right),\qquad\mu_0=4\pi\times10^{-7},\ I=1.$$

1. Determine the divergence $\nabla\cdot\mathbf B$.
2. Determine the curl $\nabla\times\mathbf B$.
3. Visualise the magnetic field.

**Theory.** In magnetostatics $\nabla\cdot\mathbf B=0$ (no magnetic monopoles). By Ampère's law $\nabla\times\mathbf B=\mu_0\mathbf J$: the curl is zero wherever there is no current — that is, everywhere except on the wire itself.

````python
mu0 = 4 * sp.pi * sp.Rational(1, 10 ** 7)
I = 1
B = sp.Matrix([-mu0 * I * y / (2 * sp.pi * (x ** 2 + y ** 2)), mu0 * I * x / (2 * sp.pi * (x ** 2 + y ** 2)), 0])

print("Magnetic field B   = (μ0 I / 2π) · (−y, x, 0) / (x² + y²),   with μ0 I / 2π =", mu0 * I / (2 * sp.pi))
print("Divergence ∇·B     =", sp.simplify(div(B)))
print("Curl ∇×B           =", [sp.simplify(c) for c in curl(B)], "   (for all points off the wire, x² + y² ≠ 0)")

# Ampère's law: the circulation around any loop that ENCLOSES the wire equals μ0·I; around a loop that does not, it is zero
from scipy.integrate import quad
Bx = sp.lambdify((x, y), B[0], 'numpy'); By = sp.lambdify((x, y), B[1], 'numpy')

def circulation(cx, cy, R):
    f = lambda t: Bx(cx + R * np.cos(t), cy + R * np.sin(t)) * (-R * np.sin(t)) + By(cx + R * np.cos(t), cy + R * np.sin(t)) * (R * np.cos(t))
    return quad(f, 0, 2 * np.pi)[0]

print(f"\nμ0·I = {float(mu0 * I):.6e}  T·m")
for (cx, cy, R, text) in ((0, 0, 0.5, "circle of radius 0.5 around the wire"), (0, 0, 2.0, "circle of radius 2 around the wire"),
                          (0.5, 0, 1.0, "off-centre circle (radius 1, wire inside)"), (3, 0, 1.0, "circle NOT enclosing the wire")):
    print(f"  ∮ B·dr, {text:45s} = {circulation(cx, cy, R):.6e}")
````

````python
Xs, Ys = np.meshgrid(np.linspace(-3, 3, 200), np.linspace(-3, 3, 200))
Xs = np.where(np.hypot(Xs, Ys) < 0.15, np.nan, Xs)                     # remove the singularity on the wire
Bxs, Bys = Bx(Xs, Ys), By(Xs, Ys)
mag = np.hypot(Bxs, Bys)

fig, ax = plt.subplots(1, 3, figsize=(20, 6))
ax[0].streamplot(Xs[0], Ys[:, 0], np.nan_to_num(Bxs), np.nan_to_num(Bys), color=np.log10(np.nan_to_num(mag) + 1e-12), cmap='plasma', density=1.5)
ax[0].plot(0, 0, 'ko', ms=10); ax[0].text(0.15, 0.15, 'wire (I out of the page)', fontsize=9)
ax[0].set_title('Magnetic field lines: circles around the wire'); ax[0].set_aspect('equal'); ax[0].set_xlabel('x (m)'); ax[0].set_ylabel('y (m)')

Xq, Yq = np.meshgrid(np.linspace(-3, 3, 13), np.linspace(-3, 3, 13))
ok = np.hypot(Xq, Yq) > 0.2                                           # leave out the wire itself
Bxq, Byq = Bx(Xq[ok], Yq[ok]), By(Xq[ok], Yq[ok])
ax[1].quiver(Xq[ok], Yq[ok], Bxq / np.hypot(Bxq, Byq), Byq / np.hypot(Bxq, Byq), np.hypot(Bxq, Byq), cmap='plasma', pivot='mid')
ax[1].set_title('Direction of B (arrows normalised)'); ax[1].set_aspect('equal'); ax[1].set_xlabel('x (m)')

rr = np.linspace(0.1, 3, 200)
ax[2].plot(rr, float(mu0 * I) / (2 * np.pi * rr), 'r')
ax[2].set_title('|B| = μ0 I / (2π r): decreases like 1/r'); ax[2].set_xlabel('distance r from the wire (m)'); ax[2].set_ylabel('|B| (tesla)'); ax[2].grid(True)
plt.tight_layout()
plt.show()
````

# **Result**

* $\mathbf F=(xy,yz,zx)$: $\ \nabla\cdot\mathbf F=x+y+z$ and $\nabla\times\mathbf F=(-y,-z,-x)$; the results of the hand-written functions and of `sympy.vector` agree, and $\nabla\cdot(\nabla\times\mathbf F)=0$, $\nabla\times\nabla\phi=\mathbf 0$ were verified.
* Fluid flow $\mathbf V=(-y,x,0)$: $\ \nabla\cdot\mathbf V=0$ (incompressible) and $\nabla\times\mathbf V=(0,0,2)$ — rigid rotation with angular velocity $1$ rad/s; the circulation around a circle of radius $r$ is $2\pi r^2$, equal to $(\text{curl})\times(\text{area})$.
* Magnetic field of a wire: $\nabla\cdot\mathbf B=0$ and $\nabla\times\mathbf B=\mathbf 0$ everywhere off the wire; nevertheless the circulation around any loop enclosing the wire is $\mu_0I\approx1.2566\times10^{-6}$ T·m, and zero for a loop that does not enclose it.

# **Discussion**

Divergence is a *scalar* that detects sources and sinks, curl is a *vector* that detects rotation. The pipe flow has zero divergence but non-zero curl: the fluid swirls without being created or destroyed. The magnetic field presents a subtle case: its curl is zero at every point off the wire, so it looks "irrotational", yet its circulation around the wire is *not* zero — the whole curl is concentrated on the axis (a delta function, $\nabla\times\mathbf B=\mu_0\mathbf J$), which is exactly what Ampère's law states; the region outside the wire is not simply connected, so zero curl does not imply a conservative field there. Pictures of the field lines (circles around the wire) show the same thing as the formulae.

# **Calculating and Visualizing Line Integrals**

# **Aim:**

To evaluate the line integral (work done) of a vector field along a parametrised space curve by the direct parametrisation method, and to visualise the 3-D vector field and the path of the particle or charge.

# **Software required:**

Python 3, SymPy, NumPy, SciPy, Matplotlib (with 3-D support)

# **Theory:**

The **line integral** of a vector field $\mathbf F=(P,Q,R)$ along a curve $C$ given by $\mathbf r(t)=(x(t),y(t),z(t))$, $a\le t\le b$, is

$$\int_C\mathbf F\cdot d\mathbf r=\int_C P\,dx+Q\,dy+R\,dz=\int_a^b\mathbf F\bigl(\mathbf r(t)\bigr)\cdot\frac{d\mathbf r}{dt}\,dt .$$

It represents the **work** done by the force $\mathbf F$ on a particle moving along $C$ (or the circulation of a flow along $C$). Reversing the direction of the curve changes the sign.

**Conservative fields.** If $\mathbf F=\nabla\phi$ (equivalently $\nabla\times\mathbf F=\mathbf 0$ on a simply connected region) the line integral does **not depend on the path**, only on the end points:

$$\int_C\nabla\phi\cdot d\mathbf r=\phi(\mathbf r(b))-\phi(\mathbf r(a)),$$

and $\oint_C\mathbf F\cdot d\mathbf r=0$ for every closed curve.

# **Algorithm:**

1. Import SymPy, NumPy, SciPy and Matplotlib (with 3-D support).
2. Define the vector field $\mathbf F(x,y,z)=(P,Q,R)$.
3. Define the parametrised curve $\mathbf r(t)=(x(t),y(t),z(t))$.
4. Substitute $\mathbf r(t)$ into the field to get $\mathbf F(\mathbf r(t))$.
5. Compute $d\mathbf r/dt$.
6. Form the scalar integrand $\mathbf F(\mathbf r(t))\cdot d\mathbf r/dt$.
7. Integrate with respect to $t$ over the given interval.
8. Check the answer: numerically (`scipy.integrate.quad`) and, if the field is conservative, with the potential.
9. Create a 3-D grid, evaluate the vector field (`lambdify`).
10. Plot the 3-D quiver of the vector field together with the curve.

**Case 1.** Evaluate the line integral of $\mathbf F(x,y,z)=(xy,\,yz,\,zx)$ along the curve $\mathbf r(t)=(\sin t,\ \cos t,\ t)$, $0\le t\le\pi$.

# **Functions used in this experiment**

* `line_integral(F, r, t_range)` – steps 4–7 of the algorithm.
* `plot_field_and_curve(F, r, t_range, ...)` – steps 9–10: the field (light arrows on a 3-D grid), the curve (coloured by the parameter $t$) and the field vectors along the curve (red).
* `potential(F)` – finds a scalar potential $\phi$ if the field is conservative.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
from mpl_toolkits.mplot3d import Axes3D

x, y, z, t = sp.symbols('x y z t')

def curl(F):
    return sp.Matrix([sp.diff(F[2], y) - sp.diff(F[1], z), sp.diff(F[0], z) - sp.diff(F[2], x), sp.diff(F[1], x) - sp.diff(F[0], y)])

def line_integral(F, r, t_range):
    """∫ F·dr along r(t), t in t_range = (a, b).  Returns the integrand F(r(t))·r'(t) and the value."""
    F_rt = F.subs({x: r[0], y: r[1], z: r[2]}, simultaneous=True)       # F(r(t))
    dr_dt = r.diff(t)                                                    # r'(t)
    integrand = sp.simplify(F_rt.dot(dr_dt))                             # F(r(t)) · r'(t)
    return integrand, sp.integrate(integrand, (t, *t_range))

def potential(F):
    """Scalar potential φ with ∇φ = F (only if curl F = 0), by successive integration."""
    if curl(F) != sp.zeros(3, 1):
        return None
    phi = sp.integrate(F[0], x)
    phi += sp.integrate(sp.simplify(F[1] - sp.diff(phi, y)), y)
    phi += sp.integrate(sp.simplify(F[2] - sp.diff(phi, z)), z)
    return sp.simplify(phi)

def plot_field_and_curve(F, r, t_range, lims, title, n=5, ax=None):
    Fx, Fy, Fz = [sp.lambdify((x, y, z), c, 'numpy') for c in F]
    rf = sp.lambdify(t, list(r), 'numpy')
    own = ax is None
    if own:
        fig = plt.figure(figsize=(9, 7.5)); ax = fig.add_subplot(projection='3d')
    gx, gy, gz = [np.linspace(a, b, n) for a, b in lims]
    X, Y, Z = np.meshgrid(gx, gy, gz)
    one = np.ones_like(X)
    ax.quiver(X, Y, Z, Fx(X, Y, Z) * one, Fy(X, Y, Z) * one, Fz(X, Y, Z) * one, length=0.18 * max(b - a for a, b in lims), normalize=True, color='lightsteelblue', alpha=0.7)
    tt = np.linspace(*[float(c) for c in t_range], 200)
    px, py, pz = [np.asarray(c) * np.ones_like(tt) for c in rf(tt)]
    ax.plot(px, py, pz, 'k', lw=3, label='curve C')
    tk = np.linspace(*[float(c) for c in t_range], 9)
    qx, qy, qz = [np.asarray(c) * np.ones_like(tk) for c in rf(tk)]
    ax.quiver(qx, qy, qz, Fx(qx, qy, qz) * np.ones_like(tk), Fy(qx, qy, qz) * np.ones_like(tk), Fz(qx, qy, qz) * np.ones_like(tk),
              length=0.18 * max(b - a for a, b in lims), normalize=True, color='red', label='F on the curve')
    ax.scatter(px[0], py[0], pz[0], color='green', s=70, label='start'); ax.scatter(px[-1], py[-1], pz[-1], color='purple', s=70, label='end')
    ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z'); ax.set_title(title); ax.legend(loc='upper left', fontsize=8)
    if own: plt.show()
````

# **5.1 Case 1: $\mathbf F=(xy,\,yz,\,zx)$ along $\mathbf r(t)=(\sin t,\cos t,t)$, $0\le t\le\pi$**

The curve is a helix. Along it $x=\sin t$, $y=\cos t$, $z=t$, and $d\mathbf r/dt=(\cos t,-\sin t,1)$, so
$$\mathbf F\cdot d\mathbf r/dt=\sin t\cos^2t-t\sin t\cos t+t\sin t .$$

````python
F1 = sp.Matrix([x * y, y * z, z * x])
r1 = sp.Matrix([sp.sin(t), sp.cos(t), t])

integrand, value = line_integral(F1, r1, (0, sp.pi))
print("Field F       =", list(F1))
print("Curve r(t)    =", list(r1), "  0 <= t <= pi")
print("F(r(t))·r'(t) =", integrand)
print("\nLine integral ∫ F·dr =", value, " ≈", float(value))

# independent numerical check
f_num = sp.lambdify(t, integrand, 'numpy')
print("Numerical value (scipy.integrate.quad) =", quad(f_num, 0, np.pi)[0])
````

````python
fig = plt.figure(figsize=(17, 7))
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
plot_field_and_curve(F1, r1, (0, np.pi), [(-1, 1), (-1, 1), (0, np.pi)], r'Field $\mathbf{F}=(xy,yz,zx)$ and the helix $\mathbf{r}(t)=(\sin t,\cos t,t)$', ax=ax1)

ax2 = fig.add_subplot(1, 2, 2)
tt = np.linspace(0, np.pi, 300)
ax2.plot(tt, f_num(tt), label=r'$\mathbf{F}(\mathbf{r}(t))\cdot\mathbf{r}\,\prime(t)$')
ax2.fill_between(tt, f_num(tt), alpha=0.25, label=f'area = {float(value):.4f}')
ax2.axhline(0, color='k', lw=0.7); ax2.set_xlabel('t'); ax2.set_title('The integrand: the line integral is the area under this curve'); ax2.legend(); ax2.grid(True)
plt.tight_layout()
plt.show()
````

**Is the field conservative?** $\nabla\times\mathbf F=(-y,-z,-x)\neq\mathbf 0$, so the value of the line integral **depends on the path**. We check this by integrating along the *straight line* joining the same end points $(0,1,0)$ and $(0,-1,\pi)$: the result is different.

````python
print("curl F =", list(curl(F1)), " -> not zero: F is not conservative")

s = sp.symbols('s')
start, end = r1.subs(t, 0), r1.subs(t, sp.pi)
straight = start + s * (end - start)                      # straight segment from start to end, 0 <= s <= 1
integrand2, value2 = line_integral(F1, straight.subs(s, t), (0, 1))
print("End points:", list(start), "->", list(end))
print("Along the helix            :", value, "≈", round(float(value), 5))
print("Along the straight segment :", value2, "≈", round(float(value2), 5))
print("Different values -> the work depends on the path.")
````

# **Application Problem 1 – Work done by a force field on a particle**

**Problem statement.** A particle moves along the curve $C$ given by $\mathbf r(t)=(t,\ t^2,\ t^3)$, $0\le t\le1$, in the force field $\mathbf F(x,y,z)=(2xy,\ x^2-y^2,\ z)$. Determine the work done by the force field on the particle.

````python
F2 = sp.Matrix([2 * x * y, x ** 2 - y ** 2, z])
r2 = sp.Matrix([t, t ** 2, t ** 3])

integrand, W = line_integral(F2, r2, (0, 1))
print("F(r(t))·r'(t) =", integrand)
print("Work done W = ∫ F·dr =", W, "≈", float(W), "  (force and length units)")
print("Numerical check      =", quad(sp.lambdify(t, integrand, 'numpy'), 0, 1)[0])

# the field is conservative: use the potential
print("\ncurl F =", list(curl(F2)), " -> conservative")
phi = potential(F2)
print("Potential φ with ∇φ = F :", phi)
print("φ(end) - φ(start) = φ(1,1,1) - φ(0,0,0) =", phi.subs({x: 1, y: 1, z: 1}) - phi.subs({x: 0, y: 0, z: 0}), "  = W")

# path independence: straight line from (0,0,0) to (1,1,1)
_, W_straight = line_integral(F2, sp.Matrix([t, t, t]), (0, 1))
print("Work along the straight line (0,0,0)->(1,1,1) =", W_straight, " (the same, as it must be)")
````

# **Evaluating Line Integrals Using Green's Theorem**

# **Aim:**

To evaluate the line integral of a vector field along a closed plane curve using Green's theorem, to verify the result by direct evaluation of the line integral, and to visualise the vector field and the region.

# **Software required:**

Python 3, SymPy, NumPy, Matplotlib

# **Theory:**

**Green's theorem.** Let $C$ be a positively oriented (counter-clockwise), piecewise-smooth, simple closed curve in the plane bounding a region $D$, and let $\mathbf F=(P,Q)$ have continuous first partial derivatives on $D$. Then

$$\oint_C P\,dx+Q\,dy=\iint_D\left(\frac{\partial Q}{\partial x}-\frac{\partial P}{\partial y}\right)dA .$$

The integrand $\dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}$ is the $z$-component of $\nabla\times\mathbf F$ — the *circulation per unit area* — so Green's theorem says that the circulation round the boundary equals the total curl inside.

**Useful consequences**

* Area of $D$: $A=\dfrac12\oint_C(x\,dy-y\,dx)$ (take $\mathbf F=\tfrac12(-y,x)$, curl $=1$).
* If $\partial Q/\partial x=\partial P/\partial y$ the circulation around every closed curve is zero.
* For regions bounded by circles or ellipses, polar coordinates ($dA=r\,dr\,d\theta$) or generalised polar coordinates $x=ar\cos\theta,\ y=br\sin\theta$ ($dA=abr\,dr\,d\theta$) are convenient.

# **Algorithm:**

1. Import the libraries SymPy and NumPy (and Matplotlib).
2. Define the vector field $\mathbf F=(P,Q)$.
3. Define the region $D$.
4. Compute $\partial Q/\partial x-\partial P/\partial y$ and the double integral in polar coordinates.
5. Check by evaluating the line integral directly on the parametrised boundary.
6. Compute the vector field at specific points / on a grid.
7. Plot the vector field and the region $D$ with its boundary.

**Case.** Evaluate the line integral of the vector field $\mathbf F(x,y)=(x^2-y,\ x+y^2)$ along the positively oriented unit circle $C:\ x^2+y^2=1$ using Green's theorem.

# **6.1 Case: $\mathbf F=(x^2-y,\ x+y^2)$ on the unit circle**

*Method 1 – Green's theorem:* $\dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}=1-(-1)=2$, so the line integral is $\iint_D2\,dA=2\times(\text{area})=2\pi$ — the integral is evaluated in polar coordinates over $0\le r\le1$, $0\le\theta\le2\pi$ with the Jacobian $r$.

*Method 2 – direct evaluation:* parametrise $C$ by $x=\cos t,\ y=\sin t$ and evaluate $\int_0^{2\pi}(P\,x'+Q\,y')\,dt$. Here $P\,x'+Q\,y'=1+\sin^2t\cos t-\sin t\cos^2t$: the constant part gives $2\pi$ and the two oscillating terms integrate to zero.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x, y, r, theta, t = sp.symbols('x y r theta t')

# STEP 2: vector field F = (P, Q)
P = x ** 2 - y
Q = x + y ** 2

# STEP 4: Green's integrand  ∂Q/∂x − ∂P/∂y
dQ_dx, dP_dy = sp.diff(Q, x), sp.diff(P, y)
integrand = sp.simplify(dQ_dx - dP_dy)
print("∂Q/∂x =", dQ_dx, "    ∂P/∂y =", dP_dy, "    ∂Q/∂x − ∂P/∂y =", integrand)

# double integral over the unit disc in polar coordinates (Jacobian r)
x_polar, y_polar = r * sp.cos(theta), r * sp.sin(theta)
integrand_polar = integrand.subs({x: x_polar, y: y_polar}) * r
by_green = sp.integrate(integrand_polar, (r, 0, 1), (theta, 0, 2 * sp.pi))
print("\nMethod 1 (Green's theorem):   ∬ (∂Q/∂x − ∂P/∂y) dA =", by_green)

# STEP 5: direct line integral on the circle  x = cos t, y = sin t
xt, yt = sp.cos(t), sp.sin(t)
direct_integrand = sp.expand((P * sp.diff(xt, t) + Q * sp.diff(yt, t)).subs({x: xt, y: yt}))   # = 1 + sin²t cos t − sin t cos²t
direct = sp.integrate(direct_integrand, (t, 0, 2 * sp.pi))
print("Method 2 (direct):            ∮ P dx + Q dy  = ∫₀^2π", direct_integrand, "dt =", direct)
print("Both methods agree:", sp.simplify(by_green - direct) == 0)
````

````python
Px, Qx = sp.lambdify((x, y), P, 'numpy'), sp.lambdify((x, y), Q, 'numpy')
Xg, Yg = np.meshgrid(np.linspace(-1.6, 1.6, 17), np.linspace(-1.6, 1.6, 17))
tt = np.linspace(0, 2 * np.pi, 400)

fig, ax = plt.subplots(1, 2, figsize=(15, 6.5))
ax[0].fill(np.cos(tt), np.sin(tt), color='lightyellow', alpha=0.8, label='region D (unit disc)')
ax[0].quiver(Xg, Yg, Px(Xg, Yg), Qx(Xg, Yg), np.hypot(Px(Xg, Yg), Qx(Xg, Yg)), cmap='viridis')
ax[0].plot(np.cos(tt), np.sin(tt), 'r', lw=3, label='boundary C (counter-clockwise)')
for a in (0.5, 2.5, 4.4):
    ax[0].annotate('', xy=(np.cos(a + 0.12), np.sin(a + 0.12)), xytext=(np.cos(a), np.sin(a)), arrowprops=dict(arrowstyle='-|>', color='red', lw=2.5))
ax[0].set_title(r'Field $\mathbf{F}=(x^2-y,\ x+y^2)$ and the unit disc'); ax[0].set_aspect('equal'); ax[0].legend(loc='upper right', fontsize=8)

f_dir = sp.lambdify(t, direct_integrand, 'numpy')
ax[1].plot(tt, f_dir(tt) * np.ones_like(tt), label=r'$(P\,x\prime + Q\,y\prime)(t)$ along the circle')
ax[1].fill_between(tt, f_dir(tt) * np.ones_like(tt), alpha=0.3, label=f'area = {float(direct):.4f} = 2π')
ax[1].set_xlabel('t'); ax[1].set_title('Direct line integral: integrand along the circle (area = line integral)'); ax[1].legend(); ax[1].set_ylim(0, 2)
for a in ax: a.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

# **Application Problem 1 – Circulation of a velocity field (airflow)**

**Problem statement.** Consider an airflow over a flat plate in the plane, as in an aerodynamic test of aircraft wings or flat surfaces. The velocity field is $\mathbf F(x,y)=(-y,\,x)$. Determine the **circulation** of this airflow around an elliptical loop with semi-major axis $a$ and semi-minor axis $b$, i.e. $\dfrac{x^2}{a^2}+\dfrac{y^2}{b^2}=1$. (Circulation determines, for instance, the lift of a wing, which is why it is important for optimising designs.)

*Green's theorem:* here $P=-y$, $Q=x$, so $\dfrac{\partial Q}{\partial x}-\dfrac{\partial P}{\partial y}=2$ and the circulation is $2\times\text{area of the ellipse}=2\pi ab$. The area integral is computed with generalised polar coordinates $x=ar\cos\theta,\ y=br\sin\theta$ ($dA=ab\,r\,dr\,d\theta$). The direct parametrisation $x=a\cos t,\ y=b\sin t$ serves as a check.

````python
a, b, R = sp.symbols('a b R', positive=True)
P2, Q2 = -y, x
curl_z = sp.diff(Q2, x) - sp.diff(P2, y)
print("Integrand ∂Q/∂x − ∂P/∂y =", curl_z)

# Method 1: Green's theorem with generalised polar coordinates
circ_ellipse = sp.integrate(curl_z * a * b * r, (r, 0, 1), (theta, 0, 2 * sp.pi))
print("\nMethod 1 (Green):  circulation around the ellipse = ∬ 2 dA =", circ_ellipse, "  (= 2 × area, area = π a b)")

# Method 2: direct line integral around the ellipse
xt, yt = a * sp.cos(t), b * sp.sin(t)
direct_e = sp.integrate(sp.simplify((P2 * sp.diff(xt, t) + Q2 * sp.diff(yt, t)).subs({x: xt, y: yt})), (t, 0, 2 * sp.pi))
print("Method 2 (direct): circulation around the ellipse =", sp.simplify(direct_e))

# area from Green's theorem:  A = ½ ∮ (x dy − y dx)
area = sp.integrate(sp.simplify(sp.Rational(1, 2) * (xt * sp.diff(yt, t) - yt * sp.diff(xt, t))), (t, 0, 2 * sp.pi))
print("Area of the ellipse from  ½∮(x dy − y dx) =", sp.simplify(area))

# the circle is the special case a = b = R
print("\nSpecial case a = b = R (a circle):", sp.simplify(circ_ellipse.subs({a: R, b: R})))
````

````python
import pandas as pd

rows = []
for av, bv in ((1, 1), (2, 1), (3, 2), (4, 4)):
    green = float(circ_ellipse.subs({a: av, b: bv}))
    direct_v = float(sp.simplify(direct_e).subs({a: av, b: bv}))
    rows.append({"a": av, "b": bv, "area πab": round(np.pi * av * bv, 4), "circulation (Green) 2πab": round(green, 4), "circulation (direct)": round(direct_v, 4)})
display(pd.DataFrame(rows).set_index(["a", "b"]))

av, bv = 3, 2
tt = np.linspace(0, 2 * np.pi, 400)
Xg, Yg = np.meshgrid(np.linspace(-4, 4, 17), np.linspace(-3, 3, 13))

fig, ax = plt.subplots(figsize=(8.5, 6.8))
ax.fill(av * np.cos(tt), bv * np.sin(tt), color='lightyellow', label=f'region D: area = πab = {np.pi * av * bv:.2f}')
ax.quiver(Xg, Yg, -Yg, Xg, np.hypot(Xg, Yg), cmap='viridis')
ax.plot(av * np.cos(tt), bv * np.sin(tt), 'r', lw=3, label='ellipse C (counter-clockwise)')
ax.set_title(f'Airflow F = (−y, x) around the ellipse a = {av}, b = {bv}:  circulation = 2πab = {2 * np.pi * av * bv:.2f}')
ax.set_aspect('equal'); ax.legend(loc='upper right', fontsize=8); ax.grid(True, alpha=0.3)
plt.show()
````

# **Application Problem 2 – Circulation of an induced electric field**

**Problem statement.** A rotating magnetic field in a plane induces an electric field (Faraday's law of induction) that can be described by $\mathbf F(x,y)=(-y,\,x)$. Determine the circulation of this electric field around a circular loop of radius $R$, $x^2+y^2=R^2$. (This is important for understanding induced fields in applications such as MRI or the design of electric generators and motors.)

The circulation of the induced field around the loop is the **induced EMF**; by Green's theorem it equals the total curl $2$ times the area $\pi R^2$ of the loop.

````python
# Green's theorem in polar coordinates over the disc of radius R
circ_circle = sp.integrate(sp.diff(Q2, x) * r - sp.diff(P2, y) * r, (r, 0, R), (theta, 0, 2 * sp.pi))
print("Circulation of the electric field around the circle of radius R:", circ_circle)

# direct check
xt, yt = R * sp.cos(t), R * sp.sin(t)
print("Direct line integral:", sp.integrate(sp.simplify((P2 * sp.diff(xt, t) + Q2 * sp.diff(yt, t)).subs({x: xt, y: yt})), (t, 0, 2 * sp.pi)))
print("Interpretation: the circulation grows with the AREA of the loop (∝ R²):  R = 1, 2, 3  ->", [float(circ_circle.subs(R, v)) for v in (1, 2, 3)])

Rv = 2
tt = np.linspace(0, 2 * np.pi, 400)
Xg, Yg = np.meshgrid(np.linspace(-3, 3, 15), np.linspace(-3, 3, 15))
fig, ax = plt.subplots(1, 2, figsize=(15, 6.3))
ax[0].fill(Rv * np.cos(tt), Rv * np.sin(tt), color='lightyellow')
ax[0].quiver(Xg, Yg, -Yg, Xg, np.hypot(Xg, Yg), cmap='plasma')
ax[0].plot(Rv * np.cos(tt), Rv * np.sin(tt), 'r', lw=3)
ax[0].set_title(f'Induced electric field E = (−y, x) and the loop R = {Rv}'); ax[0].set_aspect('equal')

Rs = np.linspace(0, 3, 100)
ax[1].plot(Rs, 2 * np.pi * Rs ** 2, 'b', label='circulation = 2πR²')
ax[1].plot(Rv, 2 * np.pi * Rv ** 2, 'ro'); ax[1].set_xlabel('loop radius R'); ax[1].set_ylabel('circulation (induced EMF)'); ax[1].legend(); ax[1].grid(True)
ax[1].set_title('Circulation increases with the area of the loop')
plt.tight_layout()
plt.show()
````

# **Discussion**

Green's theorem replaces a line integral round a closed curve by a double integral of the curl over the enclosed region, so a complicated boundary is traded for a simple (here constant) integrand: the curl of $(-y,x)$ is the constant $2$, hence the circulation is simply $2\times$ area, whatever the shape of the loop. This explains why bigger loops give larger induced EMF. The check by the direct parametrisation in every problem guards against sign errors (the boundary must be traversed counter-clockwise, otherwise the sign of the answer changes). The formula $A=\frac12\oint(x\,dy-y\,dx)$ obtained from Green's theorem is the basis of the "planimeter" and of area computations from the coordinates of a polygon.

# **Evaluating Surface Integrals Using Stokes' Theorem**

# **Aim:**

To evaluate the surface integral of the curl of a vector field over a surface $S$ using Stokes' theorem, to verify it by the line integral around the boundary, and to visualise the surface and the vector field.

# **Software required:**

Python 3, SymPy, NumPy, Matplotlib (with 3-D support)

# **Theory:**

**Stokes' theorem.** If $S$ is an oriented, piecewise-smooth surface with boundary curve $C$, traversed so that the surface is on the left (right-hand rule with respect to the normal $\mathbf n$ of $S$), then

$$\oint_C\mathbf F\cdot d\mathbf r=\iint_S(\nabla\times\mathbf F)\cdot\mathbf n\,dS .$$

* For a surface with parametrisation $\mathbf r(u,v)$ the *oriented* surface element is $\mathbf n\,dS=\left(\dfrac{\partial\mathbf r}{\partial u}\times\dfrac{\partial\mathbf r}{\partial v}\right)du\,dv$, so $\iint_S(\nabla\times\mathbf F)\cdot d\mathbf S=\iint_D(\nabla\times\mathbf F)\cdot(\mathbf r_u\times\mathbf r_v)\,du\,dv$.
* **The orientation matters:** reversing the normal (and the direction of the boundary) changes the sign of both sides.
* The surface integral depends only on the **boundary** $C$: *any* surface with the same boundary gives the same value (as long as the orientation is consistent). Green's theorem is the special case of a flat surface in the $xy$-plane.

# **Algorithm:**

1. Import the libraries SymPy, NumPy, Matplotlib.
2. Define the vector field $\mathbf F$.
3. Define the surface $S$ (parametrisation) and its boundary curve $C$.
4. Compute the curl of the vector field.
5. Evaluate the surface integral $\iint_S(\nabla\times\mathbf F)\cdot d\mathbf S$ with the parametrisation.
6. Evaluate the line integral $\oint_C\mathbf F\cdot d\mathbf r$ on the boundary to verify Stokes' theorem.
7. Define the grid for the surface.
8. Create the 3-D plot of the surface, the boundary and the vector field.

**Case.** Evaluate $\iint_S(\nabla\times\mathbf F)\cdot d\mathbf S$ for $\mathbf F(x,y,z)=(y,z,x)$ over the part of the plane $z=1-x-y$ in the first octant, using Stokes' theorem. (The surface is the triangle with corners $(1,0,0)$, $(0,1,0)$, $(0,0,1)$ and the normal is taken upward.)

# **7.1 The surface integral of the curl**

The surface is parametrised by $\mathbf r(u,v)=(u,\,v,\,1-u-v)$ over the triangle $D:\ u\ge0,\ v\ge0,\ u+v\le1$. Then $\mathbf r_u=(1,0,-1)$, $\mathbf r_v=(0,1,-1)$ and $\mathbf r_u\times\mathbf r_v=(1,1,1)$, which points **upward** (positive $z$-component) — this is the orientation used.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

x, y, z, u, v, t = sp.symbols('x y z u v t')

def curl(F):
    return sp.Matrix([sp.diff(F[2], y) - sp.diff(F[1], z), sp.diff(F[0], z) - sp.diff(F[2], x), sp.diff(F[1], x) - sp.diff(F[0], y)])

def line_integral(F, r, t_range):
    F_rt = F.subs({x: r[0], y: r[1], z: r[2]}, simultaneous=True)
    return sp.integrate(sp.simplify(F_rt.dot(r.diff(t))), (t, *t_range))

# STEP 2: vector field
F = sp.Matrix([y, z, x])
curl_F = curl(F)
print("F =", list(F), "     curl F =", list(curl_F))

# STEP 3: surface z = 1 − x − y, parametrised by (u, v)
r_uv = sp.Matrix([u, v, 1 - u - v])
r_u, r_v = r_uv.diff(u), r_uv.diff(v)
normal = r_u.cross(r_v)
print("r_u =", list(r_u), "  r_v =", list(r_v), "  r_u × r_v =", list(normal), "  -> upward normal")

# STEP 5: surface integral over the triangle 0 <= v <= 1-u, 0 <= u <= 1
integrand = sp.simplify(curl_F.subs({x: r_uv[0], y: r_uv[1], z: r_uv[2]}, simultaneous=True).dot(normal))
surface_integral = sp.integrate(integrand, (v, 0, 1 - u), (u, 0, 1))
print("\n(curl F)·(r_u × r_v) =", integrand, "  (constant),   area of the triangle D in the uv-plane = 1/2")
print("Surface integral  ∬ (curl F)·dS  =", surface_integral)
````

# **7.2 Verification with the line integral around the boundary**

Seen from above (the tip of the normal) the boundary is traversed counter-clockwise: $(1,0,0)\to(0,1,0)\to(0,0,1)\to(1,0,0)$. Each edge is a straight segment $\mathbf r_i(s)=A+s\,(B-A)$, $0\le s\le1$. By Stokes' theorem the sum of the three line integrals must equal the surface integral.

````python
import pandas as pd

corners = [sp.Matrix([1, 0, 0]), sp.Matrix([0, 1, 0]), sp.Matrix([0, 0, 1])]
rows, total = [], 0
for i in range(3):
    A_, B_ = corners[i], corners[(i + 1) % 3]
    seg = A_ + t * (B_ - A_)
    val = line_integral(F, seg, (0, 1))
    total += val
    rows.append({"Edge": f"({', '.join(map(str, A_))}) → ({', '.join(map(str, B_))})", "∫ F·dr": val})
rows.append({"Edge": "TOTAL  ∮_C F·dr", "∫ F·dr": total})
display(pd.DataFrame(rows).set_index("Edge"))

print("Surface integral ∬ curl F · dS =", surface_integral, "     Line integral ∮ F·dr =", total, "    Stokes' theorem verified:", surface_integral == total)
print("With the opposite orientation (downward normal, boundary clockwise) both integrals change sign:", -surface_integral)
````

````python
fig = plt.figure(figsize=(17, 7.5))
ax = fig.add_subplot(1, 2, 1, projection='3d')

tri = [[(1, 0, 0), (0, 1, 0), (0, 0, 1)]]
ax.add_collection3d(Poly3DCollection(tri, alpha=0.35, facecolor='lightblue', edgecolor='k'))

# vector field F = (y, z, x) at points of the surface
uu, vv = np.meshgrid(np.linspace(0.05, 0.85, 6), np.linspace(0.05, 0.85, 6))
ok = uu + vv <= 0.95
Xp, Yp, Zp = uu[ok], vv[ok], 1 - uu[ok] - vv[ok]
ax.quiver(Xp, Yp, Zp, Yp, Zp, Xp, length=0.22, normalize=True, color='red', label='F = (y, z, x)')

# boundary orientation arrows and the normal vector
for (A_, B_) in zip(corners, corners[1:] + corners[:1]):
    A_, B_ = np.array(A_, float).ravel(), np.array(B_, float).ravel()
    m = (A_ + B_) / 2; d = (B_ - A_) / np.linalg.norm(B_ - A_)
    ax.quiver(*(m - 0.1 * d), *(0.2 * d), color='green', lw=3, arrow_length_ratio=0.5)
c = np.array([1 / 3, 1 / 3, 1 / 3])
ax.quiver(*c, 0.3, 0.3, 0.3, color='purple', lw=3, label='normal n (upward)')
ax.quiver(0.7, 0.7, 0.7, 0, 0, 0, color='white')
ax.plot([], [], color='green', lw=3, label='boundary direction')
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z'); ax.set_title('Surface z = 1 − x − y (first octant), boundary C and F')
ax.legend(loc='upper left', fontsize=8)

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.add_collection3d(Poly3DCollection(tri, alpha=0.3, facecolor='lightblue', edgecolor='k'))
ax2.quiver(Xp, Yp, Zp, -np.ones_like(Xp), -np.ones_like(Xp), -np.ones_like(Xp), length=0.2, normalize=True, color='darkorange', pivot='middle')
ax2.set_title(r'curl $\mathbf{F}=(-1,-1,-1)$ on the surface: points against the normal $\Rightarrow$ negative flux')
ax2.set_xlabel('x'); ax2.set_ylabel('y'); ax2.set_zlabel('z')
for a_ in (ax, ax2): a_.set_xlim(0, 1); a_.set_ylim(0, 1); a_.set_zlim(0, 1); a_.set_box_aspect((1, 1, 1))
plt.tight_layout()
plt.show()
````

# **Application Problem – Circulation of a fluid in a spherical tank**

**Problem statement.** A spherical tank is partially filled with water. The upper hemisphere of the sphere $x^2+y^2+z^2=R^2$ ($z\ge0$, from the equator upwards) is the surface $S$. The fluid flows with the velocity field $\mathbf V(x,y,z)=(y,\,-x,\,z)$. Compute the **circulation** of the fluid around the boundary of $S$ (the equator) by Stokes' theorem, and check it by a direct calculation.

* $\nabla\times\mathbf V=(0,\,0,\,-2)$.
* The hemisphere is parametrised by $\mathbf r(\varphi,\theta)=(R\sin\varphi\cos\theta,\ R\sin\varphi\sin\theta,\ R\cos\varphi)$, $0\le\varphi\le\pi/2$, $0\le\theta\le2\pi$; $\mathbf r_\varphi\times\mathbf r_\theta=R\sin\varphi\,\mathbf r$ points **outward** (upward on the hemisphere), so the equator is traversed counter-clockwise seen from above.

````python
R = sp.symbols('R', positive=True)
phi, th = sp.symbols('phi theta')

V = sp.Matrix([y, -x, z])
curl_V = curl(V)
print("V =", list(V), "     curl V =", list(curl_V))

# --- surface integral over the upper hemisphere (outward normal)
r_s = sp.Matrix([R * sp.sin(phi) * sp.cos(th), R * sp.sin(phi) * sp.sin(th), R * sp.cos(phi)])
n_dS = sp.simplify(r_s.diff(phi).cross(r_s.diff(th)))
print("\nr_φ × r_θ =", [sp.simplify(c) for c in n_dS], "  = R sinφ · r  (outward)")
integrand = sp.simplify(curl_V.dot(n_dS))
surface = sp.integrate(integrand, (phi, 0, sp.pi / 2), (th, 0, 2 * sp.pi))
print("(curl V)·(r_φ × r_θ) =", integrand)
print("Surface integral over the hemisphere  ∬ (curl V)·dS =", surface)

# --- line integral around the equator, counter-clockwise (seen from +z)
equator = sp.Matrix([R * sp.cos(t), R * sp.sin(t), 0])
circulation = line_integral(V, equator, (0, 2 * sp.pi))
print("\nLine integral around the equator ∮ V·dr =", circulation)
print("Stokes' theorem verified:", sp.simplify(surface - circulation) == 0)

# --- any other surface with the same boundary gives the same value: the flat disc (z = 0, normal +z)
rho = sp.symbols('rho', positive=True)
disc = sp.integrate(curl_V[2] * rho, (rho, 0, R), (th, 0, 2 * sp.pi))
print("Flat disc spanning the equator (normal +z):  ∬ (curl V)·dS =", disc, "  -> the same value")
print("\nThe circulation is NEGATIVE: V = (y, −x, ·) rotates clockwise when seen from above.  For R = 1:", float(circulation.subs(R, 1)), "= -2π")
````

````python
Rv = 1.0
ph, tt = np.meshgrid(np.linspace(0, np.pi / 2, 20), np.linspace(0, 2 * np.pi, 40))
Xs, Ys, Zs = Rv * np.sin(ph) * np.cos(tt), Rv * np.sin(ph) * np.sin(tt), Rv * np.cos(ph)

fig = plt.figure(figsize=(17, 7.5))
ax = fig.add_subplot(1, 2, 1, projection='3d')
ax.plot_surface(Xs, Ys, Zs, alpha=0.3, color='lightblue', edgecolor='none')
sel = (slice(2, None, 5), slice(None, None, 4))
ax.quiver(Xs[sel], Ys[sel], Zs[sel], Ys[sel], -Xs[sel], Zs[sel], length=0.25, normalize=True, color='red', label='V = (y, −x, z)')
te = np.linspace(0, 2 * np.pi, 200)
ax.plot(Rv * np.cos(te), Rv * np.sin(te), 0, 'g', lw=3, label='boundary C (equator)')
for a_ in np.linspace(0, 2 * np.pi, 5)[:-1] + 0.3:
    ax.quiver(Rv * np.cos(a_), Rv * np.sin(a_), 0, -np.sin(a_) * 0.35, np.cos(a_) * 0.35, 0, color='green', lw=3, arrow_length_ratio=0.5)
ax.set_title('Upper hemisphere, its boundary (counter-clockwise) and the velocity field'); ax.legend(loc='upper left', fontsize=8)
ax.set_box_aspect((2, 2, 1))
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z')

ax2 = fig.add_subplot(1, 2, 2)
Xg, Yg = np.meshgrid(np.linspace(-1.3, 1.3, 14), np.linspace(-1.3, 1.3, 14))
ax2.quiver(Xg, Yg, Yg, -Xg, np.hypot(Xg, Yg), cmap='viridis')
ax2.plot(np.cos(te), np.sin(te), 'g', lw=3)
ax2.set_title('Top view: the flow turns CLOCKWISE while C is counter-clockwise → circulation = −2πR²'); ax2.set_aspect('equal'); ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

# **Result**

* Case: $\mathbf F=(y,z,x)$ has $\nabla\times\mathbf F=(-1,-1,-1)$; over the triangle $z=1-x-y$ (first octant, upward normal $(1,1,1)$) $\ \iint_S(\nabla\times\mathbf F)\cdot d\mathbf S=-3\times\tfrac12=-\dfrac32$. The three edge integrals of the boundary are $-\tfrac12$ each, giving $\oint_C\mathbf F\cdot d\mathbf r=-\dfrac32$ — Stokes' theorem is verified. (With the opposite orientation the value is $+\tfrac32$.)
* Application: for $\mathbf V=(y,-x,z)$ the circulation around the equator of the hemisphere is $\displaystyle\oint_C\mathbf V\cdot d\mathbf r=\iint_S(\nabla\times\mathbf V)\cdot d\mathbf S=-2\pi R^2$; the flat disc bounded by the same circle gives the same value.

# **Discussion**

Stokes' theorem relates the *rotation inside* a surface to the *circulation on its boundary*. Its two sides are very different calculations — a double integral over the surface and a single integral over the edge — which makes it an excellent self-check. The **orientation** decides the sign: with the upward normal the triangle gives $-3/2$, and the same triangle with the downward normal would give $+3/2$; for the tank the clockwise swirl gives a negative circulation when the boundary is traversed counter-clockwise. The surface-independence (hemisphere versus flat disc) shows that only the boundary matters when the curl is known, so the simplest surface can be chosen — the key idea behind the integral form of Faraday's and Ampère's laws.

# **Evaluating Volume Integrals Using the Divergence Theorem**

# **Aim:**

To evaluate the volume integral of the divergence of a vector field over a region using the Divergence (Gauss) theorem, to verify the result by computing the flux through the bounding surface, and to visualise the vector field and the surface.

# **Software required:**

Python 3, SymPy, NumPy, Matplotlib (with 3-D support)

# **Theory:**

**Divergence theorem.** If $V$ is a solid region bounded by a closed, piecewise-smooth surface $S$ with the **outward** unit normal $\mathbf n$, then

$$\oiint_S\mathbf F\cdot\mathbf n\,dS=\iiint_V(\nabla\cdot\mathbf F)\,dV .$$

The left side is the **net outward flux** of $\mathbf F$ through $S$; the right side adds up the sources ($\nabla\cdot\mathbf F>0$) and sinks ($\nabla\cdot\mathbf F<0$) inside $V$. For an incompressible fluid ($\nabla\cdot\mathbf v=0$) the net outflow through any closed surface is zero.

**Coordinates.** For a ball the spherical coordinates $x=r\sin\varphi\cos\theta,\ y=r\sin\varphi\sin\theta,\ z=r\cos\varphi$, $dV=r^2\sin\varphi\,dr\,d\varphi\,d\theta$ are convenient; on a sphere of radius $R$ the oriented surface element is $\mathbf n\,dS=\mathbf r_\varphi\times\mathbf r_\theta\,d\varphi\,d\theta$ (outward).

# **Algorithm:**

1. Import the libraries SymPy, NumPy, Matplotlib.
2. Define the vector field $\mathbf F$.
3. Compute the divergence of the vector field.
4. Evaluate the volume integral $\iiint_V\nabla\cdot\mathbf F\,dV$ using suitable coordinates (e.g. spherical).
5. Evaluate the flux $\oiint_S\mathbf F\cdot\mathbf n\,dS$ through the bounding surface to verify the theorem.
6. Create a grid of points for the visualisation.
7. Create the 3-D plot.
8. Plot the vector field and the bounding surface.

**Case.** Evaluate $\iiint_V\nabla\cdot\mathbf F\,dV$ for $\mathbf F(x,y,z)=(x^2,y^2,z^2)$ over the region enclosed by the unit sphere $x^2+y^2+z^2\le1$, using the Divergence theorem.

# **8.1 Case: $\mathbf F=(x^2,y^2,z^2)$ over the unit ball**

$\nabla\cdot\mathbf F=2x+2y+2z$. This is an **odd** function: the ball is symmetric about the origin, and the values at $\mathbf r$ and $-\mathbf r$ cancel, so the volume integral is **zero**. The flux through the sphere must therefore also be zero: the field flows out through the positive octants ($x,y,z>0$) and in through the negative ones, in equal amounts.

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
import pandas as pd

x, y, z = sp.symbols('x y z')
r, ph, th = sp.symbols('r varphi theta', positive=True)

def div(F):
    return sp.diff(F[0], x) + sp.diff(F[1], y) + sp.diff(F[2], z)

def ball_volume_integral(F, R):
    """∭ div F dV over the ball of radius R (spherical coordinates)."""
    sph = {x: r * sp.sin(ph) * sp.cos(th), y: r * sp.sin(ph) * sp.sin(th), z: r * sp.cos(ph)}
    integrand = sp.simplify(div(F).subs(sph, simultaneous=True) * r ** 2 * sp.sin(ph))
    return sp.integrate(integrand, (r, 0, R), (th, 0, 2 * sp.pi), (ph, 0, sp.pi))

def sphere_flux(F, R):
    """∯ F·n dS through the sphere of radius R, using n dS = r_φ × r_θ dφ dθ (outward)."""
    rs = sp.Matrix([R * sp.sin(ph) * sp.cos(th), R * sp.sin(ph) * sp.sin(th), R * sp.cos(ph)])
    n_dS = rs.diff(ph).cross(rs.diff(th))
    F_s = F.subs({x: rs[0], y: rs[1], z: rs[2]}, simultaneous=True)
    return sp.integrate(sp.simplify(F_s.dot(n_dS)), (th, 0, 2 * sp.pi), (ph, 0, sp.pi))

# STEPS 2-3: field and divergence
F = sp.Matrix([x ** 2, y ** 2, z ** 2])
print("F =", list(F), "      div F =", div(F))

# STEP 4: volume integral over the unit ball
vol = ball_volume_integral(F, 1)
print("\nVolume integral  ∭ div F dV =", vol)

# STEP 5: flux through the unit sphere
flux = sphere_flux(F, 1)
print("Flux through the sphere  ∯ F·n dS =", flux)
print("Divergence theorem verified:", sp.simplify(vol - flux) == 0)
````

**Why zero? – the flux through each octant.** On the unit sphere $\mathbf F\cdot\mathbf n=x^3+y^3+z^3$, positive where the coordinates are positive and negative where they are negative. In the picture the sphere is coloured by $\mathbf F\cdot\mathbf n$ (red = outflow, blue = inflow); the red and blue parts balance exactly.

````python
phi_g, theta_g = np.meshgrid(np.linspace(0, np.pi, 60), np.linspace(0, 2 * np.pi, 120))
Xs, Ys, Zs = np.sin(phi_g) * np.cos(theta_g), np.sin(phi_g) * np.sin(theta_g), np.cos(phi_g)
Fn = Xs ** 3 + Ys ** 3 + Zs ** 3                                     # F·n on the unit sphere
norm = plt.Normalize(-1, 1)

fig = plt.figure(figsize=(17, 7.5))
ax = fig.add_subplot(1, 2, 1, projection='3d')
ax.plot_surface(Xs, Ys, Zs, facecolors=cm.coolwarm(norm(Fn)), rstride=1, cstride=1, shade=False, alpha=0.9)
sel = (slice(4, None, 10), slice(None, None, 12))
ax.quiver(Xs[sel], Ys[sel], Zs[sel], Xs[sel] ** 2, Ys[sel] ** 2, Zs[sel] ** 2, length=0.35, normalize=True, color='k')
fig.colorbar(cm.ScalarMappable(norm=norm, cmap='coolwarm'), ax=ax, shrink=0.6, label='F·n  (red: outflow, blue: inflow)')
ax.set_title(r'Flux density $\mathbf{F}\cdot\mathbf{n}$ of $\mathbf{F}=(x^2,y^2,z^2)$ on the unit sphere'); ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z'); ax.set_box_aspect((1, 1, 1))

ax2 = fig.add_subplot(1, 2, 2)
ax2.hist(Fn.ravel(), bins=40, weights=np.abs(np.sin(phi_g)).ravel(), color='tab:purple', alpha=0.8)
ax2.axvline(0, color='k'); ax2.set_xlabel('F·n'); ax2.set_ylabel('surface area with this value'); ax2.set_title('Outflow and inflow areas are symmetric → net flux 0')
plt.tight_layout()
plt.show()
````

**Other fields over the unit ball** — the theorem is checked for fields with a non-zero answer. For $\mathbf F=(x,y,z)$, $\nabla\cdot\mathbf F=3$ and the integral is $3\times\frac43\pi=4\pi$; for $\mathbf F=(x^3,y^3,z^3)$, $\nabla\cdot\mathbf F=3(x^2+y^2+z^2)=3r^2$ and the integral is $3\int_0^1r^2\cdot4\pi r^2dr=\dfrac{12\pi}{5}$.

````python
fields = {"(x², y², z²)": sp.Matrix([x ** 2, y ** 2, z ** 2]),
          "(x, y, z)":    sp.Matrix([x, y, z]),
          "(x³, y³, z³)": sp.Matrix([x ** 3, y ** 3, z ** 3]),
          "(y, z, x)":    sp.Matrix([y, z, x])}
rows = []
for name, Fv in fields.items():
    vol, fl = ball_volume_integral(Fv, 1), sphere_flux(Fv, 1)
    rows.append({"F": name, "div F": div(Fv), "∭ div F dV": vol, "∯ F·n dS": fl, "equal?": sp.simplify(vol - fl) == 0})
display(pd.DataFrame(rows).set_index("F"))
````

# **Application Problem 1 – Net outflow of water from a cubic tank**

**Problem statement.** A cubic water tank with side length $2a$ is centred at the origin ($-a\le x,y,z\le a$). The water flows with the velocity field $\mathbf V(x,y,z)=(2x,\,-y,\,z)$. Determine the net outflow of water through the surfaces of the tank.

$\nabla\cdot\mathbf V=2-1+1=2$, so $\iiint\nabla\cdot\mathbf V\,dV=2\times(2a)^3=16a^3$. The outflow through each of the six faces is computed separately as a check.

````python
a = sp.symbols('a', positive=True)
V = sp.Matrix([2 * x, -y, z])
print("V =", list(V), "     div V =", div(V))
vol_cube = sp.integrate(div(V), (x, -a, a), (y, -a, a), (z, -a, a))
print("Volume integral ∭ div V dV = 2·(2a)³ =", vol_cube)

# flux through each face: ∬ V·n dA, with n = ± unit vector of the axis
axes = (x, y, z)
rows, total = [], 0
for i, name in enumerate("xyz"):
    others = [axes[j] for j in range(3) if j != i]
    for s, label in ((1, "+"), (-1, "−")):
        comp = V[i].subs(axes[i], s * a)                         # V on the face  axis_i = s·a
        flux_face = sp.integrate(s * comp, (others[0], -a, a), (others[1], -a, a))   # V·n = s·V_i
        total += flux_face
        rows.append({"Face": f"{name} = {label}a", "normal n": f"{label}e_{name}", "V·n on the face": sp.simplify(s * comp), "Flux": flux_face,
                     "": "outflow" if sp.simplify(flux_face).subs(a, 1) > 0 else "inflow"})
rows.append({"Face": "TOTAL", "normal n": "", "V·n on the face": "", "Flux": sp.simplify(total), "": ""})
display(pd.DataFrame(rows).set_index("Face"))
print("Net outflow through the surface =", sp.simplify(total), "   = volume integral:", sp.simplify(total - vol_cube) == 0)
print("For a tank of side 2 m (a = 1 m):  net outflow =", vol_cube.subs(a, 1), "m³ per unit time (in the units of V)")
````

````python
av = 1.0
fig = plt.figure(figsize=(9, 8))
ax = fig.add_subplot(projection='3d')
g = np.linspace(-av, av, 4)
A_, B_ = np.meshgrid(g, g)
O = np.ones_like(A_)
faces = [  # (X, Y, Z, outward normal sign, axis)   colour: red = outflow, blue = inflow
    (av * O, A_, B_, 'x+'), (-av * O, A_, B_, 'x-'), (A_, av * O, B_, 'y+'), (A_, -av * O, B_, 'y-'), (A_, B_, av * O, 'z+'), (A_, B_, -av * O, 'z-')]
for X, Y, Z, name in faces:
    Vx, Vy, Vz = 2 * X, -Y, Z
    nvec = {'x+': (1, 0, 0), 'x-': (-1, 0, 0), 'y+': (0, 1, 0), 'y-': (0, -1, 0), 'z+': (0, 0, 1), 'z-': (0, 0, -1)}[name]
    out = (Vx * nvec[0] + Vy * nvec[1] + Vz * nvec[2]).mean() > 0
    ax.quiver(X, Y, Z, Vx, Vy, Vz, length=0.5, normalize=True, color='red' if out else 'blue')
corners = np.array([[i, j, k] for i in (-av, av) for j in (-av, av) for k in (-av, av)])
for p in corners:
    for q in corners:
        if np.sum(p != q) == 1: ax.plot(*zip(p, q), color='gray', lw=1)
ax.set_title(r'Cubic tank, $\mathbf{V}=(2x,-y,z)$: red = flow out of the face, blue = flow into it' + '\n(outflow through the faces x = ±a and z = ±a exceeds the inflow through y = ±a)', fontsize=10)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z'); ax.set_box_aspect((1, 1, 1))
plt.show()
````

# **Application Problem 2 – Net electric flux through a spherical balloon**

**Problem statement.** A spherical balloon of radius $R$ encloses a region with the electric field $\mathbf E(x,y,z)=(x,\,2y,\,3z)$. Determine the net electric flux through the surface of the balloon.

$\nabla\cdot\mathbf E=1+2+3=6$, so the flux is $6\times\frac43\pi R^3=8\pi R^3$. Gauss's law $\oiint\mathbf E\cdot d\mathbf S=Q_{\text{enclosed}}/\varepsilon_0$ then implies that the enclosed charge density is $\rho=6\varepsilon_0$.

````python
R = sp.symbols('R', positive=True)
E = sp.Matrix([x, 2 * y, 3 * z])
print("E =", list(E), "     div E =", div(E))

vol = ball_volume_integral(E, R)
flux = sphere_flux(E, R)
print("\nVolume integral  ∭ div E dV  =", vol, "   (= 6 × (4/3)πR³)")
print("Surface flux     ∯ E·n dS    =", flux)
print("Divergence theorem verified:", sp.simplify(vol - flux) == 0)
print("For R = 1:  net flux =", float(flux.subs(R, 1)), "  = 8π")
print("Charge density from Gauss's law ρ = ε0 div E = 6 ε0  (uniform positive charge inside)")
````

````python
Rv = 1.0
phi_g, theta_g = np.meshgrid(np.linspace(0, np.pi, 50), np.linspace(0, 2 * np.pi, 100))
Xs, Ys, Zs = Rv * np.sin(phi_g) * np.cos(theta_g), Rv * np.sin(phi_g) * np.sin(theta_g), Rv * np.cos(phi_g)
En = (Xs ** 2 + 2 * Ys ** 2 + 3 * Zs ** 2) / Rv                    # E·n on the sphere (n = r/R)
norm = plt.Normalize(0, En.max())

fig = plt.figure(figsize=(17, 7.5))
ax = fig.add_subplot(1, 2, 1, projection='3d')
ax.plot_surface(Xs, Ys, Zs, facecolors=cm.YlOrRd(norm(En)), shade=False, alpha=0.9)
sel = (slice(4, None, 9), slice(None, None, 10))
ax.quiver(Xs[sel], Ys[sel], Zs[sel], Xs[sel], 2 * Ys[sel], 3 * Zs[sel], length=0.4, normalize=True, color='k')
fig.colorbar(cm.ScalarMappable(norm=norm, cmap='YlOrRd'), ax=ax, shrink=0.6, label='E·n')
ax.set_title(r'Field $\mathbf{E}=(x,2y,3z)$ on the balloon; colour = E·n (everywhere outward)'); ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_zlabel('z'); ax.set_box_aspect((1, 1, 1))

ax2 = fig.add_subplot(1, 2, 2)
Rs = np.linspace(0, 3, 100)
ax2.plot(Rs, 8 * np.pi * Rs ** 3, 'r', lw=2, label=r'flux = $8\pi R^3$'); ax2.plot(Rv, 8 * np.pi, 'ko')
ax2.set_xlabel('balloon radius R'); ax2.set_ylabel('net flux'); ax2.set_title('The flux grows with the enclosed volume (∝ R³)'); ax2.legend(); ax2.grid(True)
plt.tight_layout()
plt.show()
````

# **Result**

* Case: $\mathbf F=(x^2,y^2,z^2)$, $\nabla\cdot\mathbf F=2(x+y+z)$: $\ \iiint_V\nabla\cdot\mathbf F\,dV=0$ over the unit ball, and the flux through the unit sphere is also $0$ — the outflow in the positive octants cancels the inflow in the negative ones. Other fields over the unit ball: $(x,y,z)\to4\pi$, $(x^3,y^3,z^3)\to12\pi/5$, $(y,z,x)\to0$; in every case the volume integral equals the surface flux.
* Cubic tank, $\mathbf V=(2x,-y,z)$: $\nabla\cdot\mathbf V=2$ and the net outflow is $2\,(2a)^3=16a^3$ (face by face: $+8a^3$ through each of the $x$-faces, $-4a^3$ through each of the $y$-faces and $+4a^3$ through each of the $z$-faces).
* Balloon, $\mathbf E=(x,2y,3z)$: $\nabla\cdot\mathbf E=6$ and the net flux is $6\cdot\frac43\pi R^3=8\pi R^3$.

# **Discussion**

The Divergence theorem converts a flux through a closed surface into a volume integral of the divergence. A zero answer is not "no result": for $(x^2,y^2,z^2)$ the divergence is an odd function and the symmetric ball makes the integral vanish, which the colour picture of $\mathbf F\cdot\mathbf n$ on the sphere shows as equal red and blue areas. The cube example shows the theorem at work: water enters through the two $y$-faces and leaves through the four others, with a net outflow of $16a^3$ which is exactly the total source strength $\nabla\cdot\mathbf V=2$ times the volume $8a^3$. Gauss's law in electrostatics, the continuity equation and the heat-flow equation are all statements of this theorem.
