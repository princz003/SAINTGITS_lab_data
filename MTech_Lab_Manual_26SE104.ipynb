# **Theory:**

Computational mathematics uses a computer to *approximate* mathematical quantities (numbers, functions, series, solutions of equations). Python is widely used for this because it is:

* **Readable** – the code looks close to the mathematics.
* **Rich in libraries** – NumPy, SciPy, SymPy, Matplotlib, Pandas (Experiment 2).

Two ideas matter throughout the course:

* **Floating-point arithmetic.** Real numbers are stored with finite precision (about 16 decimal digits), so results are approximate and equality tests must be done with a tolerance.
* **Convergence.** Many numerical methods generate a sequence of approximations $x_0, x_1, x_2, \dots$ that approach the exact answer; we measure the *error* of each approximation.

# **1.4 Functions, loops and list comprehensions**

**Leibniz series for $\pi$:**

$$\frac{\pi}{4} = 1 - \frac13 + \frac15 - \frac17 + \cdots = \sum_{k=0}^{\infty} \frac{(-1)^k}{2k+1}$$

The function below adds the first $N$ terms. The *error* $|\pi_N - \pi|$ shows how the approximation improves as $N$ grows.

```python
def leibniz_pi(terms):
    """Approximate pi using the first `terms` terms of the Leibniz series."""
    total = 0.0
    for k in range(terms):
        total += (-1) ** k / (2 * k + 1)
    return 4 * total

for N in (10, 100, 1000, 100000):
    approx = leibniz_pi(N)
    print(f"N = {N:>6d}   pi ≈ {approx:.8f}   error = {abs(approx - math.pi):.2e}")
```

# **1.5 Basic statistics**

For data $x_1,\dots,x_n$:

$$\bar{x}=\frac1n\sum x_i, \qquad s^2=\frac{1}{n-1}\sum (x_i-\bar{x})^2, \qquad s=\sqrt{s^2}$$

We compute them from the definition and compare with Python's built-in `statistics` module.

```python
import statistics as st

data = [12, 15, 11, 18, 20, 14, 16, 13]

mean = sum(data) / len(data)
var  = sum((d - mean) ** 2 for d in data) / (len(data) - 1)
print(f"From definition : mean = {mean:.4f}, variance = {var:.4f}, std = {var ** 0.5:.4f}")
print(f"statistics module: mean = {st.mean(data):.4f}, variance = {st.variance(data):.4f}, "
      f"std = {st.stdev(data):.4f}, median = {st.median(data)}")
```

# **1.6 Basic plotting**

Plot $f(x)=\dfrac{\sin x}{x}$ and the error of the Leibniz approximation against the number of terms (log–log scale). A straight line on a log–log plot indicates that the error behaves like $N^{-p}$.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-20, 20, 800)
y = np.sinc(x / np.pi)          # np.sinc(t) = sin(pi t)/(pi t)  ->  sin(x)/x

k = np.arange(1, 5001)
partial_sums = np.cumsum(4 * (-1.0) ** (k - 1) / (2 * k - 1))
error = np.abs(partial_sums - np.pi)

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))

ax[0].plot(x, y, color='blue')
ax[0].set_title(r'$f(x)=\sin(x)/x$')
ax[0].set_xlabel('x'); ax[0].set_ylabel('f(x)'); ax[0].grid(True)

ax[1].loglog(k, error, color='red')
ax[1].set_title('Error of the Leibniz approximation of $\\pi$')
ax[1].set_xlabel('number of terms N'); ax[1].set_ylabel('|approximation - pi|'); ax[1].grid(True, which='both')

plt.tight_layout()
plt.show()
```

# **Result**

Python was used as a calculator, for floating-point experiments, for data types, loops, functions and comprehensions, for basic statistics and for plotting. The Leibniz series approximated $\pi$ with the error decreasing roughly like $1/N$ (slope $-1$ on the log–log plot), and the statistics computed from the definition matched the built-in `statistics` module.

# **Discussion**

Floating-point numbers are only approximations (e.g. $0.1+0.2\neq 0.3$), so numerical results should be compared with a tolerance. The Leibniz series converges very slowly — about 5000 terms still leave an error of order $10^{-4}$ — which motivates the faster numerical methods used in later experiments.

# **1. NumPy**

**Purpose:** arrays, broadcasting and vectorised computation.

A NumPy array stores numbers of one type in a block of memory; operations act on the *whole array at once* (vectorisation), which is much faster than a Python loop.

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
M = np.arange(12).reshape(3, 4)

print("a        :", a)
print("a * 2    :", a * 2)
print("a ** 2   :", a ** 2)
print("mean(a)  :", a.mean(), "  std(a):", round(a.std(), 4))
print("\nM =\n", M)
print("shape of M:", M.shape, "  shape of M.T:", M.T.shape)
print("\nlinspace(0, 1, 5):", np.linspace(0, 1, 5))
print("identity(3):\n", np.eye(3))
```

**Broadcasting** lets NumPy combine arrays of different shapes: a column of shape $(3,1)$ plus a row of shape $(4,)$ gives a $3\times4$ table.

```python
col = np.array([[1], [2], [3]])
row = np.array([10, 20, 30, 40])
print(col + row)
```

**Vectorisation vs. loop** — compute $\sum x_i^2$ for one million numbers in two ways.

```python
import time

x = np.random.default_rng(0).random(1_000_000)

t0 = time.perf_counter()
s_loop = sum(v * v for v in x)          # pure Python loop
t1 = time.perf_counter()
s_vec = np.dot(x, x)                     # vectorised NumPy
t2 = time.perf_counter()

print(f"Python loop : {s_loop:.4f}   ({(t1 - t0) * 1000:.1f} ms)")
print(f"NumPy dot   : {s_vec:.4f}   ({(t2 - t1) * 1000:.1f} ms)")
print("Same result :", np.isclose(s_loop, s_vec))
```

# **2. SciPy**

**Purpose:** numerical integration, root finding, scalar optimisation and linear systems.

```python
from scipy import integrate, optimize, linalg

value, abs_err = integrate.quad(np.sin, 0, np.pi)
print(f"Integral of sin(x) on [0, pi] = {value:.10f}   (estimated error {abs_err:.1e})")

res = optimize.minimize_scalar(lambda t: (t - 2) ** 2 + 1)
print(f"Minimum of (x-2)^2 + 1 at x = {res.x:.6f}, value = {res.fun:.6f}")

root = optimize.brentq(lambda t: t ** 3 - 2 * t - 5, 2, 3)
print(f"Root of x^3 - 2x - 5 = 0 in [2, 3]: x = {root:.10f}")

A = np.array([[3., 2.], [1., 4.]])
b = np.array([5., 6.])
print("Solution of A x = b:", linalg.solve(A, b))
```

# **Vector and Matrix Operations Using Python and Visualizing the Results**

# **Aim:**

To perform vector and matrix operations in Python (NumPy) and to visualise vectors and linear transformations.

# **Software required:**

Python 3, NumPy, Matplotlib

# **Theory:**

**Vector operations** for $\mathbf u,\mathbf v\in\mathbb R^n$:

* Dot product: $\mathbf u\cdot\mathbf v=\sum u_iv_i=\|\mathbf u\|\|\mathbf v\|\cos\theta$
* Norm: $\|\mathbf u\|=\sqrt{\mathbf u\cdot\mathbf u}$
* Projection of $\mathbf u$ on $\mathbf v$: $\dfrac{\mathbf u\cdot\mathbf v}{\mathbf v\cdot\mathbf v}\,\mathbf v$
* Cross product (in $\mathbb R^3$): a vector perpendicular to both $\mathbf u$ and $\mathbf v$

**Matrix operations:** addition, product $AB$ (not commutative), transpose $A^T$, determinant, inverse $A^{-1}$ (exists only if $\det A\neq0$), rank and trace.

**Linear transformation:** a matrix $A$ maps a point $\mathbf x$ to $A\mathbf x$. A $2\times2$ matrix rotates, scales or shears the plane and $|\det A|$ is the factor by which areas change.

# **Procedure:**

* Create vectors and compute dot product, cross product, norm, angle and projection.
* Draw 2-D vectors with arrows.
* Perform matrix operations and verify identities.
* Apply rotation, scaling and shear matrices to a shape and plot the result.

# **3.1 Vector operations**

For $\mathbf u=(3,4,0)$ and $\mathbf v=(1,2,2)$: $\mathbf u\cdot\mathbf v=11$, $\|\mathbf u\|=5$, $\|\mathbf v\|=3$.

```python
import numpy as np
import matplotlib.pyplot as plt

u = np.array([3, 4, 0])
v = np.array([1, 2, 2])

dot = np.dot(u, v)
cross = np.cross(u, v)
norm_u, norm_v = np.linalg.norm(u), np.linalg.norm(v)
angle = np.degrees(np.arccos(dot / (norm_u * norm_v)))
proj_u_on_v = (dot / np.dot(v, v)) * v

print("u + v            :", u + v)
print("u - v            :", u - v)
print("3u               :", 3 * u)
print("u . v            :", dot)
print("u x v            :", cross)
print("||u||, ||v||     :", norm_u, norm_v)
print("unit vector of u :", u / norm_u)
print(f"angle between u and v: {angle:.4f} degrees")
print("projection of u on v :", proj_u_on_v.round(4))
print("u x v is perpendicular to u and v:", np.dot(cross, u) == 0 and np.dot(cross, v) == 0)
```

# **3.2 Visualising vectors in 2-D**

Plot $\mathbf a=(3,1)$, $\mathbf b=(1,3)$, their sum and difference, and the projection of $\mathbf a$ on $\mathbf b$ (the dotted line shows the perpendicular component).

```python
a = np.array([3, 1])
b = np.array([1, 3])
proj = (a @ b) / (b @ b) * b

fig, ax = plt.subplots(figsize=(6.5, 6.5))

def arrow(vec, color, label, origin=(0, 0)):
    ax.quiver(*origin, *vec, angles='xy', scale_units='xy', scale=1, color=color, label=label, width=0.008)

arrow(a, 'blue', 'a = (3, 1)')
arrow(b, 'red', 'b = (1, 3)')
arrow(a + b, 'green', 'a + b')
arrow(a - b, 'purple', 'a - b')
arrow(proj, 'orange', 'projection of a on b')
ax.plot([a[0], proj[0]], [a[1], proj[1]], 'k:', lw=1)

ax.set_xlim(-3, 5); ax.set_ylim(-3, 5)
ax.set_aspect('equal'); ax.grid(True)
ax.axhline(0, color='black', lw=0.8); ax.axvline(0, color='black', lw=0.8)
ax.legend(loc='upper left'); ax.set_title('Vector addition, subtraction and projection')
plt.show()
```

# **3.3 Matrix operations**

Take $A=\begin{pmatrix}2&1&0\\1&3&1\\0&1&4\end{pmatrix}$ and $B=\begin{pmatrix}1&0&2\\0&1&1\\3&0&1\end{pmatrix}$. Note that `A @ B` is the matrix product while `A * B` multiplies element by element.

```python
A = np.array([[2, 1, 0],
              [1, 3, 1],
              [0, 1, 4]], dtype=float)
B = np.array([[1, 0, 2],
              [0, 1, 1],
              [3, 0, 1]], dtype=float)

print("A + B =\n", A + B)
print("\nA @ B (matrix product) =\n", A @ B)
print("\nA * B (element-wise)  =\n", A * B)
print("\nAB == BA ?", np.allclose(A @ B, B @ A))
print("\nTranspose of A =\n", A.T)
print("\ndet(A)  =", round(np.linalg.det(A), 6))
print("trace(A) =", np.trace(A))
print("rank(A)  =", np.linalg.matrix_rank(A))

A_inv = np.linalg.inv(A)
print("\nInverse of A =\n", A_inv.round(4))
print("A @ A^-1 = I ?", np.allclose(A @ A_inv, np.eye(3)))
print("det(AB) = det(A) det(B) ?", np.isclose(np.linalg.det(A @ B), np.linalg.det(A) * np.linalg.det(B)))
print("(AB)^T = B^T A^T ?", np.allclose((A @ B).T, B.T @ A.T))
```

A **singular** matrix has $\det=0$ and no inverse: its rows are linearly dependent and its rank is less than $n$.

```python
C = np.array([[1., 2.],
              [2., 4.]])
print("det(C) =", np.linalg.det(C), "   rank(C) =", np.linalg.matrix_rank(C))
try:
    np.linalg.inv(C)
except np.linalg.LinAlgError as err:
    print("Cannot invert C ->", err)
```

Matrices can be visualised as **heat maps** — useful for large matrices.

```python
fig, ax = plt.subplots(1, 3, figsize=(13, 4))
for axis, mat, name in zip(ax, (A, B, A @ B), ('A', 'B', 'A @ B')):
    im = axis.imshow(mat, cmap='coolwarm')
    axis.set_title(name)
    for (i, j), val in np.ndenumerate(mat):
        axis.text(j, i, f"{val:g}", ha='center', va='center', fontsize=12)
    axis.set_xticks(range(3)); axis.set_yticks(range(3))
    plt.colorbar(im, ax=axis, fraction=0.046)
plt.tight_layout()
plt.show()
```

# **3.4 Linear transformations of the plane**

Apply three matrices to a "house" shape and watch how it changes:

* **Rotation** by $45^\circ$: $R=\begin{pmatrix}\cos\theta&-\sin\theta\\ \sin\theta&\cos\theta\end{pmatrix}$ ($\det R=1$, area preserved)
* **Scaling**: $S=\begin{pmatrix}2&0\\0&0.5\end{pmatrix}$ ($\det S=1$)
* **Shear**: $H=\begin{pmatrix}1&1\\0&1\end{pmatrix}$ ($\det H=1$)
* **Projection** onto the $x$-axis: $P=\begin{pmatrix}1&0\\0&0\end{pmatrix}$ ($\det P=0$, the shape collapses to a line)

```python
house = np.array([[0, 0], [1, 0], [1, 1], [0.5, 1.5], [0, 1], [0, 0]]).T    # 2 x 6 array of points

theta = np.radians(45)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
S = np.array([[2, 0], [0, 0.5]])
H = np.array([[1, 1], [0, 1]])
P = np.array([[1, 0], [0, 0]])

transforms = [('Original', np.eye(2)), ('Rotation 45°', R), ('Scaling (2, 0.5)', S),
              ('Shear', H), ('Projection on x-axis', P), ('Rotation then Shear (H R)', H @ R)]

fig, ax = plt.subplots(2, 3, figsize=(14, 8))
for axis, (name, T) in zip(ax.ravel(), transforms):
    new = T @ house
    axis.fill(house[0], house[1], color='lightgray', alpha=0.6, label='original')
    axis.fill(new[0], new[1], color='tab:blue', alpha=0.5, label='transformed')
    axis.set_title(f"{name}\ndet = {np.linalg.det(T):.2f}")
    axis.set_xlim(-2, 3); axis.set_ylim(-2, 3)
    axis.set_aspect('equal'); axis.grid(True)
ax[0, 0].legend()
plt.tight_layout()
plt.show()
```

# **Solving Systems of Linear Equations Using Matrix Methods and Visualizing the Solutions**

# **Aim:**

To solve systems of linear equations $A\mathbf x=\mathbf b$ using matrix methods (inverse, Gaussian elimination, LU decomposition), to decide whether a solution exists, and to visualise the solutions. The extension covers QR decomposition, least squares and the iterative methods of Jacobi and Gauss–Seidel.

# **Software required:**

Python 3, NumPy, SciPy, Matplotlib, Pandas

# **Theory:**

A system of $m$ equations in $n$ unknowns is written $A\mathbf x=\mathbf b$.

**Existence of solutions (Rouché–Capelli theorem).** Let $r=\operatorname{rank}(A)$ and $r_a=\operatorname{rank}([A\,|\,\mathbf b])$:

* $r\neq r_a$ → **no solution** (inconsistent)
* $r=r_a=n$ → **unique solution**
* $r=r_a<n$ → **infinitely many solutions**

**Methods for a square invertible system:**

* *Inverse:* $\mathbf x=A^{-1}\mathbf b$ (simple, but rarely the best numerically)
* *Gaussian elimination:* reduce to upper-triangular form with row operations, then back-substitute (use **partial pivoting** for stability)
* *LU decomposition:* $PA=LU$; then solve $L\mathbf y=P\mathbf b$ and $U\mathbf x=\mathbf y$ — the factorisation is reused for many right-hand sides

**Extension:**

* *Least squares:* for an over-determined system ($m>n$) minimise $\|A\mathbf x-\mathbf b\|_2$; the solution satisfies the normal equations $A^TA\hat{\mathbf x}=A^T\mathbf b$, or with $A=QR$: $R\hat{\mathbf x}=Q^T\mathbf b$.
* *Jacobi:* $x_i^{(k+1)}=\dfrac{1}{a_{ii}}\Big(b_i-\sum_{j\ne i}a_{ij}x_j^{(k)}\Big)$
* *Gauss–Seidel:* same, but uses the newest values $x_j^{(k+1)}$ ($j<i$) as soon as they are available.
  Both converge if $A$ is strictly diagonally dominant (more precisely if the spectral radius of the iteration matrix is $<1$).

# **Procedure:**

* Test the existence of a solution using ranks.
* Solve the same system by inverse, Gaussian elimination, LU and `np.linalg.solve`; compare the residuals.
* Plot a $2\times2$ system as lines (unique / infinite / no solution) and a $3\times3$ system as planes.
* Extension: fit a line by least squares with the normal equations and with QR; implement Jacobi and Gauss–Seidel and plot their convergence.

# **Part A – Matrix methods**

We solve

$$\begin{aligned}2x+y-z&=8\\-3x-y+2z&=-11\\-2x+y+2z&=-3\end{aligned}$$

whose solution is $(x,y,z)=(2,3,-1)$.

# **4.1 Does a solution exist?**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import linalg

A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

def classify(A, b):
    """Rouché–Capelli test."""
    r  = np.linalg.matrix_rank(A)
    ra = np.linalg.matrix_rank(np.column_stack([A, b]))
    n  = A.shape[1]
    if r != ra:
        return f"rank(A)={r}, rank([A|b])={ra} -> NO solution"
    if r == n:
        return f"rank(A)=rank([A|b])={r}=n -> UNIQUE solution"
    return f"rank(A)=rank([A|b])={r}<n={n} -> INFINITELY many solutions"

print(classify(A, b))
print("det(A) =", round(np.linalg.det(A), 6))
```

# **4.2 Solving the system – four ways**

The function `gaussian_elimination` below performs forward elimination with partial pivoting followed by back-substitution.

```python
def gaussian_elimination(A, b):
    A = A.astype(float).copy()
    b = b.astype(float).copy()
    n = len(b)
    # Forward elimination with partial pivoting
    for k in range(n - 1):
        p = np.argmax(np.abs(A[k:, k])) + k          # row with the largest pivot
        if abs(A[p, k]) < 1e-12:
            raise ValueError("Matrix is singular")
        if p != k:
            A[[k, p]] = A[[p, k]]
            b[[k, p]] = b[[p, k]]
        for i in range(k + 1, n):
            m = A[i, k] / A[k, k]                    # multiplier
            A[i, k:] -= m * A[k, k:]
            b[i] -= m * b[k]
    # Back substitution
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - A[i, i + 1:] @ x[i + 1:]) / A[i, i]
    return x

x_inv   = np.linalg.inv(A) @ b                       # 1. inverse
x_gauss = gaussian_elimination(A, b)                  # 2. Gaussian elimination
P, L, U = linalg.lu(A)                                # 3. LU decomposition  A = P L U
lu_piv  = linalg.lu_factor(A)
x_lu    = linalg.lu_solve(lu_piv, b)
x_np    = np.linalg.solve(A, b)                       # 4. library routine (LU based)

print("L =\n", L.round(4))
print("U =\n", U.round(4))
print("P L U == A ?", np.allclose(P @ L @ U, A))

methods = {"Inverse": x_inv, "Gaussian elimination": x_gauss, "LU decomposition": x_lu, "np.linalg.solve": x_np}
table = pd.DataFrame({name: [*x, np.linalg.norm(A @ x - b)] for name, x in methods.items()},
                     index=["x", "y", "z", "residual ||Ax-b||"]).T
display(table)
```

# **Part B – Extension: QR decomposition, least squares and iterative methods**

# **4.5 Least squares and QR decomposition**

Fit a straight line $y=c_0+c_1x$ to six noisy measurements. This gives 6 equations in 2 unknowns — an over-determined system with no exact solution — so we minimise $\|A\mathbf c-\mathbf y\|_2$.

We use the **QR decomposition** $A=QR$ ($Q$ has orthonormal columns, $R$ is upper triangular). Below we implement the classical Gram–Schmidt process to obtain $Q$ and $R$, and then solve $R\mathbf c=Q^T\mathbf y$.

```python
def gram_schmidt_qr(A):
    m, n = A.shape
    Q = np.zeros((m, n)); R = np.zeros((n, n))
    for j in range(n):
        v = A[:, j].copy()
        for i in range(j):
            R[i, j] = Q[:, i] @ A[:, j]
            v -= R[i, j] * Q[:, i]
        R[j, j] = np.linalg.norm(v)
        Q[:, j] = v / R[j, j]
    return Q, R

xd = np.array([0, 1, 2, 3, 4, 5], dtype=float)
yd = np.array([1.1, 2.9, 5.2, 7.1, 8.8, 11.2])
Amat = np.column_stack([np.ones_like(xd), xd])          # columns: 1, x

Q, R = gram_schmidt_qr(Amat)
print("Q^T Q = I ?", np.allclose(Q.T @ Q, np.eye(2)))
print("QR == A  ?", np.allclose(Q @ R, Amat))
print("R =\n", R.round(4))

c_normal = np.linalg.solve(Amat.T @ Amat, Amat.T @ yd)   # normal equations
c_qr     = linalg.solve_triangular(R, Q.T @ yd)           # QR:  R c = Q^T y
c_lstsq  = np.linalg.lstsq(Amat, yd, rcond=None)[0]        # library routine

print("\nNormal equations :", c_normal.round(6))
print("QR method        :", c_qr.round(6))
print("np.linalg.lstsq  :", c_lstsq.round(6))
res = yd - Amat @ c_qr
print(f"\nBest line: y = {c_qr[0]:.4f} + {c_qr[1]:.4f} x      ||residual|| = {np.linalg.norm(res):.4f}")
```

```python
xx = np.linspace(-0.5, 5.5, 100)
fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))

ax[0].plot(xd, yd, 'ro', label='data')
ax[0].plot(xx, c_qr[0] + c_qr[1] * xx, 'b-', label='least-squares line')
for xi, yi, fi in zip(xd, yd, Amat @ c_qr):
    ax[0].plot([xi, xi], [yi, fi], 'g--', lw=1)
ax[0].set_title('Least-squares line (dashed = residuals)'); ax[0].legend(); ax[0].grid(True)

ax[1].stem(xd, res)
ax[1].axhline(0, color='black', lw=0.8)
ax[1].set_title('Residuals'); ax[1].set_xlabel('x'); ax[1].grid(True)
plt.tight_layout()
plt.show()
```

# **4.6 Iterative methods: Jacobi and Gauss–Seidel**

Solve the strictly diagonally dominant system

$$\begin{pmatrix}10&-1&2&0\\-1&11&-1&3\\2&-1&10&-1\\0&3&-1&8\end{pmatrix}\mathbf x=\begin{pmatrix}6\\25\\-11\\15\end{pmatrix},\qquad \mathbf x^*=(1,\,2,\,-1,\,1)$$

and compare how fast the two methods converge.

```python
A4 = np.array([[10., -1.,  2.,  0.],
               [-1., 11., -1.,  3.],
               [ 2., -1., 10., -1.],
               [ 0.,  3., -1.,  8.]])
b4 = np.array([6., 25., -11., 15.])
x_true = np.linalg.solve(A4, b4)

def jacobi(A, b, tol=1e-10, max_iter=200):
    n = len(b); x = np.zeros(n); D = np.diag(A); R = A - np.diagflat(D); errors = []
    for k in range(max_iter):
        x_new = (b - R @ x) / D
        errors.append(np.linalg.norm(x_new - x_true, np.inf))
        done = np.linalg.norm(x_new - x, np.inf) < tol
        x = x_new
        if done: break
    return x, errors

def gauss_seidel(A, b, tol=1e-10, max_iter=200):
    n = len(b); x = np.zeros(n); errors = []
    for k in range(max_iter):
        x_old = x.copy()
        for i in range(n):
            x[i] = (b[i] - A[i, :i] @ x[:i] - A[i, i + 1:] @ x[i + 1:]) / A[i, i]
        errors.append(np.linalg.norm(x - x_true, np.inf))
        if np.linalg.norm(x - x_old, np.inf) < tol: break
    return x, errors

xj, ej = jacobi(A4, b4)
xg, eg = gauss_seidel(A4, b4)
print("Exact solution :", x_true.round(8))
print(f"Jacobi         : {xj.round(8)}   iterations = {len(ej)}")
print(f"Gauss-Seidel   : {xg.round(8)}   iterations = {len(eg)}")

# Spectral radius of the iteration matrices decides convergence (need rho < 1)
D = np.diagflat(np.diag(A4)); Lw = np.tril(A4, -1); Up = np.triu(A4, 1)
T_jacobi = -np.linalg.inv(D) @ (Lw + Up)
T_gs     = -np.linalg.inv(D + Lw) @ Up
rho = lambda T: max(abs(np.linalg.eigvals(T)))
print(f"\nSpectral radius  Jacobi = {rho(T_jacobi):.4f},  Gauss-Seidel = {rho(T_gs):.4f}")

plt.figure(figsize=(7, 4.5))
plt.semilogy(range(1, len(ej) + 1), ej, 'o-', label='Jacobi')
plt.semilogy(range(1, len(eg) + 1), eg, 's-', label='Gauss-Seidel')
plt.xlabel('iteration'); plt.ylabel('error  ||x_k - x*||∞'); plt.grid(True, which='both')
plt.title('Convergence of iterative methods'); plt.legend()
plt.show()
```

The two-equation systems were drawn as lines (intersecting, coincident, parallel) and the three-equation system as three planes meeting at $(2,3,-1)$. # **Eigenvalues, Eigenvectors and Singular Value Decomposition (SVD) and Their Applications**

# **Aim:**

To compute eigenvalues, eigenvectors and the Singular Value Decomposition (SVD) of matrices using Python and to visualise their applications — diagonalisation, Markov chains, image compression and principal component analysis.

# **Software required:**

Python 3, NumPy, Matplotlib

# **Theory:**

**Eigenvalues and eigenvectors.** A non-zero vector $\mathbf v$ is an eigenvector of a square matrix $A$ with eigenvalue $\lambda$ if $A\mathbf v=\lambda\mathbf v$. The eigenvalues are the roots of $\det(A-\lambda I)=0$; $\operatorname{trace}(A)=\sum\lambda_i$ and $\det(A)=\prod\lambda_i$.

**Diagonalisation.** If $A$ has $n$ independent eigenvectors then $A=PDP^{-1}$ with $D=\operatorname{diag}(\lambda_i)$ and $P=[\mathbf v_1\cdots\mathbf v_n]$; hence $A^k=PD^kP^{-1}$. A *symmetric* matrix has real eigenvalues and orthonormal eigenvectors: $A=Q\Lambda Q^T$.

**Singular Value Decomposition.** Every $m\times n$ matrix can be written

$$A=U\Sigma V^T,$$

where $U$ ($m\times m$) and $V$ ($n\times n$) are orthogonal and $\Sigma$ is diagonal with singular values $\sigma_1\ge\sigma_2\ge\dots\ge0$. The $\sigma_i$ are the square roots of the eigenvalues of $A^TA$. Geometrically, $A$ rotates ($V^T$), stretches ($\Sigma$) and rotates ($U$) again.

**Low-rank approximation (Eckart–Young).** Keeping the $k$ largest singular values gives $A_k=\sum_{i=1}^{k}\sigma_i\mathbf u_i\mathbf v_i^T$, the best rank-$k$ approximation, with $\|A-A_k\|_F=\sqrt{\sigma_{k+1}^2+\dots+\sigma_r^2}$.

# **Procedure:**

* Compute eigenvalues/eigenvectors, verify $A\mathbf v=\lambda\mathbf v$ and diagonalise.
* Visualise how a matrix acts on the unit circle; apply eigenvectors to a Markov chain.
* Compute the SVD, verify the factorisation and visualise it geometrically.
* Compress an image by low-rank approximation and perform PCA using the SVD.

# **Part 1 – Eigenvalues and eigenvectors**

# **5.1 Computing eigenvalues and eigenvectors**

For $A=\begin{pmatrix}4&1\\2&3\end{pmatrix}$ the characteristic equation $\lambda^2-7\lambda+10=0$ gives $\lambda=5$ and $\lambda=2$.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

A = np.array([[4., 1.],
              [2., 3.]])

eigvals, eigvecs = np.linalg.eig(A)
print("Eigenvalues :", eigvals)
print("Eigenvectors (columns):\n", eigvecs.round(4))

for i in range(2):
    v = eigvecs[:, i]
    print(f"\nlambda_{i+1} = {eigvals[i]:.4f}")
    print("  A v       =", (A @ v).round(4))
    print("  lambda v  =", (eigvals[i] * v).round(4))
    print("  A v = lambda v ?", np.allclose(A @ v, eigvals[i] * v))

print("\ntrace(A) =", np.trace(A), "   sum of eigenvalues     =", round(eigvals.sum(), 6))
print("det(A)   =", round(np.linalg.det(A), 6), "   product of eigenvalues =", round(eigvals.prod(), 6))
```

# **5.2 Diagonalisation and matrix powers**

With $P$ the matrix of eigenvectors and $D$ the diagonal matrix of eigenvalues, $A=PDP^{-1}$ and $A^{k}=PD^{k}P^{-1}$ — raising a matrix to a power only needs the eigenvalues to be raised to that power.

```python
P = eigvecs
D = np.diag(eigvals)
A_rebuilt = P @ D @ np.linalg.inv(P)
print("P D P^-1 =\n", A_rebuilt.round(8))
print("equals A ?", np.allclose(A, A_rebuilt))

k = 5
A_k = P @ np.diag(eigvals ** k) @ np.linalg.inv(P)
print(f"\nA^{k} via diagonalisation =\n", A_k.round(4))
print(f"A^{k} via matrix_power    =\n", np.linalg.matrix_power(A, k))

# Symmetric matrix: orthogonal diagonalisation  S = Q Λ Q^T
S = np.array([[2., 1.],
              [1., 2.]])
w, Q = np.linalg.eigh(S)
print("\nSymmetric S: eigenvalues =", w, "\nQ =\n", Q.round(4))
print("Q^T Q = I ?", np.allclose(Q.T @ Q, np.eye(2)), "   S = Q Λ Q^T ?", np.allclose(Q @ np.diag(w) @ Q.T, S))
```

# **5.3 Visualising eigenvectors**

$A$ maps the unit circle to an ellipse. Most vectors change direction, but eigenvectors stay on their own line — they are only *stretched* by $\lambda$. The solid arrows are the eigenvectors and the faint arrows are $A\mathbf v=\lambda\mathbf v$.

```python
theta = np.linspace(0, 2 * np.pi, 400)
circle = np.vstack([np.cos(theta), np.sin(theta)])
ellipse = A @ circle

fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(*circle, color='tab:blue', label='unit circle')
ax.plot(*ellipse, color='tab:red', label='A × unit circle')

colors = ['green', 'purple']
for i in range(2):
    v = eigvecs[:, i]
    ax.quiver(0, 0, *v, angles='xy', scale_units='xy', scale=1, color=colors[i], width=0.007,
              label=f'eigenvector, λ = {eigvals[i]:.0f}')
    ax.quiver(0, 0, *(A @ v), angles='xy', scale_units='xy', scale=1, color=colors[i], alpha=0.4, width=0.007)
    ax.plot([-6 * v[0], 6 * v[0]], [-6 * v[1], 6 * v[1]], '--', color=colors[i], lw=0.8)

ax.set_xlim(-6, 6); ax.set_ylim(-6, 6); ax.set_aspect('equal'); ax.grid(True)
ax.axhline(0, color='black', lw=0.6); ax.axvline(0, color='black', lw=0.6)
ax.legend(loc='upper left'); ax.set_title('Action of A on the unit circle')
plt.show()
```

# **5.4 Application – a Markov chain**

A web surfer is either on page 1 or page 2. Each step she stays on page 1 with probability $0.9$ and stays on page 2 with probability $0.5$:

$$P=\begin{pmatrix}0.9&0.1\\0.5&0.5\end{pmatrix}.$$

The long-run (steady-state) distribution $\boldsymbol\pi$ satisfies $\boldsymbol\pi P=\boldsymbol\pi$, i.e. it is the eigenvector of $P^T$ for the eigenvalue $\lambda=1$ (normalised so its entries sum to 1). The other eigenvalue ($0.4$) controls *how fast* the chain converges.

```python
P_m = np.array([[0.9, 0.1],
                [0.5, 0.5]])

w, V = np.linalg.eig(P_m.T)
idx = np.argmin(np.abs(w - 1))
pi = np.real(V[:, idx]); pi = pi / pi.sum()
print("Eigenvalues of P :", w.round(4))
print("Steady-state distribution:", pi.round(4), " (exact: 5/6, 1/6 =", round(5/6, 4), round(1/6, 4), ")")

state = np.array([0.0, 1.0])            # start on page 2
history = [state]
for step in range(15):
    state = state @ P_m
    history.append(state)
history = np.array(history)

plt.figure(figsize=(7, 4.5))
plt.plot(history[:, 0], 'o-', label='P(page 1)')
plt.plot(history[:, 1], 's-', label='P(page 2)')
plt.axhline(pi[0], color='tab:blue', ls=':'); plt.axhline(pi[1], color='tab:orange', ls=':')
plt.xlabel('step'); plt.ylabel('probability'); plt.grid(True); plt.legend()
plt.title('Convergence to the steady state')
plt.show()
```

# **5.5 Computing the SVD**

For $M=\begin{pmatrix}3&1&1\\-1&3&1\end{pmatrix}$ the singular values are $\sqrt{12}\approx3.4641$ and $\sqrt{10}\approx3.1623$.

```python
M = np.array([[ 3., 1., 1.],
              [-1., 3., 1.]])

U, s, Vt = np.linalg.svd(M)
Sigma = np.zeros(M.shape); np.fill_diagonal(Sigma, s)

print("U =\n", U.round(4))
print("Singular values:", s.round(4))
print("V^T =\n", Vt.round(4))
print("\nU Σ V^T == M ?", np.allclose(U @ Sigma @ Vt, M))
print("sqrt(eigenvalues of M M^T):", np.sqrt(np.sort(np.linalg.eigvalsh(M @ M.T))[::-1]).round(4))
print(f"rank = {np.linalg.matrix_rank(M)},   ||M||_2 = σ1 = {s[0]:.4f},   condition number σ1/σ_min = {s[0]/s[-1]:.4f}")
```

# **5.6 Geometry of the SVD**

For a $2\times2$ matrix, $A\mathbf x=U\Sigma V^T\mathbf x$ is a sequence of three simple steps applied to the unit circle: rotate/reflect by $V^T$, stretch the axes by $\sigma_1,\sigma_2$, and rotate/reflect by $U$.

```python
A2 = np.array([[2.0, 1.0],
               [0.0, 1.5]])
U2, s2, Vt2 = np.linalg.svd(A2)

theta = np.linspace(0, 2 * np.pi, 300)
circle = np.vstack([np.cos(theta), np.sin(theta)])
step1 = Vt2 @ circle                 # rotate / reflect
step2 = np.diag(s2) @ step1          # stretch along the axes
step3 = U2 @ step2                   # rotate / reflect  -> this is A2 @ circle

titles = ['Unit circle', r'after $V^T$', r'after $\Sigma$', r'after $U$  (= $A$ x circle)']
fig, ax = plt.subplots(1, 4, figsize=(16, 4.2))
for axis, pts, title in zip(ax, (circle, step1, step2, step3), titles):
    axis.plot(*pts, color='tab:red')
    axis.plot(*pts[:, :1], 'ko')                          # marks the starting point
    axis.set_xlim(-3, 3); axis.set_ylim(-3, 3); axis.set_aspect('equal'); axis.grid(True); axis.set_title(title)
plt.tight_layout()
plt.show()
print("Singular values (semi-axes of the final ellipse):", s2.round(4))
print("A x circle equals U Σ V^T x circle ?", np.allclose(step3, A2 @ circle))
```

# **5.7 Application 1 – Image compression by low-rank approximation**

A grey-scale image is a matrix. Keeping only the $k$ largest singular values stores $k(m+n+1)$ numbers instead of $mn$. We create a synthetic $128\times128$ image (a gradient with a disc and a checker-board) so that no file is needed.

```python
rng_img = np.random.default_rng(7)
yy, xx = np.mgrid[0:128, 0:128]
img = 0.5 * (xx / 127) + 0.3 * np.sin(xx / 9) * np.cos(yy / 13)
img += ((xx - 64) ** 2 + (yy - 64) ** 2 < 30 ** 2) * 0.6          # a bright disc
img += ((xx // 16 + yy // 16) % 2) * 0.15                         # checker-board
img += 0.08 * np.sin((xx + yy) / 4)                               # diagonal stripes
img += rng_img.normal(0, 0.05, img.shape)                         # noise / fine detail
img = (img - img.min()) / (img.max() - img.min())

Ui, si, Vti = np.linalg.svd(img, full_matrices=False)

def rank_k(k):
    return (Ui[:, :k] * si[:k]) @ Vti[:k]

m, n = img.shape
ks = [1, 5, 15, 40]
fig, ax = plt.subplots(1, 5, figsize=(18, 4))
ax[0].imshow(img, cmap='gray'); ax[0].set_title('Original (rank %d)' % np.linalg.matrix_rank(img))
for axis, k in zip(ax[1:], ks):
    approx = rank_k(k)
    err = np.linalg.norm(img - approx) / np.linalg.norm(img)
    storage = k * (m + n + 1) / (m * n) * 100
    axis.imshow(approx, cmap='gray')
    axis.set_title(f'k = {k}\nerror {err*100:.2f}%, storage {storage:.1f}%')
for axis in ax: axis.axis('off')
plt.tight_layout()
plt.show()
```

```python
# relative error of the rank-k approximation:  sqrt(sum of discarded sigma_i^2) / sqrt(sum of all sigma_i^2)
rel_err = np.sqrt(np.maximum(0, 1 - np.cumsum(si ** 2) / np.sum(si ** 2)))
k5 = int(np.argmax(rel_err < 0.05) + 1)
k1 = int(np.argmax(rel_err < 0.01) + 1)
print(f"Smallest k with error below 5%: {k5}  (storage {k5*(m+n+1)/(m*n)*100:.1f}% of the original)")
print(f"Smallest k with error below 1%: {k1}  (storage {k1*(m+n+1)/(m*n)*100:.1f}% of the original)")

# Eckart-Young check:  ||A - A_k||_F = sqrt(sum of the discarded sigma_i^2)
k = 10
print(f"\n||img - A_{k}||_F             = {np.linalg.norm(img - rank_k(k)):.6f}")
print(f"sqrt(sum of sigma_i^2, i>{k}) = {np.sqrt(np.sum(si[k:] ** 2)):.6f}")

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].semilogy(si, 'o-', ms=3); ax[0].set_title('Singular values of the image'); ax[0].set_xlabel('index'); ax[0].grid(True)
ax[1].semilogy(range(1, len(si) + 1), rel_err * 100)
ax[1].axhline(5, color='gray', ls=':'); ax[1].axhline(1, color='gray', ls=':')
ax[1].set_ylim(0.1, 100)
ax[1].set_title('Relative error (%) of the rank-k approximation'); ax[1].set_xlabel('k'); ax[1].grid(True, which='both')
plt.tight_layout()
plt.show()
```

# **5.8 Application 2 – Principal Component Analysis (PCA) via SVD**

Centre the data matrix $X$ ($n$ samples) and compute $X=U\Sigma V^T$. The rows of $V^T$ are the **principal directions** and the variance along them is $\sigma_i^2/(n-1)$. Below, 300 correlated 2-D points are generated and their principal axes are drawn.

```python
rng = np.random.default_rng(3)
cov = np.array([[3.0, 1.8],
                [1.8, 1.5]])
X = rng.multivariate_normal([2, 1], cov, size=300)

Xc = X - X.mean(axis=0)
Up, sp_, Vtp = np.linalg.svd(Xc, full_matrices=False)
variances = sp_ ** 2 / (len(X) - 1)
ratio = variances / variances.sum()

print("Principal directions (rows):\n", Vtp.round(4))
print("Variance along each direction:", variances.round(4))
print("Explained variance ratio     :", ratio.round(4))
print("Same as eigenvalues of the sample covariance matrix:", np.sort(np.linalg.eigvalsh(np.cov(X.T)))[::-1].round(4))

fig, ax = plt.subplots(figsize=(6.5, 6))
ax.scatter(X[:, 0], X[:, 1], s=10, alpha=0.5)
mean = X.mean(axis=0)
for i, color in enumerate(['red', 'green']):
    d = Vtp[i] * 2 * np.sqrt(variances[i])
    ax.quiver(*mean, *d, angles='xy', scale_units='xy', scale=1, color=color, width=0.009,
              label=f'PC{i+1} ({ratio[i]*100:.1f}% of variance)')
ax.set_aspect('equal'); ax.grid(True); ax.legend(); ax.set_title('Principal components of the data')
plt.show()
```

# **Result**

* Eigenvalues of $A$ were $5$ and $2$ with eigenvectors along $(1,1)$ and $(1,-2)$; $A\mathbf v=\lambda\mathbf v$, $\operatorname{tr}A=\sum\lambda_i$ and $\det A=\prod\lambda_i$ were verified, and $A=PDP^{-1}$ was used to compute $A^5$.
* The steady state of the Markov chain was $(5/6,\,1/6)$.
* The SVD of $M$ gave singular values $\sqrt{12}$ and $\sqrt{10}$ and reproduced $M=U\Sigma V^T$.
* A rank-$k$ SVD approximation compressed an image (error below 5% with $k=18$), and PCA recovered the principal axes of correlated data (first component explained about 94% of the variance).

# **Discussion**

Eigenvectors are the directions that a matrix leaves unchanged apart from scaling; eigenvalues give the scaling and the long-run behaviour of repeated multiplication (here, the convergence of the Markov chain is governed by the second eigenvalue $0.4$). The SVD exists for *every* matrix and is the basis of image compression, noise reduction, PCA and recommender systems: for the test image about 18 singular values (28% of the storage) reproduce it with a 5% error, whereas the random noise component can only be captured by using almost all singular values — so noisy data compress poorly. The Eckart–Young identity was confirmed numerically.

# **Implementing the Gradient Descent Algorithm to Solve Optimisation Problems and Visualising Convergence**

# **Aim:**

To implement the gradient descent algorithm (and its momentum and stochastic variants) in Python, to study the effect of the step size (learning rate) and to visualise convergence.

# **Software required:**

Python 3, NumPy, Matplotlib, Pandas

# **Theory:**

To minimise a differentiable function $f:\mathbb R^n\to\mathbb R$, gradient descent repeatedly moves in the direction of steepest descent, the negative gradient:

$$\mathbf x_{k+1}=\mathbf x_k-\alpha\,\nabla f(\mathbf x_k)$$

where $\alpha>0$ is the **step size** (learning rate). The algorithm stops when $\|\nabla f(\mathbf x_k)\|<\varepsilon$.

* If $\alpha$ is too small, convergence is slow; if it is too large, the iterates oscillate or diverge.
* For a quadratic $f(\mathbf x)=\tfrac12\mathbf x^TQ\mathbf x$ with largest eigenvalue $L$ of $Q$, the method converges iff $0<\alpha<2/L$.
* The speed depends on the **condition number** $\kappa=\lambda_{\max}/\lambda_{\min}$: long narrow valleys ($\kappa$ large) give slow, zig-zag convergence.

**Variants**

* *Momentum:* $\mathbf v_{k+1}=\beta\mathbf v_k+\nabla f(\mathbf x_k),\;\; \mathbf x_{k+1}=\mathbf x_k-\alpha\mathbf v_{k+1}$ — accumulates past gradients and damps zig-zag.
* *Stochastic / mini-batch gradient descent:* for a loss that is an average over data, $f=\frac1N\sum f_i$, the gradient is estimated from a single sample (SGD) or a small batch — cheaper per step but noisy.

# **Procedure:**

* Implement gradient descent for a one-variable function and study different step sizes.
* Implement a general $n$-dimensional gradient descent and apply it to $f(x,y)=x^2+10y^2$; plot the path on a contour plot and the convergence curve.
* Find the range of step sizes that converge; compare with momentum.
* Apply batch, mini-batch and stochastic gradient descent to linear regression.

# **6.1 One variable: effect of the step size**

Minimise $f(x)=(x-3)^2+2$ with $f'(x)=2(x-3)$. The iteration is $x_{k+1}-3=(1-2\alpha)(x_k-3)$, so the error is multiplied by $(1-2\alpha)$ at every step:

| $\alpha$ | factor $1-2\alpha$ | behaviour |
|---|---|---|
| 0.1 | 0.8 | slow, monotone convergence |
| 0.5 | 0 | exact in one step |
| 0.9 | −0.8 | oscillating convergence |
| 1.05 | −1.1 | oscillation grows → **divergence** |

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

f  = lambda x: (x - 3) ** 2 + 2
df = lambda x: 2 * (x - 3)

def gd_1d(x0, lr, n_iter=15):
    xs = [x0]
    for _ in range(n_iter):
        xs.append(xs[-1] - lr * df(xs[-1]))
    return np.array(xs)

rates = [0.1, 0.5, 0.9, 1.05]
xx = np.linspace(-3, 9, 300)

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].plot(xx, f(xx), 'k', lw=1)
for lr in rates:
    xs = gd_1d(0.0, lr)
    ax[0].plot(xs, f(xs), 'o-', ms=4, label=f'α = {lr}')
    ax[1].semilogy(np.abs(xs - 3) + 1e-16, 'o-', ms=4, label=f'α = {lr}')
ax[0].set_ylim(0, 40); ax[0].set_xlim(-3, 9)
ax[0].set_title('Iterates on the graph of f(x) = (x-3)² + 2'); ax[0].set_xlabel('x'); ax[0].set_ylabel('f(x)'); ax[0].legend(); ax[0].grid(True)
ax[1].set_title('Distance from the minimiser |x_k - 3|'); ax[1].set_xlabel('iteration'); ax[1].legend(); ax[1].grid(True, which='both')
plt.tight_layout()
plt.show()
```

# **6.2 Two variables: path on a contour plot**

Take the "narrow valley" $f(x,y)=x^2+10y^2$ with gradient $\nabla f=(2x,\,20y)$. The Hessian eigenvalues are $2$ and $20$, so $L=20$, $\kappa=10$ and gradient descent converges iff $\alpha<2/20=0.1$.

The function `gradient_descent` below is general: pass any gradient function.

```python
def gradient_descent(grad, x0, lr, max_iter=1000, tol=1e-6):
    """Return the path of iterates and a status message."""
    x = np.array(x0, dtype=float)
    path = [x.copy()]
    status = "max_iter reached"
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            status = "converged"
            break
        x = x - lr * g
        path.append(x.copy())
        if np.linalg.norm(x) > 1e8:
            status = "diverged"
            break
    return np.array(path), status

f2    = lambda p: p[0] ** 2 + 10 * p[1] ** 2
grad2 = lambda p: np.array([2 * p[0], 20 * p[1]])
start = (8.0, 2.0)

X, Y = np.meshgrid(np.linspace(-9, 9, 300), np.linspace(-3, 3, 300))
Z = X ** 2 + 10 * Y ** 2

fig, ax = plt.subplots(1, 2, figsize=(15, 5.5))
ax[0].contour(X, Y, Z, levels=25, cmap='viridis')
for lr, color in zip([0.01, 0.05, 0.09], ['tab:blue', 'tab:green', 'tab:red']):
    path, status = gradient_descent(grad2, start, lr, max_iter=60)
    ax[0].plot(path[:, 0], path[:, 1], 'o-', ms=3, color=color, label=f'α = {lr}')
    ax[1].semilogy([f2(p) for p in path], color=color, label=f'α = {lr}')
ax[0].plot(0, 0, 'k*', ms=14)
ax[0].set_title('Gradient descent paths on f(x, y) = x² + 10y²'); ax[0].set_xlabel('x'); ax[0].set_ylabel('y'); ax[0].legend()
ax[1].set_title('Convergence: f(x_k) versus iteration'); ax[1].set_xlabel('iteration'); ax[1].set_ylabel('f(x_k)'); ax[1].legend(); ax[1].grid(True, which='both')
plt.tight_layout()
plt.show()
```

# **6.3 Which step sizes converge?**

For each $\alpha$ count the iterations needed to reach $\|\nabla f\|<10^{-6}$ (maximum 2000). The theory predicts convergence for $\alpha<0.1$; the fastest rate is obtained when the two error factors $|1-2\alpha|$ and $|1-20\alpha|$ are equal, i.e. $\alpha=2/(2+20)\approx0.091$, and there is no convergence at $\alpha=0.1$ (the $y$-error just flips sign) or beyond.

```python
rows = []
for lr in [0.005, 0.01, 0.02, 0.05, 0.09, 0.099, 0.10, 0.11]:
    path, status = gradient_descent(grad2, start, lr, max_iter=2000)
    rows.append({"step size α": lr, "iterations": len(path) - 1, "status": status,
                 "final f": f"{f2(path[-1]):.2e}" if status != "diverged" else "—"})
display(pd.DataFrame(rows))
```

# **6.4 Momentum**

Momentum accumulates a velocity and damps the zig-zag across a narrow valley. Compare plain gradient descent with momentum ($\beta=0.9$) at the same small step size $\alpha=0.01$.

```python
def gd_momentum(grad, x0, lr, beta=0.9, max_iter=2000, tol=1e-6):
    x = np.array(x0, dtype=float); v = np.zeros_like(x)
    path = [x.copy()]
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            break
        v = beta * v + g
        x = x - lr * v
        path.append(x.copy())
    return np.array(path)

p_plain, st_plain = gradient_descent(grad2, start, 0.01, max_iter=3000)
p_mom = gd_momentum(grad2, start, 0.01, beta=0.9, max_iter=3000)
print(f"Plain gradient descent : {len(p_plain) - 1} iterations ({st_plain})")
print(f"Momentum (beta = 0.9)  : {len(p_mom) - 1} iterations")

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].contour(X, Y, Z, levels=25, cmap='viridis', alpha=0.6)
ax[0].plot(p_plain[:200, 0], p_plain[:200, 1], '.-', ms=3, label='plain GD (first 200 steps)')
ax[0].plot(p_mom[:200, 0], p_mom[:200, 1], '.-', ms=3, label='momentum (first 200 steps)')
ax[0].plot(0, 0, 'k*', ms=14); ax[0].legend(); ax[0].set_title('Paths, α = 0.01')
ax[1].semilogy([f2(p) for p in p_plain[:400]], label='plain GD')
ax[1].semilogy([f2(p) for p in p_mom[:400]], label='momentum')
ax[1].set_xlabel('iteration'); ax[1].set_ylabel('f(x_k)'); ax[1].grid(True, which='both'); ax[1].legend()
ax[1].set_title('Convergence')
plt.tight_layout()
plt.show()
```

# **6.5 Application – linear regression by gradient descent**

Fit $y=wx+b$ to 200 noisy points (true values $w=3,\;b=2$) by minimising the mean squared error

$$J(w,b)=\frac1N\sum_{i=1}^{N}(wx_i+b-y_i)^2,\qquad \nabla J=\frac{2}{N}X^T(X\boldsymbol\theta-\mathbf y),\quad \boldsymbol\theta=(w,b)^T .$$

Three versions: **batch** (all 200 samples per update), **mini-batch** (20 samples) and **stochastic** (1 sample). The closed-form least-squares solution (Experiment 4) is used as the reference.

```python
rng = np.random.default_rng(1)
N = 200
x_data = rng.random(N)
y_data = 3 * x_data + 2 + rng.normal(0, 0.2, N)
Xd = np.column_stack([x_data, np.ones(N)])

def mse(theta):
    return np.mean((Xd @ theta - y_data) ** 2)

def train(batch_size, lr, epochs, seed=0):
    r = np.random.default_rng(seed)
    theta = np.zeros(2); history = [mse(theta)]
    for epoch in range(epochs):
        order = r.permutation(N)
        for s in range(0, N, batch_size):
            idx = order[s:s + batch_size]
            grad = 2 * Xd[idx].T @ (Xd[idx] @ theta - y_data[idx]) / len(idx)
            theta = theta - lr * grad
        history.append(mse(theta))
    return theta, np.array(history)

theta_batch, h_batch = train(N,  0.5,  300)
theta_mini,  h_mini  = train(20, 0.2,  100)
theta_sgd,   h_sgd   = train(1,  0.05, 100)
theta_cf = np.linalg.lstsq(Xd, y_data, rcond=None)[0]

display(pd.DataFrame({
    "slope w":   [theta_batch[0], theta_mini[0], theta_sgd[0], theta_cf[0]],
    "intercept b": [theta_batch[1], theta_mini[1], theta_sgd[1], theta_cf[1]],
    "final MSE": [h_batch[-1], h_mini[-1], h_sgd[-1], mse(theta_cf)],
}, index=["Batch GD (300 epochs)", "Mini-batch GD (100 epochs)", "SGD (100 epochs)", "Closed form"]).round(5))

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
ax[0].scatter(x_data, y_data, s=10, alpha=0.5, label='data')
xl = np.array([0, 1])
ax[0].plot(xl, theta_cf[0] * xl + theta_cf[1], 'k-', lw=2, label='closed form')
ax[0].plot(xl, theta_batch[0] * xl + theta_batch[1], 'r--', lw=2, label='batch GD')
ax[0].set_title('Fitted line'); ax[0].legend(); ax[0].grid(True)

floor = mse(theta_cf)
ax[1].semilogy(h_batch - floor + 1e-12, label='batch (lr 0.5)')
ax[1].semilogy(h_mini - floor + 1e-12, label='mini-batch 20 (lr 0.2)')
ax[1].semilogy(h_sgd - floor + 1e-12, label='SGD (lr 0.05)')
ax[1].set_xlabel('epoch'); ax[1].set_ylabel('MSE − minimum MSE'); ax[1].set_title('Convergence of the loss')
ax[1].grid(True, which='both'); ax[1].legend()
plt.tight_layout()
plt.show()
```

# **Result**

Gradient descent was implemented for one and two variables and for linear regression. For $f=(x-3)^2+2$ the step sizes $0.1,\,0.5,\,0.9,\,1.05$ gave slow convergence, one-step convergence, oscillating convergence and divergence respectively, exactly as predicted by the factor $1-2\alpha$. For $f=x^2+10y^2$ the method converged for $\alpha<0.1$ and failed at $\alpha\ge0.1$; momentum reduced the iterations needed at $\alpha=0.01$ from 822 to 317. Batch, mini-batch and stochastic gradient descent all recovered a line close to the true $w=3,\;b=2$.

# **Discussion**

The step size must be matched to the curvature of the function: $\alpha<2/L$ is necessary for a quadratic. In a narrow valley the gradient points mostly across the valley, so plain gradient descent zig-zags and the slow direction (small eigenvalue) dominates the number of iterations; momentum averages out the zig-zag. SGD makes cheap, noisy updates and therefore stalls at a "noise floor" around the optimum, while batch gradient descent converges smoothly but needs the whole data set for every update — the central trade-off in machine-learning training.

# **Solving Optimization Problems Using Newton's Method and Visualizing Results**

# **Aim:**

To solve unconstrained optimisation problems with Newton's method, to compare it with gradient descent and to visualise the iterations.

# **Software required:**

Python 3, NumPy, SymPy, Matplotlib, Pandas

# **Theory:**

Newton's method approximates $f$ near $\mathbf x_k$ by its second-order Taylor polynomial and jumps to the minimiser of that quadratic model.

* **One variable:** $\;x_{k+1}=x_k-\dfrac{f'(x_k)}{f''(x_k)}$  (this is Newton's root-finding method applied to $f'$)
* **Several variables:** $\;\mathbf x_{k+1}=\mathbf x_k-H_f(\mathbf x_k)^{-1}\nabla f(\mathbf x_k)$, where $H_f$ is the Hessian matrix. In practice the linear system $H\,\mathbf s=\nabla f$ is solved instead of inverting $H$.

**Properties**

* Near a minimiser where $f''>0$ (or $H$ positive definite) the convergence is **quadratic**: the number of correct digits roughly doubles each iteration.
* For a quadratic function Newton's method converges in **one step**.
* It needs the Hessian (expensive in high dimension) and may fail when $f''=0$, when $H$ is not positive definite (it can head to a maximum or saddle point), or when started far from the minimiser.

# **Procedure:**

* Use SymPy to obtain $f'$ and $f''$ and run Newton's method for a one-variable convex function; verify quadratic convergence.
* Show a case where Newton's method does not find a minimum.
* Apply the multivariable method to the Rosenbrock function and compare with gradient descent.

# **7.1 One variable: minimise $f(x)=e^x-2x$**

$f'(x)=e^x-2$ and $f''(x)=e^x>0$, so $f$ is convex with its minimum at $x^*=\ln2\approx0.693147$. SymPy computes the derivatives and `lambdify` turns them into fast numerical functions.

```python
import numpy as np
import pandas as pd
import sympy as sp
import matplotlib.pyplot as plt

xs_ = sp.symbols('x')
f_sym   = sp.exp(xs_) - 2 * xs_
df_sym  = sp.diff(f_sym, xs_)
d2f_sym = sp.diff(df_sym, xs_)
print("f(x)   =", f_sym, "\nf'(x)  =", df_sym, "\nf''(x) =", d2f_sym)

f   = sp.lambdify(xs_, f_sym,   'numpy')
df  = sp.lambdify(xs_, df_sym,  'numpy')
d2f = sp.lambdify(xs_, d2f_sym, 'numpy')

def newton_1d(df, d2f, x0, tol=1e-12, max_iter=60):
    xs = [x0]
    for _ in range(max_iter):
        xs.append(xs[-1] - df(xs[-1]) / d2f(xs[-1]))
        if abs(xs[-1] - xs[-2]) < tol:
            break
    return np.array(xs)

path = newton_1d(df, d2f, x0=3.0)
x_star = np.log(2)
table = pd.DataFrame({"k": range(len(path)), "x_k": path, "f'(x_k)": df(path), "error |x_k - ln 2|": np.abs(path - x_star)})
pd.options.display.float_format = '{:.3e}'.format
display(table.set_index("k"))
pd.reset_option('display.float_format')
print(f"\nNewton converged to x = {path[-1]:.12f} in {len(path) - 1} iterations (ln 2 = {x_star:.12f})")
```

Look at the error column: each error is roughly the *square* of the previous one (here $1.4\times10^{-2}\to9.9\times10^{-5}\to4.9\times10^{-9}$) — this is **quadratic convergence**. The picture shows the geometry: each step follows the tangent of $f'$ down to the $x$-axis.

```python
xx = np.linspace(0, 3.2, 300)
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].plot(xx, f(xx), 'k')
ax[0].plot(path[:6], f(path[:6]), 'ro-', ms=5)
ax[0].plot(x_star, f(x_star), 'g*', ms=15, label='minimum  x* = ln 2')
ax[0].set_title('f(x) = e^x - 2x and the Newton iterates'); ax[0].set_xlabel('x'); ax[0].grid(True); ax[0].legend()

ax[1].plot(xx, df(xx), 'k', label="f'(x)")
ax[1].axhline(0, color='gray')
for k in range(4):
    ax[1].plot([path[k], path[k]], [0, df(path[k])], 'b:', lw=1)
    ax[1].plot([path[k], path[k + 1]], [df(path[k]), 0], 'r--', lw=1.5)
    ax[1].plot(path[k], df(path[k]), 'ro')
    ax[1].annotate(f'$x_{k}$', (path[k], 0), textcoords='offset points', xytext=(0, -16), ha='center')
ax[1].set_title("Newton step = tangent line of f' hitting the x-axis"); ax[1].grid(True); ax[1].legend()
plt.tight_layout()
plt.show()
```

# **7.2 When Newton's method fails to find a minimum**

Take $g(x)=x^4-3x^3+2$. Then $g'(x)=x^2(4x-9)$ has stationary points $x=2.25$ (a minimum, $g''>0$) and $x=0$ (an inflection point, $g''(0)=0$). Newton's method only looks for $g'(x)=0$ and does not know which kind of point it reaches: the starting point decides.

```python
g_sym = xs_ ** 4 - 3 * xs_ ** 3 + 2
dg  = sp.lambdify(xs_, sp.diff(g_sym, xs_), 'numpy')
d2g = sp.lambdify(xs_, sp.diff(g_sym, xs_, 2), 'numpy')
g   = sp.lambdify(xs_, g_sym, 'numpy')

for x0 in (3.0, 0.5):
    p = newton_1d(dg, d2g, x0)
    kind = "minimum" if d2g(p[-1]) > 1e-6 else "NOT a minimum (g'' ≈ 0: inflection point)"
    print(f"x0 = {x0}:  stopped at x = {p[-1]:.6f} after {len(p) - 1} iterations,  g'' = {d2g(p[-1]):.2e}  -> {kind}")

xx = np.linspace(-1, 3.5, 300)
plt.figure(figsize=(7, 4.5))
plt.plot(xx, g(xx), 'k')
for x0, color in ((3.0, 'tab:green'), (0.5, 'tab:red')):
    p = newton_1d(dg, d2g, x0)
    plt.plot(p, g(p), 'o-', color=color, ms=4, label=f'start {x0}')
plt.title("g(x) = x⁴ − 3x³ + 2"); plt.xlabel('x'); plt.grid(True); plt.legend()
plt.show()
```

# **7.3 Several variables: the Rosenbrock function**

$$f(x,y)=(1-x)^2+100\,(y-x^2)^2$$

has its minimum at $(1,1)$ inside a long, curved, narrow valley — a classic test for optimisation algorithms. SymPy provides the gradient and the Hessian exactly. Both algorithms start from $(-1.2,\,1)$.

```python
x1, x2 = sp.symbols('x1 x2')
fr = (1 - x1) ** 2 + 100 * (x2 - x1 ** 2) ** 2
grad_sym = sp.Matrix([sp.diff(fr, v) for v in (x1, x2)])
hess_sym = sp.hessian(fr, (x1, x2))
print("Gradient:", list(grad_sym))
print("Hessian :", hess_sym.tolist())

_f = sp.lambdify((x1, x2), fr, 'numpy')
_g = sp.lambdify((x1, x2), grad_sym, 'numpy')
_h = sp.lambdify((x1, x2), hess_sym, 'numpy')
F = lambda p: float(_f(*p))
G = lambda p: np.array(_g(*p), dtype=float).ravel()
H = lambda p: np.array(_h(*p), dtype=float)

def newton_nd(grad, hess, x0, tol=1e-10, max_iter=50):
    x = np.array(x0, dtype=float); path = [x.copy()]
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            break
        step = np.linalg.solve(hess(x), g)       # solve H s = grad f   (no explicit inverse)
        x = x - step
        path.append(x.copy())
    return np.array(path)

def gradient_descent(grad, x0, lr, max_iter=1000, tol=1e-6):
    x = np.array(x0, dtype=float); path = [x.copy()]
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol: break
        x = x - lr * g; path.append(x.copy())
    return np.array(path)

x0 = (-1.2, 1.0)
p_newton = newton_nd(G, H, x0)
p_gd = gradient_descent(G, x0, lr=5e-4, max_iter=5000)

print("\nNewton iterates:")
for k, p in enumerate(p_newton):
    print(f"  k={k}:  ({p[0]:+.6f}, {p[1]:+.6f})   f = {F(p):.3e}   ||grad|| = {np.linalg.norm(G(p)):.3e}")
print(f"\nGradient descent after {len(p_gd) - 1} iterations: ({p_gd[-1][0]:.4f}, {p_gd[-1][1]:.4f}),  f = {F(p_gd[-1]):.2e}")
```

```python
X, Y = np.meshgrid(np.linspace(-1.8, 1.8, 400), np.linspace(-1, 3.2, 400))
Z = _f(X, Y)

fig, ax = plt.subplots(1, 2, figsize=(15, 5.5))
ax[0].contour(X, Y, Z, levels=np.logspace(-1, 3.5, 25), cmap='viridis')
ax[0].plot(p_gd[:, 0], p_gd[:, 1], 'b-', lw=1, label=f'gradient descent ({len(p_gd) - 1} iterations, not yet converged)')
ax[0].plot(p_newton[:, 0], p_newton[:, 1], 'ro-', ms=5, label=f'Newton ({len(p_newton) - 1} iterations)')
ax[0].plot(1, 1, 'k*', ms=15, label='minimum (1, 1)')
ax[0].set_title('Rosenbrock function: optimisation paths'); ax[0].set_xlabel('x'); ax[0].set_ylabel('y'); ax[0].legend(fontsize=8)

ax[1].semilogy([np.linalg.norm(G(p)) + 1e-18 for p in p_newton], 'ro-', label='Newton')
ax[1].semilogy([np.linalg.norm(G(p)) for p in p_gd], 'b-', label='gradient descent')
ax[1].set_xscale('symlog', linthresh=10)
ax[1].set_xlabel('iteration'); ax[1].set_ylabel('||∇f(x_k)||'); ax[1].set_title('Gradient norm versus iteration')
ax[1].grid(True, which='both'); ax[1].legend()
plt.tight_layout()
plt.show()
```

# **7.4 Newton versus gradient descent on a quadratic**

For $f(x,y)=x^2+10y^2$ the Hessian is constant, $H=\operatorname{diag}(2,20)$, so Newton's quadratic model is *exact* and one step reaches the minimum, whereas gradient descent (best safe step size close to $\alpha=0.09$) needs many iterations.

```python
grad_q = lambda p: np.array([2 * p[0], 20 * p[1]])
hess_q = lambda p: np.array([[2., 0.], [0., 20.]])
f_q    = lambda p: p[0] ** 2 + 10 * p[1] ** 2

pn = newton_nd(grad_q, hess_q, (8.0, 2.0))
pg = gradient_descent(grad_q, (8.0, 2.0), lr=0.09, max_iter=5000, tol=1e-10)

comparison = pd.DataFrame({
    "iterations": [len(pn) - 1, len(pg) - 1],
    "final point": [str(pn[-1].round(8)), str(pg[-1].round(8))],
    "final f": [f"{f_q(pn[-1]):.2e}", f"{f_q(pg[-1]):.2e}"],
}, index=["Newton's method", "Gradient descent (α = 0.09)"])
display(comparison)
```

# **Result**

* For $f(x)=e^x-2x$, Newton's method converged to $x^*=\ln2$ from $x_0=3$ in 7 iterations and the error showed quadratic convergence.
* For $g(x)=x^4-3x^3+2$ it reached the minimum from $x_0=3$ but converged slowly to the non-minimising stationary point $x=0$ from $x_0=0.5$.
* For the Rosenbrock function Newton's method reached $(1,1)$ in 6–7 iterations from $(-1.2,1)$ (including one wild intermediate jump to $(0.76,-3.18)$), whereas gradient descent had only reached about $(0.79,0.63)$ after 5000 iterations.
* On the quadratic $x^2+10y^2$ Newton's method needed a single step against 131 iterations for gradient descent with $\alpha=0.09$.

# **Discussion**

Newton's method uses curvature information ($f''$ or the Hessian) to choose both the direction and the length of the step, so it converges in very few iterations near a minimiser, independent of the conditioning that slows gradient descent. The price is the cost of forming and solving with the Hessian ($O(n^3)$ per iteration) and a lack of robustness: it only seeks $\nabla f=0$ and may converge to a saddle point or maximum, or diverge from a poor starting point. Practical algorithms therefore use a line search or trust region ("damped" Newton) or approximate the Hessian (quasi-Newton methods such as BFGS).

# **Solving Constrained Optimization Problems Using Lagrange Multipliers and Visualizing Solutions**

# **Aim:**

To solve constrained optimisation problems with the method of Lagrange multipliers (equality constraints) and the Kuhn–Tucker (KKT) conditions (inequality constraints), and to visualise the solutions.

# **Software required:**

Python 3, NumPy, SymPy, SciPy, Matplotlib, Pandas

# **Theory:**

**Equality constraints.** To optimise $f(\mathbf x)$ subject to $g(\mathbf x)=0$, form the **Lagrangian**

$$\mathcal L(\mathbf x,\lambda)=f(\mathbf x)-\lambda\,g(\mathbf x).$$

At a constrained optimum, $\nabla_{\mathbf x}\mathcal L=0$ and $g(\mathbf x)=0$, i.e.

$$\nabla f(\mathbf x^*)=\lambda\,\nabla g(\mathbf x^*):$$

the gradient of the objective is parallel to the gradient of the constraint — a level curve of $f$ touches the constraint curve without crossing it. The multiplier $\lambda$ measures how fast the optimal value changes when the constraint is relaxed ("shadow price").

**Inequality constraints and KKT conditions.** For $\min f(\mathbf x)$ subject to $g_i(\mathbf x)\ge0$, a local minimum $\mathbf x^*$ (under regularity conditions) satisfies, for some multipliers $\mu_i$:

1. *Stationarity:* $\nabla f(\mathbf x^*)=\sum_i\mu_i\nabla g_i(\mathbf x^*)$
2. *Primal feasibility:* $g_i(\mathbf x^*)\ge0$
3. *Dual feasibility:* $\mu_i\ge0$
4. *Complementary slackness:* $\mu_i\,g_i(\mathbf x^*)=0$ — a constraint is either active ($g_i=0$) or has zero multiplier.

For convex problems the KKT conditions are also sufficient.

# **Procedure:**

* Form the Lagrangian symbolically with SymPy and solve $\nabla\mathcal L=0$.
* Plot the level curves of $f$, the constraint and the optimum; check that the gradients are parallel.
* Solve a problem with inequality constraints with SciPy (SLSQP), recover the multipliers and verify the KKT conditions.
* Apply the method to a minimum-variance portfolio.

# **8.1 Lagrange multipliers with SymPy**

**Example 1.** Minimise $f(x,y)=x^2+y^2$ (squared distance from the origin) subject to $x+2y=5$.

$\mathcal L=x^2+y^2-\lambda(x+2y-5)\;\Rightarrow\;2x=\lambda,\;2y=2\lambda,\;x+2y=5\;\Rightarrow\;(x,y)=(1,2),\;\lambda=2,\;f=5.$

```python
import numpy as np
import pandas as pd
import sympy as sp
import matplotlib.pyplot as plt
from scipy.optimize import minimize

x, y, lam = sp.symbols('x y lam', real=True)

f = x ** 2 + y ** 2
g = x + 2 * y - 5
L = f - lam * g

equations = [sp.diff(L, v) for v in (x, y, lam)]
print("Lagrangian L =", L)
print("Equations (dL/dx, dL/dy, dL/dlam = 0):", equations)

solutions = sp.solve(equations, [x, y, lam], dict=True)
for s in solutions:
    print("Solution:", s, "   f =", f.subs(s))
```

**Example 2.** Maximise and minimise $f(x,y)=xy$ on the circle $x^2+y^2=8$.

$y=2\lambda x,\;x=2\lambda y\Rightarrow\lambda=\pm\tfrac12$, so $y=\pm x$ and there are four stationary points $(\pm2,\pm2)$. We evaluate $f$ at each to decide which is the maximum and which the minimum.

```python
f2 = x * y
g2 = x ** 2 + y ** 2 - 8
L2 = f2 - lam * g2
sols = sp.solve([sp.diff(L2, v) for v in (x, y, lam)], [x, y, lam], dict=True)

rows = [{"x": s[x], "y": s[y], "λ": s[lam], "f = xy": f2.subs(s)} for s in sols]
table = pd.DataFrame(rows)
display(table)
print("Maximum of xy on the circle:", table["f = xy"].max(), "   Minimum:", table["f = xy"].min())
```

**Example 3 (application).** A closed rectangular box has surface area $24\ \text{m}^2$. Find the dimensions of maximum volume: maximise $V=xyz$ subject to $2(xy+yz+zx)=24$.

```python
a, b, c, mu = sp.symbols('a b c mu', positive=True)
V = a * b * c
area = 2 * (a * b + b * c + c * a) - 24
Lbox = V - mu * area
sol = sp.solve([sp.diff(Lbox, v) for v in (a, b, c, mu)], [a, b, c, mu], dict=True)
print("Solution:", sol)
print("Maximum volume =", V.subs(sol[0]), "m^3  (the cube of side", sol[0][a], "m)")
```

# **8.3 Inequality constraints and the KKT conditions**

Minimise $f(x,y)=(x-1)^2+(y-2.5)^2$ (distance$^2$ to the point $(1,2.5)$) subject to

$$g_1=x-2y+2\ge0,\quad g_2=-x-2y+6\ge0,\quad g_3=-x+2y+2\ge0,\quad x\ge0,\;y\ge0 .$$

The unconstrained minimum $(1,2.5)$ violates $g_1$, so the solution lies on the boundary $g_1=0$. SciPy's SLSQP finds it; we then recover the multipliers from $\nabla f=\sum\mu_i\nabla g_i$ over the active constraints and check all four KKT conditions.

```python
fobj = lambda v: (v[0] - 1) ** 2 + (v[1] - 2.5) ** 2
grad_f = lambda v: np.array([2 * (v[0] - 1), 2 * (v[1] - 2.5)])

# constraints g_i(v) >= 0 and their (constant) gradients
G = [lambda v: v[0] - 2 * v[1] + 2, lambda v: -v[0] - 2 * v[1] + 6, lambda v: -v[0] + 2 * v[1] + 2]
dG = [np.array([1., -2.]), np.array([-1., -2.]), np.array([-1., 2.])]
cons = [{'type': 'ineq', 'fun': g} for g in G]

res = minimize(fobj, x0=[2.0, 0.0], method='SLSQP', bounds=[(0, None), (0, None)], constraints=cons)
xs = res.x
print(f"Optimum  x* = ({xs[0]:.4f}, {xs[1]:.4f}),   f(x*) = {res.fun:.4f},   success = {res.success}")
print("Unconstrained minimum (1, 2.5) is feasible?  g1 =", G[0]([1, 2.5]), "< 0  -> no")

# multipliers: solve  grad f = sum mu_i grad g_i  over the active constraints
active = [i for i in range(3) if abs(G[i](xs)) < 1e-5]
mu = np.zeros(3)
if active:
    Mat = np.column_stack([dG[i] for i in active])
    mu[active] = np.linalg.lstsq(Mat, grad_f(xs), rcond=None)[0]

kkt = pd.DataFrame({
    "g_i(x*)": [G[i](xs) for i in range(3)],
    "active?": ["yes" if i in active else "no" for i in range(3)],
    "multiplier μ_i": mu,
    "μ_i · g_i(x*)": [mu[i] * G[i](xs) for i in range(3)],
}, index=["g1 = x - 2y + 2", "g2 = -x - 2y + 6", "g3 = -x + 2y + 2"]).round(6)
display(kkt)

stationarity = np.linalg.norm(grad_f(xs) - sum(mu[i] * dG[i] for i in range(3)))
print(f"\n1. Stationarity residual ||∇f - Σ μ_i ∇g_i|| = {stationarity:.2e}")
print("2. Primal feasibility  (all g_i ≥ 0):", all(G[i](xs) > -1e-6 for i in range(3)))
print("3. Dual feasibility    (all μ_i ≥ 0):", all(mu >= -1e-8))
print("4. Complementary slackness (μ_i g_i = 0):", np.allclose([mu[i] * G[i](xs) for i in range(3)], 0, atol=1e-6))
```

```python
X, Y = np.meshgrid(np.linspace(0, 4, 400), np.linspace(0, 4, 400))
feasible = (X - 2 * Y + 2 >= 0) & (-X - 2 * Y + 6 >= 0) & (-X + 2 * Y + 2 >= 0)

fig, ax = plt.subplots(figsize=(7.5, 6.5))
ax.contourf(X, Y, feasible, levels=[0.5, 1.5], colors=['lightgreen'], alpha=0.6)
cs = ax.contour(X, Y, (X - 1) ** 2 + (Y - 2.5) ** 2, levels=[0.25, 0.5, 1, 2, 4], cmap='viridis')
ax.clabel(cs, fmt='%g', fontsize=8)
xl = np.linspace(0, 4, 100)
ax.plot(xl, (xl + 2) / 2, 'r-', label='g1 = 0 (active)')
ax.plot(xl, (6 - xl) / 2, 'b--', label='g2 = 0')
ax.plot(xl, (xl - 2) / 2, 'm--', label='g3 = 0')
ax.plot(1, 2.5, 'k^', ms=10, label='unconstrained minimum (infeasible)')
ax.plot(*xs, 'k*', ms=16, label=f'constrained optimum ({xs[0]:.1f}, {xs[1]:.1f})')
ax.set_xlim(0, 4); ax.set_ylim(0, 4); ax.set_aspect('equal'); ax.grid(True); ax.legend(fontsize=8, loc='lower right')
ax.set_title('Feasible region (green), level curves of f and the KKT point')
plt.show()
```

# **8.4 Application – minimum-variance portfolio**

An investor splits one unit of money among three assets with covariance matrix $\Sigma$ and wants the least risk (variance $\mathbf w^T\Sigma\mathbf w$) subject to the budget constraint $\mathbf 1^T\mathbf w=1$. The Lagrangian conditions give $2\Sigma\mathbf w=\lambda\mathbf 1$, so

$$\mathbf w^*=\frac{\Sigma^{-1}\mathbf 1}{\mathbf 1^T\Sigma^{-1}\mathbf 1},\qquad \lambda=2\,\mathbf w^{*T}\Sigma\mathbf w^* .$$

(Negative weights would mean short-selling.) The closed-form answer is checked against a numerical constrained optimiser.

```python
Sigma = np.array([[0.10, 0.02, 0.04],
                  [0.02, 0.08, 0.01],
                  [0.04, 0.01, 0.09]])
ones = np.ones(3)

w = np.linalg.solve(Sigma, ones)
w = w / (ones @ w)
risk = w @ Sigma @ w
print("Lagrange solution  w* =", w.round(4), "  sum =", w.sum().round(6))
print(f"Minimum variance = {risk:.5f}   (std. deviation {np.sqrt(risk):.4f}),   multiplier λ = 2·variance = {2 * risk:.5f}")

num = minimize(lambda v: v @ Sigma @ v, x0=ones / 3, method='SLSQP',
               constraints=[{'type': 'eq', 'fun': lambda v: v.sum() - 1}])
print("SciPy (SLSQP)      w  =", num.x.round(4), "  variance =", round(num.fun, 5))
print("Same answer (to 3 decimals)?", np.allclose(w, num.x, atol=1e-3))

# compare with the equal-weight portfolio
eq = ones / 3
print(f"Equal-weight portfolio variance = {eq @ Sigma @ eq:.5f}  (higher risk)")

plt.figure(figsize=(6, 4))
plt.bar(['Asset 1', 'Asset 2', 'Asset 3'], w, color=['tab:blue', 'tab:orange', 'tab:green'])
plt.axhline(0, color='black', lw=0.8); plt.ylabel('weight'); plt.title('Minimum-variance portfolio weights')
plt.show()
```

# **Result**

* Example 1: minimum of $x^2+y^2$ on $x+2y=5$ at $(1,2)$ with $\lambda=2$ and $f=5$.
* Example 2: the four stationary points $(\pm2,\pm2)$; maximum $xy=4$ at $(2,2),(-2,-2)$ and minimum $xy=-4$ at $(2,-2),(-2,2)$.
* Example 3: the box of maximum volume is the cube of side $2$ m with $V=8\ \text{m}^3$.
* The inequality-constrained problem had its optimum at $(1.4,\,1.7)$ with $f=0.8$, only $g_1$ active and multiplier $\mu_1=0.8$; all four KKT conditions were verified.
* The minimum-variance portfolio from the Lagrange formula agreed with the numerical optimiser.

# **Discussion**

At a constrained optimum the gradient of the objective is a multiple of the gradient of the (active) constraint — in the pictures, a level curve touches the constraint tangentially. The multiplier has a meaning: $\mu_1=0.8$ says that relaxing $g_1\ge0$ by a small amount $\varepsilon$ would lower the minimum value by about $0.8\varepsilon$. Complementary slackness explains why the inactive constraints $g_2,g_3$ and the bounds carry zero multipliers. Lagrange multipliers turn a constrained problem into the *unconstrained* problem of finding stationary points of $\mathcal L$, which is also the starting point of the theory of duality used in machine learning (e.g. support vector machines).

# **Simulating Random Variables and Probability Distributions and Visualising Their Outcomes**

# **Aim:**

To simulate random variables from the binomial, Poisson, normal and exponential distributions, to compare the simulated outcomes with the theoretical distributions, and to illustrate the law of large numbers, the central limit theorem and independence by simulation.

# **Software required:**

Python 3, NumPy, SciPy (`scipy.stats`), Pandas, Matplotlib

# **Theory:**

| Distribution | Type | pmf / pdf | Mean | Variance |
|---|---|---|---|---|
| Binomial$(n,p)$ | discrete | $\binom nk p^k(1-p)^{n-k}$ | $np$ | $np(1-p)$ |
| Poisson$(\lambda)$ | discrete | $e^{-\lambda}\lambda^k/k!$ | $\lambda$ | $\lambda$ |
| Normal$(\mu,\sigma^2)$ | continuous | $\frac{1}{\sigma\sqrt{2\pi}}e^{-(x-\mu)^2/2\sigma^2}$ | $\mu$ | $\sigma^2$ |
| Exponential (rate $\theta$) | continuous | $\theta e^{-\theta x},\;x\ge0$ | $1/\theta$ | $1/\theta^2$ |

* **Law of large numbers:** the sample mean $\bar X_n$ approaches $E[X]$ as $n\to\infty$.
* **Central limit theorem:** for large $n$ the mean of $n$ independent observations is approximately normal, $\bar X_n\approx N(\mu,\sigma^2/n)$, whatever the distribution of $X$ (finite variance).
* **Independence:** $X,Y$ are independent if $P(X\in A,Y\in B)=P(X\in A)P(Y\in B)$; then their covariance is $0$.
* **Inverse-transform sampling:** if $U\sim\text{Uniform}(0,1)$ and $F$ is a cdf, then $F^{-1}(U)$ has cdf $F$. For the exponential distribution $X=-\ln(U)/\theta$.

# **Procedure:**

* Generate samples with a seeded random number generator and compare sample statistics with theory.
* Draw histograms/bar charts against the exact pmf/pdf and the empirical cdf.
* Demonstrate the law of large numbers and the central limit theorem.
* Simulate independent and correlated pairs; use inverse-transform sampling and a Monte Carlo estimate of $\pi$.

# **9.1 Generating samples and comparing with theory**

`np.random.default_rng(seed)` creates a random number generator; a fixed *seed* makes the results reproducible. `scipy.stats` provides the exact mean, variance, pmf/pdf and cdf of each distribution.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(2026)
N = 10_000

dists = {
    "Binomial(n=10, p=0.3)":  (stats.binom(10, 0.3), rng.binomial(10, 0.3, N)),
    "Poisson(λ=4)":           (stats.poisson(4),     rng.poisson(4, N)),
    "Normal(μ=50, σ=10)":     (stats.norm(50, 10),   rng.normal(50, 10, N)),
    "Exponential(mean=2)":    (stats.expon(scale=2), rng.exponential(2, N)),
}

rows = [{"Distribution": name,
         "Theoretical mean": d.mean(), "Sample mean": s.mean(),
         "Theoretical variance": d.var(), "Sample variance": s.var(ddof=1)}
        for name, (d, s) in dists.items()]
display(pd.DataFrame(rows).round(4).set_index("Distribution"))
```

# **9.2 Visualising the outcomes**

Discrete variables: bars show the **relative frequency** of each value and dots the exact pmf. Continuous variables: the histogram is scaled to unit area (density) and the curve is the exact pdf.

```python
fig, ax = plt.subplots(2, 2, figsize=(13, 9))

for axis, name in zip(ax[0], ["Binomial(n=10, p=0.3)", "Poisson(λ=4)"]):
    d, s = dists[name]
    vals, counts = np.unique(s, return_counts=True)
    axis.bar(vals, counts / N, alpha=0.6, label='simulated relative frequency')
    axis.plot(vals, d.pmf(vals), 'ro-', label='exact pmf')
    axis.set_title(name); axis.set_xlabel('k'); axis.legend(); axis.grid(True)

for axis, name in zip(ax[1], ["Normal(μ=50, σ=10)", "Exponential(mean=2)"]):
    d, s = dists[name]
    grid = np.linspace(s.min(), s.max(), 400)
    axis.hist(s, bins=50, density=True, alpha=0.6, label='simulated (density)')
    axis.plot(grid, d.pdf(grid), 'r-', lw=2, label='exact pdf')
    axis.set_title(name); axis.set_xlabel('x'); axis.legend(); axis.grid(True)
plt.tight_layout()
plt.show()
```

**Empirical cdf.** For sorted data $x_{(1)}\le\dots\le x_{(n)}$ the empirical cdf rises by $1/n$ at each observation and approximates the true cdf $F$.

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
for axis, name in zip(ax, ["Normal(μ=50, σ=10)", "Exponential(mean=2)"]):
    d, s = dists[name]
    xs_sorted = np.sort(s)
    axis.step(xs_sorted, np.arange(1, N + 1) / N, where='post', label='empirical cdf')
    axis.plot(xs_sorted, d.cdf(xs_sorted), 'r--', label='exact cdf')
    axis.set_title(name); axis.legend(); axis.grid(True)
plt.tight_layout()
plt.show()

# Probability of an event: simulated frequency vs exact value
d, s = dists["Normal(μ=50, σ=10)"]
print("P(40 < X < 60) for Normal(50, 10):  simulated =", np.mean((s > 40) & (s < 60)).round(4), "  exact =", round(d.cdf(60) - d.cdf(40), 4))
d, s = dists["Binomial(n=10, p=0.3)"]
print("P(X >= 5)      for Binomial(10, 0.3): simulated =", np.mean(s >= 5).round(4), "  exact =", round(d.sf(4), 4))
```

# **9.4 Central limit theorem**

# **9.5 Joint distributions and independence**

Compare two *independent* standard normals with a *correlated* pair (correlation $\rho=0.8$). For the independent pair $P(X>0,Y>0)=P(X>0)P(Y>0)=0.25$; for the correlated pair the exact value is $\tfrac14+\tfrac{\arcsin\rho}{2\pi}\approx0.3976$.

```python
n = 5000
indep = rng.normal(size=(n, 2))
corr = rng.multivariate_normal([0, 0], [[1, 0.8], [0.8, 1]], size=n)

for name, data in (("Independent", indep), ("Correlated (ρ = 0.8)", corr)):
    X, Y = data[:, 0], data[:, 1]
    joint = np.mean((X > 0) & (Y > 0))
    print(f"{name:22s} sample correlation = {np.corrcoef(X, Y)[0, 1]:+.3f},  cov = {np.cov(X, Y)[0, 1]:+.3f},  "
          f"P(X>0,Y>0) = {joint:.3f}  vs  P(X>0)P(Y>0) = {np.mean(X > 0) * np.mean(Y > 0):.3f}")
print("Exact P(X>0, Y>0) for rho = 0.8:", round(0.25 + np.arcsin(0.8) / (2 * np.pi), 4))

fig, ax = plt.subplots(1, 2, figsize=(12, 5.5), sharex=True, sharey=True)
for axis, data, title in zip(ax, (indep, corr), ("Independent", "Correlated, ρ = 0.8")):
    axis.scatter(data[:, 0], data[:, 1], s=6, alpha=0.4)
    axis.axhline(0, color='black', lw=0.8); axis.axvline(0, color='black', lw=0.8)
    axis.set_title(title); axis.set_xlabel('X'); axis.set_ylabel('Y'); axis.set_aspect('equal'); axis.grid(True)
plt.tight_layout()
plt.show()
```

# **9.6 Inverse-transform sampling and Monte Carlo estimation**

```python
# ---- inverse-transform sampling
U = rng.random(N)
x_inv = -np.log(U) / 0.5
print(f"Inverse-transform sample: mean = {x_inv.mean():.4f} (theory 2),  variance = {x_inv.var(ddof=1):.4f} (theory 4)")
print("Kolmogorov-Smirnov test against Exponential(mean 2): p-value =", round(stats.kstest(x_inv, stats.expon(scale=2).cdf).pvalue, 4))

# ---- Monte Carlo estimate of pi
M = 20000
pts = rng.uniform(-1, 1, size=(M, 2))
inside = (pts ** 2).sum(axis=1) <= 1
print(f"\nMonte Carlo estimate of pi with {M} points: {4 * inside.mean():.4f}   (true value {np.pi:.4f})")

fig, ax = plt.subplots(1, 2, figsize=(13, 5.5))
grid = np.linspace(0, x_inv.max(), 300)
ax[0].hist(x_inv, bins=50, density=True, alpha=0.6, label='inverse-transform samples')
ax[0].plot(grid, stats.expon(scale=2).pdf(grid), 'r-', lw=2, label='exact pdf'); ax[0].legend(); ax[0].grid(True)
ax[0].set_title('Inverse-transform sampling of Exponential(mean 2)')

ax[1].scatter(pts[:2000, 0], pts[:2000, 1], c=inside[:2000], cmap='coolwarm', s=5)
t = np.linspace(0, 2 * np.pi, 300); ax[1].plot(np.cos(t), np.sin(t), 'k')
ax[1].set_aspect('equal'); ax[1].set_title(f'Monte Carlo estimation of π (first 2000 of {M} points)')
plt.tight_layout()
plt.show()
```

# **Performing Hypothesis Testing and Regression Analysis Using Python and Visualizing the Results**

# **Aim:**

To perform hypothesis tests (one-sample $z$ and $t$ tests, two-sample $t$ test), to construct confidence intervals, and to carry out simple and multiple linear regression analysis in Python, visualising the results.

# **Software required:**

Python 3, NumPy, SciPy (`scipy.stats`), Pandas, Matplotlib

# **Theory:**

**Hypothesis testing.** State a null hypothesis $H_0$ and an alternative $H_1$, compute a test statistic from the sample, and find the **$p$-value** — the probability, assuming $H_0$ is true, of a result at least as extreme as the one observed. Reject $H_0$ at significance level $\alpha$ (usually $0.05$) if $p<\alpha$.

* *One-sample $t$-test* ($H_0:\mu=\mu_0$, $\sigma$ unknown): $\;t=\dfrac{\bar x-\mu_0}{s/\sqrt n}\sim t_{n-1}$
* *One-sample $z$-test* ($\sigma$ known): $\;z=\dfrac{\bar x-\mu_0}{\sigma/\sqrt n}\sim N(0,1)$
* *Two-sample (Welch) $t$-test* ($H_0:\mu_1=\mu_2$): $\;t=\dfrac{\bar x_1-\bar x_2}{\sqrt{s_1^2/n_1+s_2^2/n_2}}$
* **Type I error** = rejecting a true $H_0$ (probability $\alpha$); **power** = probability of rejecting a false $H_0$.
* **Confidence interval** for the mean: $\bar x\pm t_{1-\alpha/2,\,n-1}\,\dfrac{s}{\sqrt n}$.

**Linear regression.** Model $y=\beta_0+\beta_1x+\varepsilon$. Least squares gives $\hat\beta_1=\dfrac{S_{xy}}{S_{xx}},\;\hat\beta_0=\bar y-\hat\beta_1\bar x$. In matrix form (several predictors) $\hat{\boldsymbol\beta}=(X^TX)^{-1}X^T\mathbf y$. The coefficient of determination is $R^2=1-\dfrac{SSE}{SST}$, and $s^2=\dfrac{SSE}{n-p}$ estimates the error variance ($p$ = number of coefficients), giving standard errors $\operatorname{SE}(\hat\beta_j)=s\sqrt{[(X^TX)^{-1}]_{jj}}$ and $t$-tests of $H_0:\beta_j=0$.

# **Procedure:**

* Test a claim about a mean with $t$ and $z$ tests, with a confidence interval and a picture of the rejection region.
* Compare two groups with a Welch $t$-test.
* Estimate the Type I error and power by simulation.
* Fit simple and multiple linear regression models; analyse residuals and visualise the fit.

# **Part A – Hypothesis testing**

# **10.1 One-sample $t$-test**

A manufacturer claims that its batteries last **50 hours** on average. A sample of 15 batteries gave the lifetimes below.

* $H_0:\ \mu=50$   against   $H_1:\ \mu\ne50$ (two-sided), $\alpha=0.05$.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

life = np.array([48.2, 51.5, 47.9, 49.3, 52.1, 46.8, 50.4, 48.7, 49.9, 47.5, 51.0, 48.3, 49.1, 50.2, 47.6])
mu0, alpha = 50, 0.05
n = len(life); xbar = life.mean(); s = life.std(ddof=1)

# --- by the formula
t_stat = (xbar - mu0) / (s / np.sqrt(n))
df_ = n - 1
p_val = 2 * stats.t.sf(abs(t_stat), df_)
t_crit = stats.t.ppf(1 - alpha / 2, df_)
print(f"n = {n},  sample mean = {xbar:.4f},  sample std = {s:.4f}")
print(f"t = {t_stat:.4f},  degrees of freedom = {df_},  critical values = ±{t_crit:.4f},  p-value = {p_val:.4f}")

# --- with SciPy
res = stats.ttest_1samp(life, mu0)
print(f"scipy.stats.ttest_1samp: t = {res.statistic:.4f},  p-value = {res.pvalue:.4f}")
print("Decision at α = 0.05:", "reject H0" if res.pvalue < alpha else "fail to reject H0 (no evidence that the mean differs from 50)")

# --- 95% confidence interval for the mean
half = t_crit * s / np.sqrt(n)
print(f"95% confidence interval for μ: ({xbar - half:.3f}, {xbar + half:.3f})")
lo, hi = stats.t.interval(0.95, df_, loc=xbar, scale=stats.sem(life))
print(f"scipy interval              : ({lo:.3f}, {hi:.3f})")
```

```python
grid = np.linspace(-4.5, 4.5, 500)
pdf = stats.t.pdf(grid, df_)

plt.figure(figsize=(9, 4.5))
plt.plot(grid, pdf, 'k')
plt.fill_between(grid, pdf, where=grid <= -t_crit, color='red', alpha=0.4, label=f'rejection region (α = {alpha})')
plt.fill_between(grid, pdf, where=grid >= t_crit,  color='red', alpha=0.4)
plt.axvline(t_stat, color='blue', lw=2, label=f'observed t = {t_stat:.2f}')
plt.title(f'Two-sided t-test, {df_} degrees of freedom'); plt.xlabel('t'); plt.legend(); plt.grid(True)
plt.show()
```

# **10.2 One-sample $z$-test (known $\sigma$)**

Suppose the standard deviation of battery life is *known* to be $\sigma=1.8$ hours from long production history. Then the statistic uses the normal distribution. With known $\sigma$ the test is slightly more powerful than the $t$-test.

```python
sigma = 1.8
z_stat = (xbar - mu0) / (sigma / np.sqrt(n))
p_z = 2 * stats.norm.sf(abs(z_stat))
z_crit = stats.norm.ppf(1 - alpha / 2)
print(f"z = {z_stat:.4f},  critical values = ±{z_crit:.4f},  p-value = {p_z:.4f}")
print("Decision at α = 0.05:", "reject H0" if p_z < alpha else "fail to reject H0")
print(f"95% CI for μ (σ known): ({xbar - z_crit * sigma / np.sqrt(n):.3f}, {xbar + z_crit * sigma / np.sqrt(n):.3f})")
```

# **10.3 Two-sample (Welch) $t$-test**

Two teaching methods are compared using the test scores of two independent groups of 12 students. $H_0:\ \mu_A=\mu_B$ against $H_1:\ \mu_A\ne\mu_B$.

```python
group_A = np.array([72, 75, 68, 80, 77, 74, 70, 79, 73, 76, 71, 78])
group_B = np.array([78, 82, 75, 85, 80, 84, 77, 86, 81, 79, 83, 88])

res2 = stats.ttest_ind(group_A, group_B, equal_var=False)       # Welch's t-test
diff = group_A.mean() - group_B.mean()
print(f"mean A = {group_A.mean():.2f},  mean B = {group_B.mean():.2f},  difference = {diff:.2f}")
print(f"Welch t = {res2.statistic:.4f},  p-value = {res2.pvalue:.5f}")
print("Decision at α = 0.05:", "reject H0 -> the methods differ" if res2.pvalue < alpha else "fail to reject H0")

pooled_sd = np.sqrt((group_A.var(ddof=1) + group_B.var(ddof=1)) / 2)
print(f"Effect size (Cohen's d) = {abs(diff) / pooled_sd:.2f}")

plt.figure(figsize=(7, 5))
plt.boxplot([group_A, group_B], patch_artist=True)
plt.xticks([1, 2], ['Method A', 'Method B'])
plt.scatter(np.ones(12) + np.random.default_rng(0).normal(0, 0.04, 12), group_A, color='tab:blue', zorder=3)
plt.scatter(2 * np.ones(12) + np.random.default_rng(1).normal(0, 0.04, 12), group_B, color='tab:orange', zorder=3)
plt.ylabel('score'); plt.title('Comparison of two groups'); plt.grid(True)
plt.show()
```

# **10.4 Type I error and power by simulation**

If $H_0$ is true, a test at $\alpha=0.05$ should wrongly reject in about 5% of repeated experiments (Type I error). If the true mean differs, the rejection rate is the **power**. Simulate 10 000 experiments with samples of size 15 from a normal population with $\sigma=2$.

```python
rng = np.random.default_rng(10)
reps, n_s, sd = 10_000, 15, 2.0

def rejection_rate(true_mean):
    samples = rng.normal(true_mean, sd, size=(reps, n_s))
    pvals = stats.ttest_1samp(samples, 50, axis=1).pvalue
    return np.mean(pvals < 0.05)

print(f"True mean = 50 (H0 true) : rejection rate = {rejection_rate(50):.4f}   (Type I error ≈ α = 0.05)")
means = np.array([50, 50.5, 51, 51.5, 52, 53])
power = [rejection_rate(m) for m in means]
for m, p in zip(means, power):
    print(f"True mean = {m:4.1f}        : power = {p:.4f}")

plt.figure(figsize=(7, 4.5))
plt.plot(means, power, 'o-'); plt.axhline(0.05, color='gray', ls=':')
plt.xlabel('true mean'); plt.ylabel('probability of rejecting H0'); plt.title('Power curve of the one-sample t-test (n = 15, σ = 2)')
plt.grid(True)
plt.show()
```

# **Part B – Regression analysis**

# **10.5 Simple linear regression**

Fifty observations are generated from $y=3+2x+\varepsilon$ with $\varepsilon\sim N(0,2^2)$ so that the estimates can be compared with the true values. The slope and intercept are computed (i) from the formulas, (ii) with `np.polyfit`, and (iii) with `scipy.stats.linregress`.

```python
rng = np.random.default_rng(42)
n = 50
x = rng.uniform(0, 10, n)
y = 3 + 2 * x + rng.normal(0, 2, n)

# (i) formulas
xbar, ybar = x.mean(), y.mean()
Sxx = np.sum((x - xbar) ** 2); Sxy = np.sum((x - xbar) * (y - ybar))
b1 = Sxy / Sxx; b0 = ybar - b1 * xbar
yhat = b0 + b1 * x
resid = y - yhat
SSE = np.sum(resid ** 2); SST = np.sum((y - ybar) ** 2)
R2 = 1 - SSE / SST
s2 = SSE / (n - 2)
se_b1 = np.sqrt(s2 / Sxx); se_b0 = np.sqrt(s2 * (1 / n + xbar ** 2 / Sxx))
t_crit = stats.t.ppf(0.975, n - 2)

print(f"Fitted line: y = {b0:.4f} + {b1:.4f} x      (true: y = 3 + 2x)")
print(f"R² = {R2:.4f},   residual standard error s = {np.sqrt(s2):.4f}")
print(f"slope     : estimate {b1:.4f},  SE {se_b1:.4f},  95% CI ({b1 - t_crit * se_b1:.3f}, {b1 + t_crit * se_b1:.3f})")
print(f"intercept : estimate {b0:.4f},  SE {se_b0:.4f},  95% CI ({b0 - t_crit * se_b0:.3f}, {b0 + t_crit * se_b0:.3f})")

# (ii) np.polyfit, (iii) scipy.stats.linregress
print("\nnp.polyfit        :", np.polyfit(x, y, 1).round(4), " (slope, intercept)")
lr = stats.linregress(x, y)
print(f"stats.linregress  : slope = {lr.slope:.4f}, intercept = {lr.intercept:.4f}, r = {lr.rvalue:.4f}, "
      f"R² = {lr.rvalue ** 2:.4f}, p-value(slope) = {lr.pvalue:.2e}, SE(slope) = {lr.stderr:.4f}")
```

**Visualising the fit.** The shaded band is the 95% confidence band for the *mean* response, $\hat y\pm t\,s\sqrt{\tfrac1n+\tfrac{(x-\bar x)^2}{S_{xx}}}$. The residual plots check the model assumptions: residuals should scatter randomly around 0 with constant spread, and be roughly normal.

```python
xg = np.linspace(0, 10, 200)
yg = b0 + b1 * xg
band = t_crit * np.sqrt(s2) * np.sqrt(1 / n + (xg - xbar) ** 2 / Sxx)

fig, ax = plt.subplots(1, 3, figsize=(17, 4.8))
ax[0].scatter(x, y, s=18, label='data')
ax[0].plot(xg, yg, 'r-', lw=2, label=f'fit: y = {b0:.2f} + {b1:.2f}x')
ax[0].plot(xg, 3 + 2 * xg, 'g--', label='true line')
ax[0].fill_between(xg, yg - band, yg + band, color='red', alpha=0.2, label='95% confidence band')
ax[0].set_title(f'Simple linear regression (R² = {R2:.3f})'); ax[0].legend(); ax[0].grid(True)

ax[1].scatter(yhat, resid, s=18); ax[1].axhline(0, color='red')
ax[1].set_xlabel('fitted value'); ax[1].set_ylabel('residual'); ax[1].set_title('Residuals versus fitted'); ax[1].grid(True)

stats.probplot(resid, dist='norm', plot=ax[2])
ax[2].set_title('Normal Q–Q plot of residuals'); ax[2].grid(True)
plt.tight_layout()
plt.show()
print("Shapiro-Wilk normality test of residuals: p-value =", round(stats.shapiro(resid).pvalue, 4), "(p > 0.05: no evidence against normality)")
```

# **10.6 Multiple linear regression**

Now $y=5+1.5x_1-2x_2+\varepsilon$ with $\varepsilon\sim N(0,1.5^2)$ and $x_1,x_2$ possibly correlated. In matrix form $\hat{\boldsymbol\beta}=(X^TX)^{-1}X^T\mathbf y$ with a column of ones for the intercept. We compute the coefficients, their standard errors, $t$-statistics and $p$-values, $R^2$, adjusted $R^2$ and the overall $F$-test.

```python
rng = np.random.default_rng(7)
n = 80
x1 = rng.uniform(0, 10, n)
x2 = 0.5 * x1 + rng.uniform(0, 5, n)                 # correlated with x1
y2 = 5 + 1.5 * x1 - 2 * x2 + rng.normal(0, 1.5, n)

X = np.column_stack([np.ones(n), x1, x2])
p = X.shape[1]
beta = np.linalg.lstsq(X, y2, rcond=None)[0]
fitted = X @ beta
res_ = y2 - fitted
SSE = res_ @ res_; SST = np.sum((y2 - y2.mean()) ** 2)
s2 = SSE / (n - p)
cov_beta = s2 * np.linalg.inv(X.T @ X)
se = np.sqrt(np.diag(cov_beta))
t_vals = beta / se
p_vals = 2 * stats.t.sf(np.abs(t_vals), n - p)
R2 = 1 - SSE / SST
R2_adj = 1 - (1 - R2) * (n - 1) / (n - p)
F = ((SST - SSE) / (p - 1)) / (SSE / (n - p))
p_F = stats.f.sf(F, p - 1, n - p)

display(pd.DataFrame({"true value": [5, 1.5, -2], "estimate": beta, "std. error": se, "t": t_vals, "p-value": p_vals},
                     index=["intercept", "x1", "x2"]).round(4))
print(f"R² = {R2:.4f},  adjusted R² = {R2_adj:.4f},  residual standard error = {np.sqrt(s2):.4f}")
print(f"Overall F-test: F = {F:.2f},  p-value = {p_F:.2e}")
print("Correlation between x1 and x2:", np.corrcoef(x1, x2)[0, 1].round(3))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5.5))
ax[0].scatter(fitted, y2, s=18)
lims = [min(y2.min(), fitted.min()), max(y2.max(), fitted.max())]
ax[0].plot(lims, lims, 'r--', label='perfect prediction')
ax[0].set_xlabel('predicted y'); ax[0].set_ylabel('observed y'); ax[0].set_title(f'Observed vs predicted (R² = {R2:.3f})')
ax[0].legend(); ax[0].grid(True)

ax[1].scatter(fitted, res_, s=18); ax[1].axhline(0, color='red')
ax[1].set_xlabel('predicted y'); ax[1].set_ylabel('residual'); ax[1].set_title('Residual plot'); ax[1].grid(True)
plt.tight_layout()
plt.show()
```

# **Result**

* **One-sample tests:** the 15 batteries had mean 49.23 h. The $t$-test gave $t=-1.88$, $p=0.081$ and the $z$-test (known $\sigma=1.8$) gave $z=-1.65$, $p=0.099$; both fail to reject $H_0:\mu=50$ at $\alpha=0.05$, and the 95% confidence interval $(48.36,\,50.11)$ contains 50.
* **Welch test:** the two teaching methods differed significantly (group means 74.4 and 81.5, $t=-4.54$, $p=0.00016$, Cohen's $d\approx1.85$).
* **Simulation:** the Type I error rate was about 4.6% (nominal 5%) and the power rose from 14% (true mean 50.5) to 77% (51.5) and 99.97% (53).
* **Regression:** the simple regression gave $y=2.54+2.02x$ (true: $3+2x$) with $R^2=0.932$ and a 95% interval $(1.87,\,2.18)$ for the slope; the formulas, `polyfit` and `linregress` agreed and the residuals showed no pattern (Shapiro–Wilk $p=0.23$). The multiple regression estimated $\boldsymbol\beta=(5.28,\,1.48,\,-2.06)$ against the true $(5,\,1.5,\,-2)$, with highly significant coefficients, $R^2=0.850$ and $F=217$.

# **Discussion**

A $p$-value above $\alpha$ does **not** prove $H_0$; it only says the data are consistent with it — the power curve shows that small departures from $H_0$ are often missed with only 15 observations. A confidence interval conveys the same information as the test but also shows the plausible size of the effect. In regression, $R^2$ alone does not validate a model: the residual plot (random scatter) and Q–Q plot (straight line) check linearity, constant variance and normality. When predictors are correlated (as $x_1,x_2$ here) the individual coefficients have larger standard errors, although the overall fit may still be excellent.
